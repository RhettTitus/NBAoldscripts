```python
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Example: Replace with your actual data (from the game) and predicted data (from your model)
predicted_data = {
    'Player': ['Donte DiVincenzo', 'Jaden McDaniels', 'Rob Dillingham', 'Anthony Edwards'],
    'Adjusted PPG': [8.5, 9.22, 3.63, 26.31],
    'Adjusted RPG': [2.8, 3.37, 0.78, 5.1],
    'Adjusted APG': [3.15, 1.76, 1.38, 3.55]
}

# Actual performance data (to be replaced with actual stats)
actual_data = {
    'Player': ['Donte DiVincenzo', 'Jaden McDaniels', 'Rob Dillingham', 'Anthony Edwards'],
    'Actual PPG': [11, 3, 3, 8],
    'Actual RPG': [5, 2, 2, 7],
    'Actual APG': [9, 0, 0, 3]
}

# Convert to DataFrame
predicted_df = pd.DataFrame(predicted_data)
actual_df = pd.DataFrame(actual_data)

# Merge the predicted and actual data on Player name
df = pd.merge(predicted_df, actual_df, on='Player')

# Calculate MAE and RMSE for each stat (PPG, RPG, APG)
mae_ppg = mean_absolute_error(df['Actual PPG'], df['Adjusted PPG'])
mae_rpg = mean_absolute_error(df['Actual RPG'], df['Adjusted RPG'])
mae_apg = mean_absolute_error(df['Actual APG'], df['Adjusted APG'])

rmse_ppg = np.sqrt(mean_squared_error(df['Actual PPG'], df['Adjusted PPG']))
rmse_rpg = np.sqrt(mean_squared_error(df['Actual RPG'], df['Adjusted RPG']))
rmse_apg = np.sqrt(mean_squared_error(df['Actual APG'], df['Adjusted APG']))

print(f"MAE for PPG: {mae_ppg}")
print(f"MAE for RPG: {mae_rpg}")
print(f"MAE for APG: {mae_apg}")
print(f"RMSE for PPG: {rmse_ppg}")
print(f"RMSE for RPG: {rmse_rpg}")
print(f"RMSE for APG: {rmse_apg}")
```
