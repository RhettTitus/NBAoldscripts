```python
!pip install nba_api
from nba_api.stats.endpoints import commonteamroster
import pandas as pd

def fetch_team_roster_positions(team_id, season):
    try:
        # Fetch the roster data
        roster_data = commonteamroster.CommonTeamRoster(
            team_id=team_id,
            season=season
        )

        # Extract the roster DataFrame
        roster_df = roster_data.get_data_frames()[0]

        # Filter relevant columns
        position_data = roster_df[["PLAYER", "POSITION", "HEIGHT", "WEIGHT", "AGE"]]

        # Display the result
        print(position_data)
        return position_data

    except Exception as e:
        print(f"Error fetching roster for team {team_id}: {e}")
        return None

# Test the function with a specific team ID and season
team_id = 1610612737  # Atlanta Hawks (example team)
season = "2024-25"    # Current NBA season

positions_df = fetch_team_roster_positions(team_id, season)
```
