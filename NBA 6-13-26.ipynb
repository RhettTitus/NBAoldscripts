```python
#@title Sportsdataverse historic regular season an playoff boxscores. This must be updated each year for the new season
!pip install sportsdataverse polars pyarrow fastparquet
from google.colab import drive
drive.mount('/content/drive')

import sportsdataverse as sdv
import pandas as pd
import time
import os
import glob
from tqdm import tqdm

# ============================================
# CONFIG
# ============================================
boxscore_folder = "/content/drive/MyDrive/NBA/Historical Data/Boxscores/"
os.makedirs(boxscore_folder, exist_ok=True)

SEASONS = list(range(2010, 2026))  # 2010–2025 inclusive (ends at 2025 for 26-27 preseason)

# ============================================
# SPORTSDATAVERSE ADAPTER
# ============================================
def prepare_sdv_df(raw_df, season, is_regular):
    raw_cols = {c.lower(): c for c in raw_df.columns}

    def col(candidates, default=None):
        for c in candidates:
            if c.lower() in raw_cols:
                return raw_df[raw_cols[c.lower()]]
        return default

    df = pd.DataFrame()
    df["season_year"] = season
    df["game_date"] = col(["game_date", "gameDate"]).astype(str) if col(["game_date", "gameDate"]) is not None else None
    df["game_id"] = col(["game_id", "gameId"]).astype(str) if col(["game_id", "gameId"]) is not None else ""
    df["team_id"] = col(["team_id", "teamId"])
    df["team_city"] = col(["team_city", "teamCity", "team_location"], default="")
    df["team_name"] = col(["team_name", "teamName"], default="")
    df["team_abbreviation"] = col(["team_abbreviation", "teamTricode", "team_tricode", "team_abbrev"], default="")

    df["player_id"] = col(["player_id", "athlete_id", "person_id", "personId"], default="")
    df["player_name"] = col(["player_name", "athlete_display_name", "person_name", "personName"], default="")
    df["start_position"] = col(["start_position", "position", "athlete_position_abbreviation", "athlete_position_name"], default=None)
    df["comment"] = col(["comment", "reason"], default=None)

    df["min"] = col(["min", "minutes"], default=None)
    df["fgm"] = col(["fgm", "field_goals_made"], default=0)
    df["fga"] = col(["fga", "field_goals_attempted"], default=0)
    df["fg_pct"] = col(["fg_pct", "field_goal_pct"], default=0.0)

    df["fg3m"] = col(["fg3m", "three_point_field_goals_made"], default=0)
    df["fg3a"] = col(["fg3a", "three_point_field_goals_attempted"], default=0)
    df["fg3_pct"] = col(["fg3_pct", "three_point_field_goal_pct"], default=0.0)

    df["ftm"] = col(["ftm", "free_throws_made"], default=0)
    df["fta"] = col(["fta", "free_throws_attempted"], default=0)
    df["ft_pct"] = col(["ft_pct", "free_throw_pct"], default=0.0)

    df["oreb"] = col(["oreb", "offensive_rebounds"], default=0)
    df["dreb"] = col(["dreb", "defensive_rebounds"], default=0)
    df["reb"] = col(["reb", "rebounds"], default=0)

    df["ast"] = col(["ast", "assists"], default=0)
    df["stl"] = col(["stl", "steals"], default=0)
    df["blk"] = col(["blk", "blocks"], default=0)
    df["tov"] = col(["tov", "turnovers"], default=0)
    df["pf"] = col(["pf", "fouls"], default=0)
    df["pts"] = col(["pts", "points"], default=0)
    df["plus_minus"] = col(["plus_minus", "plusMinusPoints", "plusminus"], default=0)

    if is_regular:
        opp_col = col(["opponent_team_abbreviation", "opp_team_abbreviation", "matchup"])
        if opp_col is not None and "matchup" in [c.lower() for c in raw_df.columns]:
            df["matchup"] = opp_col
        elif opp_col is not None and "team_abbreviation" in raw_cols:
            df["matchup"] = df["team_abbreviation"] + " @ " + opp_col.astype(str)
        else:
            df["matchup"] = "Regular"
    else:
        df["matchup"] = None

    return df

# ============================================
# MAP TO NOCTURNEBEAR SCHEMA
# ============================================
def map_to_nocturnebear(df):
    df2 = pd.DataFrame()

    df2["season_year"] = df["season_year"]
    df2["game_date"] = df["game_date"]
    df2["gameId"] = df["game_id"]

    df2["teamId"] = df["team_id"]
    df2["teamCity"] = df["team_city"]
    df2["teamName"] = df["team_name"]
    df2["teamTricode"] = df["team_abbreviation"]
    df2["teamSlug"] = df["team_name"].str.lower().str.replace(" ", "-")

    df2["personId"] = df["player_id"]
    df2["personName"] = df["player_name"]
    df2["position"] = df["start_position"]
    df2["comment"] = df["comment"]
    df2["jerseyNum"] = None

    df2["minutes"] = df["min"]

    df2["fieldGoalsMade"] = df["fgm"]
    df2["fieldGoalsAttempted"] = df["fga"]
    df2["fieldGoalsPercentage"] = df["fg_pct"]

    df2["threePointersMade"] = df["fg3m"]
    df2["threePointersAttempted"] = df["fg3a"]
    df2["threePointersPercentage"] = df["fg3_pct"]

    df2["freeThrowsMade"] = df["ftm"]
    df2["freeThrowsAttempted"] = df["fta"]
    df2["freeThrowsPercentage"] = df["ft_pct"]

    df2["reboundsOffensive"] = df["oreb"]
    df2["reboundsDefensive"] = df["dreb"]
    df2["reboundsTotal"] = df["reb"]

    df2["assists"] = df["ast"]
    df2["steals"] = df["stl"]
    df2["blocks"] = df["blk"]
    df2["turnovers"] = df["tov"]
    df2["foulsPersonal"] = df["pf"]
    df2["points"] = df["pts"]
    df2["plusMinusPoints"] = df["plus_minus"]

    df2["matchup"] = df["matchup"]

    return df2

# ============================================
# LOAD EXISTING SINGLE MASTER FILE
# ============================================
start_season = SEASONS[0]
end_season = SEASONS[-1]
master_path = f"{boxscore_folder}boxscores_regular_season_and_playoffs_{start_season}_{end_season}.parquet"

existing_files = glob.glob(os.path.join(boxscore_folder, "boxscores_regular_season_and_playoffs_*.parquet"))

old_master_path = None
df_master = pd.DataFrame()
processed_seasons = set()

if existing_files:
    old_master_path = existing_files[0]
    print(f"📦 Loading existing file: {old_master_path}")
    df_master = pd.read_parquet(old_master_path)
    if "season_year" in df_master.columns:
        processed_seasons = set(df_master["season_year"].unique())
        print(f"⏩ Seasons already present: {sorted(list(processed_seasons))}")

seasons_to_fetch = [s for s in SEASONS if s not in processed_seasons]
new_rows = []

# ============================================
# PROCESS MISSING SEASONS
# ============================================
if not seasons_to_fetch:
    print("✨ All seasons are already up to date!")
else:
    for season in tqdm(seasons_to_fetch, desc="Amending seasons"):
        print(f"\n==============================")
        print(f"📅 Fetching season {season}")
        print("==============================")

        sdv_season = season + 1  # SportsDataverse references seasons by ending year (e.g. 2010-11 is 2011)

        try:
            df_box_logs = sdv.nba.nba_box_logs(season=sdv_season)
        except Exception:
            df_box_logs = sdv.nba.nba_box_logs(sdv_season)

        if isinstance(df_box_logs, dict):
            target_df = None
            for k, v in df_box_logs.items():
                if "player" in k.lower():
                    target_df = v
                    break
            if target_df is None:
                for v in df_box_logs.values():
                    if hasattr(v, "columns") or hasattr(v, "to_pandas") or isinstance(v, (pd.DataFrame, list)):
                        target_df = v
                        break
            df_box_logs = target_df if target_df is not None else pd.DataFrame(df_box_logs)

        if hasattr(df_box_logs, "to_pandas"):
            df_box_logs = df_box_logs.to_pandas()
        elif isinstance(df_box_logs, list):
            df_box_logs = pd.DataFrame(df_box_logs)
        elif not isinstance(df_box_logs, pd.DataFrame):
            df_box_logs = pd.DataFrame(df_box_logs)

        # Filter out team summary rows if player rows are present
        player_id_col = None
        for cand in ["athlete_id", "player_id", "person_id", "personId"]:
            if cand in df_box_logs.columns:
                player_id_col = cand
                break
        if player_id_col is not None:
            df_box_logs = df_box_logs[df_box_logs[player_id_col].notna() & (df_box_logs[player_id_col] != "") & (df_box_logs[player_id_col] != 0)]

        # Separate regular season and playoffs
        if "season_type" in df_box_logs.columns:
            is_reg = (df_box_logs["season_type"] == 2) | (df_box_logs["season_type"].astype(str).str.lower().str.contains("regular"))
            is_playoff = (df_box_logs["season_type"] == 3) | (df_box_logs["season_type"].astype(str).str.lower().str.contains("playoff|post"))
            df_reg_raw = df_box_logs[is_reg]
            df_play_raw = df_box_logs[is_playoff]
        else:
            df_reg_raw = df_box_logs
            df_play_raw = pd.DataFrame()

        game_id_col = "game_id" if "game_id" in df_box_logs.columns else ("gameId" if "gameId" in df_box_logs.columns else None)
        total_games = df_box_logs[game_id_col].nunique() if game_id_col else len(df_box_logs)
        print(f"📘 Games found: {total_games}")

        if len(df_reg_raw) > 0:
            df_regular = prepare_sdv_df(df_reg_raw, season, is_regular=True)
            df_regular = map_to_nocturnebear(df_regular)
            new_rows.append(df_regular)

        if len(df_play_raw) > 0:
            df_playoffs = prepare_sdv_df(df_play_raw, season, is_regular=False)
            df_playoffs = map_to_nocturnebear(df_playoffs)
            new_rows.append(df_playoffs)

# ============================================
# AMEND AND SAVE SINGLE MASTER FILE
# ============================================
print("\n📦 Updating single master file...")

if new_rows:
    df_new = pd.concat(new_rows, ignore_index=True)
    df_master = pd.concat([df_master, df_new], ignore_index=True) if not df_master.empty else df_new

df_master["season_type"] = df_master["matchup"].apply(lambda x: "Regular" if pd.notna(x) else "Playoffs")

# Remove previous file if the end season changed
if old_master_path and os.path.exists(old_master_path) and old_master_path != master_path:
    try:
        os.remove(old_master_path)
        print(f"🗑️ Removed previous file: {old_master_path}")
    except Exception as e:
        print(f"⚠️ Could not remove previous file: {e}")

df_master.to_parquet(master_path)

print(f"🌟 Single master file saved: {master_path}")
print("✨ Boxscore update complete.")
```

```python
#@title SCRIPT 1: Restored Feature Engineering Engine (v8.4 - ADAPTED FOR STATS-API BOXSCORES)
!pip install -q pandas unidecode tqdm scikit-learn pyarrow fastparquet

import pandas as pd
import numpy as np
import os
import re
import pytz
from datetime import datetime
from google.colab import drive
from unidecode import unidecode
from tqdm import tqdm
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# --- Configuration ---
N_PLAYER_TRAITS = 10
ROLLING_WINDOWS = [5, 10, 20]
MODERN_ERA_START_SEASON = '2015-16'

# --- Helper Functions ---
def load_historical_odds(file_path):
    print("--- Loading and Processing Historical Odds Data ---")
    try:
        df = pd.read_csv(file_path)
    except FileNotFoundError:
        print(f"⚠️ Warning: Historical odds file not found at {file_path}")
        return pd.DataFrame()

    team_map = {
        'atl': 'ATL','bkn': 'BKN','bos': 'BOS','cha': 'CHA','chi': 'CHI','cle': 'CLE','dal': 'DAL','den': 'DEN',
        'det': 'DET','gs': 'GSW','hou': 'HOU','ind': 'IND','lac': 'LAC','lal': 'LAL','mem': 'MEM','mia': 'MIA',
        'mil': 'MIL','min': 'MIN','no': 'NOP','ny': 'NYK','okc': 'OKC','orl': 'ORL','phi': 'PHI','phx': 'PHX',
        'por': 'POR','sa': 'SAS','sac': 'SAC','tor': 'TOR','utah': 'UTA','wsh': 'WAS'
    }

    df['home'] = df['home'].str.lower().map(team_map)
    df['away'] = df['away'].str.lower().map(team_map)
    df.dropna(subset=['home', 'away'], inplace=True)

    df['GAME_DATE'] = pd.to_datetime(df['date'])
    df['VEGAS_SPREAD'] = np.where(df['whos_favored'] == 'home', -df['spread'], df['spread'])
    df.rename(columns={'total': 'VEGAS_TOTAL'}, inplace=True)

    df['game_pair'] = [frozenset(x) for x in zip(df['home'], df['away'])]

    print("✅ Historical odds data processed successfully.")
    return df[['GAME_DATE', 'game_pair', 'VEGAS_SPREAD', 'VEGAS_TOTAL']]

def get_current_nba_season():
    today = datetime.now()
    year, month = today.year, today.month
    return f"{year}-{str(year+1)[-2:]}" if month >= 9 else f"{year-1}-{str(year)[-2:]}"

def get_last_completed_nba_season():
    today = datetime.now()
    year, month = today.year, today.month
    # The NBA season ends in June. From July onward (preseason), the completed season ended in the current year.
    return f"{year-1}-{str(year)[-2:]}" if month >= 7 else f"{year-2}-{str(year-1)[-2:]}"

def normalize_player_name(name):
    if not isinstance(name, str): return None
    name = unidecode(name.lower())
    name = re.sub(r'\s+(jr|sr|ii|iii|iv)\.?$', '', name)
    name = re.sub(r'[^\w\s]', '', name)
    return re.sub(r'\s+', ' ', name).strip()

def convert_min_to_decimal(min_str):
    if isinstance(min_str, str) and ':' in min_str:
        try:
            minutes, seconds = map(int, min_str.split(':'))
            return minutes + seconds / 60.0
        except:
            return np.nan
    return pd.to_numeric(min_str, errors='coerce')

def calculate_weighted_rolling_averages(group, stats_to_roll):
    avg_minutes = group['MIN'].mean()
    weights = np.ones(len(group))
    weights[group['MIN'] < 0.7 * avg_minutes] = 0.5
    weights[group['MIN'] > 1.15 * avg_minutes] = 1.2
    group['WEIGHT'] = weights

    for stat in stats_to_roll:
        weighted_stat = group[stat] * group['WEIGHT']
        for w in ROLLING_WINDOWS:
            rolling_weighted_sum = weighted_stat.shift(1).rolling(w, min_periods=1).sum()
            rolling_weight_sum = group['WEIGHT'].shift(1).rolling(w, min_periods=1).sum()
            group[f'W_ROLL_AVG_{stat}_{w}'] = rolling_weighted_sum / rolling_weight_sum

    return group

def derive_player_traits(df_for_clustering):
    stats_per_min = df_for_clustering[df_for_clustering['MIN'] > 5].copy()
    if stats_per_min.empty:
        return pd.DataFrame()

    stat_cols = ['PTS','REB','AST','STL','BLK','FGA','FG3A','FG3M','FTA']
    for col in stat_cols:
        stats_per_min[f'{col}_PM_raw'] = stats_per_min[col] / stats_per_min['MIN']

    player_profiles = stats_per_min.groupby(
        ['SEASON_YEAR','PLAYER_ID','PLAYER_NAME']
    )[[f'{col}_PM_raw' for col in stat_cols]].mean().reset_index()

    if player_profiles.empty:
        return pd.DataFrame()

    features = player_profiles[[f'{col}_PM_raw' for col in stat_cols]].fillna(0)
    scaler = StandardScaler()
    scaled = scaler.fit_transform(features)

    n_samples = len(features)
    n_components = min(0.95, n_samples) if n_samples > 1 else 1
    pca = PCA(n_components=n_components)
    reduced = pca.fit_transform(scaled)

    n_clusters = min(N_PLAYER_TRAITS, n_samples)
    if n_clusters < 2:
        player_profiles['TRAIT'] = 0
    else:
        kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init='auto')
        player_profiles['TRAIT'] = kmeans.fit_predict(reduced)

    return player_profiles[['SEASON_YEAR','PLAYER_ID','TRAIT']]

def process_nba_data_v8_4():
    drive.mount('/content/drive', force_remount=True)

    BASE_DIR = "/content/drive/MyDrive/NBA/"
    DATA_DIR = os.path.join(BASE_DIR, "Historical Data/")
    os.makedirs(DATA_DIR, exist_ok=True)

    print(f"📁 Target Folder Verified/Created on Google Drive: {DATA_DIR}")

    # Stats-API boxscores folder
    BOXSCORE_DIR = os.path.join(DATA_DIR, "Boxscores")
    # Load coaches
    coach_path = os.path.join(DATA_DIR, "Coaches 2015 to Present.csv")
    coach_df = pd.read_csv(coach_path) if os.path.exists(coach_path) else pd.DataFrame()

    # Load odds
    odds_df = load_historical_odds(os.path.join(DATA_DIR, "nba_historic_odds_2007_2026.csv"))

    print("\nStep 1: Loading and cleaning box score data...")

    # Load Parquet file(s)
    file_paths = [
        os.path.join(BOXSCORE_DIR, f)
        for f in os.listdir(BOXSCORE_DIR)
        if f.endswith('.parquet')
    ]
    if not file_paths:
        raise FileNotFoundError(f"❌ No .parquet files found in {BOXSCORE_DIR}. Please run the boxscore scraping script first.")

    df = pd.concat([pd.read_parquet(path) for path in file_paths], ignore_index=True)
    print(f"📥 Loaded {len(df)} raw rows from: {[os.path.basename(p) for p in file_paths]}")

    id_col = 'gameId' if 'gameId' in df.columns else 'game_id'
    pid_col = 'personId' if 'personId' in df.columns else 'player_id'
    if id_col in df.columns and pid_col in df.columns:
        df.drop_duplicates(subset=[id_col, pid_col], inplace=True)

    # Coincide both camelCase (NocturneBear) and snake_case headers
    rename_map = {
        'season_year': 'SEASON_YEAR',
        'seasonYear': 'SEASON_YEAR',
        'season': 'SEASON_YEAR',
        'game_date': 'GAME_DATE',
        'gameId': 'GAME_ID',
        'game_id': 'GAME_ID',
        'personId': 'PLAYER_ID',
        'person_id': 'PLAYER_ID',
        'personName': 'PLAYER_NAME',
        'player_name': 'PLAYER_NAME',
        'teamTricode': 'TEAM_ABBREVIATION',
        'team_tricode': 'TEAM_ABBREVIATION',

        'points': 'PTS',
        'reboundsTotal': 'REB',
        'rebounds_total': 'REB',
        'assists': 'AST',
        'minutes': 'MIN',
        'fieldGoalsAttempted': 'FGA',
        'field_goals_attempted': 'FGA',
        'fieldGoalsMade': 'FGM',
        'field_goals_made': 'FGM',
        'freeThrowsAttempted': 'FTA',
        'free_throws_attempted': 'FTA',
        'threePointersAttempted': 'FG3A',
        'three_pointers_attempted': 'FG3A',
        'threePointersMade': 'FG3M',
        'three_pointers_made': 'FG3M',
        'steals': 'STL',
        'blocks': 'BLK',
        'turnovers': 'TOV'
    }
    df.rename(columns=rename_map, inplace=True)

    # Clean dates
    if 'GAME_DATE' in df.columns:
        df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'], errors='coerce')

    # Standardize SEASON_YEAR or derive from GAME_DATE if missing / NaN
    def derive_season(row):
        val = row.get('SEASON_YEAR', None)
        if pd.notna(val) and str(val).strip().lower() not in ['nan', 'none', '']:
            s = str(val).strip()
            if '-' in s and len(s) >= 7:
                return s
            try:
                y = int(float(s))
                return f"{y}-{str(y+1)[-2:]}"
            except Exception:
                pass
        # Fallback to GAME_DATE
        dt = row.get('GAME_DATE', None)
        if pd.notna(dt):
            year = dt.year
            month = dt.month
            start_year = year if month >= 10 else year - 1
            return f"{start_year}-{str(start_year+1)[-2:]}"
        return "Unknown"

    df['SEASON_YEAR'] = df.apply(derive_season, axis=1)
    df = df[df['SEASON_YEAR'] != "Unknown"].copy()

    # Clean minutes
    if 'MIN' in df.columns:
        df['MIN'] = df['MIN'].apply(convert_min_to_decimal)
        df = df[df['MIN'].notna() & (df['MIN'] > 0)].copy()

    # Convert numeric columns
    numeric_cols = ['PTS','REB','AST','MIN','FGM','FGA','FTA','FG3A','FG3M','STL','BLK','TOV']
    for col in numeric_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)
        else:
            df[col] = 0.0

    # Ensure required ID and date columns exist
    valid_id_cols = [c for c in ['PLAYER_ID', 'GAME_DATE', 'MIN'] if c in df.columns]
    df.dropna(subset=valid_id_cols, inplace=True)

    # Auto-update training season cutoff to the last completed season
    TRAINING_MAX_SEASON_STR = get_last_completed_nba_season()
    print(f"🎯 Training cutoff automatically set to completed season: {TRAINING_MAX_SEASON_STR}")

    # Dynamic season filter: keep from 2015-16 through the last completed season
    if 'SEASON_YEAR' in df.columns and not df.empty:
        has_modern = (df['SEASON_YEAR'] >= MODERN_ERA_START_SEASON).any()
        if has_modern:
            df = df[(df['SEASON_YEAR'] >= MODERN_ERA_START_SEASON) & (df['SEASON_YEAR'] <= TRAINING_MAX_SEASON_STR)].copy()
        else:
            df = df[df['SEASON_YEAR'] <= TRAINING_MAX_SEASON_STR].copy()

    df.sort_values(by=['PLAYER_ID','GAME_DATE'], inplace=True)

    print(f"📊 Rows after cleaning: {len(df)} across seasons: {sorted(df['SEASON_YEAR'].unique()) if 'SEASON_YEAR' in df.columns else []}")
    if df.empty:
        raise ValueError("❌ No valid rows remain after Step 1 cleaning. Please inspect the raw parquet file columns and contents.")

    # Shooting efficiency
    df['EFG_PCT_RAW'] = (df['FGM'] + 0.5 * df['FG3M']) / df['FGA'].replace(0, np.nan)
    df['TS_PCT_RAW'] = df['PTS'] / (2 * (df['FGA'] + 0.44 * df['FTA'])).replace(0, np.nan)
    df['EFG_PCT_RAW'] = df['EFG_PCT_RAW'].fillna(0)
    df['TS_PCT_RAW'] = df['TS_PCT_RAW'].fillna(0)

    # Playoff flag
    if 'season_type' in df.columns:
        df['IS_PLAYOFF_GAME'] = (df['season_type'] == 'Playoffs').astype(int)
    else:
        df['IS_PLAYOFF_GAME'] = df['GAME_ID'].astype(str).apply(
            lambda x: 1 if x.startswith('004') or x.startswith('4') else 0
        )

    print("\nStep 2: Merging historical odds data...")
    opponent_df = df[['GAME_ID','TEAM_ABBREVIATION']].drop_duplicates()
    df = pd.merge(
        df,
        opponent_df.rename(columns={'TEAM_ABBREVIATION':'OPPONENT_ABBR'}),
        on='GAME_ID'
    )
    df = df[df['TEAM_ABBREVIATION'] != df['OPPONENT_ABBR']].copy()

    if not odds_df.empty:
        df['game_pair'] = [
            frozenset(x) for x in zip(df['TEAM_ABBREVIATION'], df['OPPONENT_ABBR'])
        ]
        df = pd.merge(df, odds_df, on=['GAME_DATE','game_pair'], how='left')
        df.drop(columns=['game_pair'], inplace=True)
    else:
        df['VEGAS_SPREAD'] = np.nan
        df['VEGAS_TOTAL'] = np.nan

    print("\nStep 3: Engineering functional role ratios (Shifted)...")
    stats_for_ratios = ['PTS','REB','AST','FGA','FG3A','FTA','TOV','FG3M']
    for stat in stats_for_ratios:
        df[f'{stat}_PM'] = df[stat] / df['MIN'].replace(0, np.nan)
        df[f'{stat}_PM'] = df.groupby('PLAYER_ID')[f'{stat}_PM'].shift(1)

    df['EFG_PCT_PREV'] = df.groupby('PLAYER_ID')['EFG_PCT_RAW'].shift(1)
    df['TS_PCT_PREV'] = df.groupby('PLAYER_ID')['TS_PCT_RAW'].shift(1)

    print("\nStep 4: Mapping unsupervised player traits (DvT)...")
    all_seasons_traits = []
    for season in sorted(df['SEASON_YEAR'].unique()):
        season_df = df[df['SEASON_YEAR'] == season].copy()
        trait_map = derive_player_traits(season_df)
        if not trait_map.empty:
            all_seasons_traits.append(trait_map)

    if all_seasons_traits:
        player_trait_map = pd.concat(all_seasons_traits, ignore_index=True)
        player_trait_map.to_csv(os.path.join(DATA_DIR, "player_trait_map.csv"), index=False)
        df = pd.merge(df, player_trait_map, on=['SEASON_YEAR','PLAYER_ID'], how='left')
        df['TRAIT'] = df['TRAIT'].fillna(-1).astype(int)
    else:
        print("⚠️ Warning: No player traits derived. Assigning default trait 0.")
        df['TRAIT'] = 0

    print("\nStep 5: Mapping historical coaching tenures & continuity...")
    df['season_start_year'] = df['SEASON_YEAR'].apply(
        lambda x: int(str(x).split('-')[0]) if str(x).split('-')[0].isdigit() else 2015
    )

    if not coach_df.empty:
        def find_active_coach(row):
            team = row['TEAM_ABBREVIATION']
            year = row['season_start_year']
            match = coach_df[
                (coach_df['team'] == team) &
                (coach_df['from_year'] <= year) &
                (coach_df['to_year'] >= year)
            ]
            if not match.empty:
                return match.iloc[0]['coach']
            return "Unknown"

        df['ACTIVE_COACH'] = df.apply(find_active_coach, axis=1)

        df.sort_values(by=['TEAM_ABBREVIATION','GAME_DATE'], inplace=True)
        df['COACH_GAMES_PLAYED'] = df.groupby(
            ['TEAM_ABBREVIATION','ACTIVE_COACH']
        ).cumcount()
    else:
        df['COACH_GAMES_PLAYED'] = 0

    print("\nStep 6: Engineering Weighted Rolling Averages & Trends (Restored)...")
    df.sort_values(by=['PLAYER_ID','GAME_DATE'], inplace=True)

    stats_to_roll = [
        'PTS','REB','AST','MIN','FG3M','FG3A','EFG_PCT_RAW','TS_PCT_RAW'
    ]

    res_list = []
    for pid, group in tqdm(df.groupby('PLAYER_ID', sort=False), desc="Weighted Rolling Averages"):
        res_list.append(calculate_weighted_rolling_averages(group, stats_to_roll))

    df = pd.concat(res_list, ignore_index=True)

    for stat in stats_to_roll:
        df[f'TREND_{stat}'] = df[f'W_ROLL_AVG_{stat}_5'] - df[f'W_ROLL_AVG_{stat}_20']

    df.sort_values(by=['PLAYER_ID','GAME_DATE'], inplace=True)
    df['DAYS_OF_REST'] = df.groupby('PLAYER_ID')['GAME_DATE'].transform(
        lambda x: x.diff().dt.days - 1
    ).clip(upper=7).fillna(7)

    print("\nStep 7: Engineering Overall Opponent Context Features...")
    team_allowed_stats = df.groupby(
        ['GAME_ID','OPPONENT_ABBR']
    ).agg(
        GAME_DATE=('GAME_DATE','first'),
        PTS=('PTS','sum'),
        REB=('REB','sum'),
        AST=('AST','sum'),
        FG3M=('FG3M','sum'),
        EFG_PCT=('EFG_PCT_RAW','mean')
    ).reset_index()

    team_allowed_stats.rename(columns={'OPPONENT_ABBR':'TEAM'}, inplace=True)
    team_allowed_stats.sort_values(by=['TEAM','GAME_DATE'], inplace=True)

    for stat in ['PTS','REB','AST','FG3M','EFG_PCT']:
        team_allowed_stats[f'OPP_ALLOWED_{stat}_AVG_30'] = team_allowed_stats.groupby(
            'TEAM'
        )[stat].transform(lambda x: x.shift(1).rolling(30, min_periods=5).mean())

    df = pd.merge(
        df,
        team_allowed_stats[
            ['GAME_ID','TEAM',
             'OPP_ALLOWED_PTS_AVG_30','OPP_ALLOWED_REB_AVG_30',
             'OPP_ALLOWED_AST_AVG_30','OPP_ALLOWED_FG3M_AVG_30',
             'OPP_ALLOWED_EFG_PCT_AVG_30']
        ],
        left_on=['GAME_ID','OPPONENT_ABBR'],
        right_on=['GAME_ID','TEAM'],
        how='left'
    )
    df.drop(columns=['TEAM'], inplace=True, errors='ignore')

    print("\nStep 8: Engineering Advanced Defense-vs-Trait (DvT)...")
    trait_allowed_stats = df.groupby(
        ['GAME_ID','GAME_DATE','OPPONENT_ABBR','TRAIT']
    )[['PTS','REB','AST','FG3M']].sum().reset_index()

    trait_allowed_stats.rename(columns={'OPPONENT_ABBR':'TEAM'}, inplace=True)
    trait_allowed_stats.sort_values(by=['TEAM','GAME_DATE'], inplace=True)

    dvp_grouped = trait_allowed_stats.groupby(['TEAM','TRAIT'])
    for stat in ['PTS','REB','AST','FG3M']:
        trait_allowed_stats[f'OPP_DvT_{stat}'] = dvp_grouped[stat].transform(
            lambda x: x.shift(1).rolling(30, min_periods=5).mean()
        )

    dvp_cols = ['GAME_ID','TEAM','TRAIT'] + [
        c for c in trait_allowed_stats.columns if c.startswith('OPP_DvT')
    ]

    df = pd.merge(
        df,
        trait_allowed_stats[dvp_cols],
        left_on=['GAME_ID','OPPONENT_ABBR','TRAIT'],
        right_on=['GAME_ID','TEAM','TRAIT'],
        how='left'
    )

    print("\nStep 9: Finalizing data and saving training set...")
    for stat in ['MIN','PTS','REB','AST','FG3M']:
        df[f'TARGET_{stat}'] = df[stat]

    dd_td_stats = ['PTS','REB','AST','STL','BLK']
    df['DD_COUNT'] = df[dd_td_stats].ge(10).sum(axis=1)
    df['TARGET_DD'] = (df['DD_COUNT'] >= 2).astype(int)
    df['TARGET_TD'] = (df['DD_COUNT'] >= 3).astype(int)

    feature_columns = [
        c for c in df.columns
        if 'W_ROLL_AVG' in c
        or 'TREND' in c
        or 'OPP_DvT' in c
        or 'OPP_ALLOWED' in c
        or '_PM' in c
        or 'VEGAS' in c
        or '_PREV' in c
    ] + ['DAYS_OF_REST', 'COACH_GAMES_PLAYED', 'IS_PLAYOFF_GAME']

    target_columns = [c for c in df.columns if 'TARGET' in c]

    # Build final training dataset
    final_df = df[
        ['GAME_DATE', 'PLAYER_ID', 'PLAYER_NAME'] +
        feature_columns +
        target_columns
    ].copy()

    # Drop rows missing targets or rolling averages
    final_df.dropna(
        subset=[c for c in target_columns if c not in ['TARGET_DD', 'TARGET_TD']],
        inplace=True
    )
    final_df.dropna(
        subset=[c for c in feature_columns if 'W_ROLL_AVG' in c],
        inplace=True
    )

    # Save output explicitly
    output_path = os.path.join(DATA_DIR, "nba_advanced_training_data_v8.csv")
    print(f"💾 Writing {len(final_df)} rows to {output_path}...")
    final_df.to_csv(output_path, index=False)

    # Verify write
    if os.path.exists(output_path):
        size_mb = os.path.getsize(output_path) / (1024 * 1024)
        print(f"\n✅ Training dataset verified on disk: {output_path} ({size_mb:.2f} MB)")
        print(f"📊 Total rows: {len(final_df)}")
        print("🎯 SCRIPT 1 adaptation complete — fully compatible with Stats‑API boxscores.")
    else:
        raise IOError(f"❌ Failed to verify file on disk: {output_path}")

# Run script
if __name__ == "__main__":
    tqdm.pandas()
    process_nba_data_v8_4()
```

### **SCRIPT 2: Model Training Engine (v8.3)**

This script serves as the machine learning core of the pipeline. It reads the advanced, leak-free training dataset (`nba_advanced_training_data_v8.csv`) and executes a multi-stage training routine using LightGBM and Optuna hyperparameter optimization.

#### **Key Operations & Features:**
1. **Chronological Validation Split:** Employs a time-series split (using the last 2 months of historical data as a test window) to validate predictions and prevent future-data leakage during training.
2. **Optuna Hyperparameter Tuning:** Dynamically searches for the most efficient model parameters (learning rate, leaves, depth, regularization) over 20 trials for each individual target metric.
3. **Multi-Stage Training Pipeline:**
   * **Stage 1 (Minutes):** Trains the playing-time model based on historical rest, performance trends, and coaching tenures.
   * **Stage 2 (Core Stats):** Feeds the predicted minutes into the statistical models to project Points, Rebounds, Assists, and Threes, incorporating Defense-vs-Trait (DvT) matchups and playoff flags.
   * **Stage 3 (Classifiers):** Uses the predicted stat outputs to train balanced binary classifiers for Double-Double (DD) and Triple-Double (TD) probability.
4. **Google Drive Export:** Saves the final trained models as lightweight `.pkl` files and exports validation errors to `mae_values.json` for live error-range calculations.

```python
#@title SCRIPT 2: Model Training Engine (v8.3 - LEAK-FREE, COHESIVE & PLAYOFF-AWARE)
# ==============================================================================
!pip install -q lightgbm scikit-learn pandas joblib optuna tqdm

import pandas as pd
import numpy as np
import joblib
import os
import json
import lightgbm as lgb
import optuna
import warnings
from tqdm import tqdm
from google.colab import drive
from sklearn.metrics import mean_squared_error, mean_absolute_error, roc_auc_score
from sklearn.impute import SimpleImputer
from sklearn.model_selection import KFold

# Suppress feature name warnings and Optuna verbosity
warnings.filterwarnings("ignore", category=UserWarning)
optuna.logging.set_verbosity(optuna.logging.WARNING)

# --- Optuna Objective Function ---
def objective_lgbm(trial, X, y):
    params = {
        'objective': 'rmse',
        'metric': 'rmse',
        'verbosity': -1,
        'boosting_type': 'gbdt',
        'random_state': 42,
        'n_jobs': -1,
        'n_estimators': trial.suggest_int('n_estimators', 100, 1000),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3),
        'num_leaves': trial.suggest_int('num_leaves', 20, 150),
        'max_depth': trial.suggest_int('max_depth', 3, 12),
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 100),
        'subsample': trial.suggest_float('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0),
        'reg_alpha': trial.suggest_float('reg_alpha', 0.0, 1.0),
        'reg_lambda': trial.suggest_float('reg_lambda', 0.0, 1.0)
    }

    cv = KFold(n_splits=3, shuffle=True, random_state=42)
    rmse_scores = []

    # Fast Cross-Validation for Tuning
    for train_idx, val_idx in cv.split(X, y):
        X_train_fold, X_val_fold = X[train_idx], X[val_idx]
        y_train_fold, y_val_fold = y.iloc[train_idx], y.iloc[val_idx]

        model = lgb.LGBMRegressor(**params)
        model.fit(X_train_fold, y_train_fold)
        preds = model.predict(X_val_fold)
        rmse_scores.append(np.sqrt(mean_squared_error(y_val_fold, preds)))

    return np.mean(rmse_scores)

def run_optuna_tuning(X, y, target_name, n_trials=20):
    print(f"   >>> Tuning Hyperparameters for {target_name}...")
    study = optuna.create_study(direction='minimize')

    # Subsample for fast tuning if dataset is large, while final model trains on 100%
    if len(X) > 40000:
        sample_idx = np.random.RandomState(42).choice(len(X), size=40000, replace=False)
        X_tune = X[sample_idx]
        y_tune = y.iloc[sample_idx].reset_index(drop=True)
    else:
        X_tune = X
        y_tune = y.reset_index(drop=True)

    # Progress bar for trials
    with tqdm(total=n_trials, desc=f"      {target_name} Tuning", unit="trial", leave=True) as pbar:
        def callback(study, trial):
            pbar.update(1)
        study.optimize(lambda trial: objective_lgbm(trial, X_tune, y_tune), n_trials=n_trials, callbacks=[callback])

    print(f"   ✅ Best params for {target_name}: {study.best_params}")
    return study.best_params

def train_and_validate_models_v8_3():
    drive.mount('/content/drive', force_remount=True)
    BASE_DIR = "/content/drive/MyDrive/NBA/"
    DATA_DIR = os.path.join(BASE_DIR, "Historical Data/")
    MODEL_DIR = os.path.join(BASE_DIR, "Training Model Advanced")
    os.makedirs(MODEL_DIR, exist_ok=True)

    # Read the updated v8 dataset with Trait and Coaching features
    df = pd.read_csv(os.path.join(DATA_DIR, "nba_advanced_training_data_v8.csv"))
    df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])
    print(f"Loaded {len(df)} rows from v8 file.")

    print("\n--- Performing Chronological Train/Test Split ---")
    df = df.sort_values('GAME_DATE')
    split_date = df['GAME_DATE'].max() - pd.DateOffset(months=2)
    train_df = df[df['GAME_DATE'] < split_date].copy()
    test_df = df[df['GAME_DATE'] >= split_date].copy()
    print(f"Training data: {len(train_df)} rows | Test data: {len(test_df)} rows")
    mae_values = {}

    print("\n--- STAGE 1: Training & Validating Minutes Model (w/ Optuna) ---")
    # Added Coaching Games and Playoff flags directly into the minutes training features
    min_features = [c for c in df.columns if 'W_ROLL_AVG_MIN' in c or 'TREND_MIN' in c] + ['DAYS_OF_REST', 'COACH_GAMES_PLAYED', 'IS_PLAYOFF_GAME']
    X_train_min, y_train_min = train_df[min_features], train_df['TARGET_MIN']
    X_test_min, y_test_min = test_df[min_features], test_df['TARGET_MIN']

    imputer_min = SimpleImputer(strategy='mean')
    X_train_min_imputed = imputer_min.fit_transform(X_train_min)

    # Optuna tuning for minutes
    best_params_min = run_optuna_tuning(X_train_min_imputed, y_train_min, "Minutes")
    minutes_model = lgb.LGBMRegressor(**best_params_min, random_state=42, n_jobs=-1)
    minutes_model.fit(X_train_min_imputed, y_train_min)

    X_test_min_imputed = imputer_min.transform(X_test_min)
    preds_min = minutes_model.predict(X_test_min_imputed)
    mae_values['MIN'] = mean_absolute_error(y_test_min, preds_min)
    print(f"✅ Minutes Model Validation RMSE: {np.sqrt(mean_squared_error(y_test_min, preds_min)):.3f} | MAE: {mae_values['MIN']:.3f}")

    print("\n--- STAGE 2: Training & Validating Core Stats Models (w/ Optuna) ---")
    # Adding predicted minutes as a feature for stats
    train_df['Proj_MIN'] = minutes_model.predict(X_train_min_imputed)
    test_df['Proj_MIN'] = minutes_model.predict(X_test_min_imputed)

    # Added Coaching Games, Playoff flags, and DvT variables directly into stats training features
    stat_features = [c for c in df.columns if ('W_ROLL_AVG' in c or 'TREND' in c or 'OPP_ALLOWED' in c or 'OPP_DvT' in c or '_PM' in c or 'VEGAS' in c or '_PREV' in c) and 'MIN' not in c] + ['DAYS_OF_REST', 'Proj_MIN', 'COACH_GAMES_PLAYED', 'IS_PLAYOFF_GAME']

    X_train_stats = train_df[stat_features]
    X_test_stats = test_df[stat_features]

    imputer_stats = SimpleImputer(strategy='mean')
    X_train_stats_imputed = imputer_stats.fit_transform(X_train_stats)
    X_test_stats_imputed = imputer_stats.transform(X_test_stats)

    test_preds = {}
    best_params_stats = {}

    # Optuna tuning for stats
    for prop in ['PTS', 'REB', 'AST', 'FG3M']:
        best_params_stats[prop] = run_optuna_tuning(X_train_stats_imputed, train_df[f'TARGET_{prop}'], prop)
        model = lgb.LGBMRegressor(**best_params_stats[prop], random_state=42, n_jobs=-1)
        model.fit(X_train_stats_imputed, train_df[f'TARGET_{prop}'])

        preds_stat = model.predict(X_test_stats_imputed)
        test_preds[prop] = preds_stat
        mae_values[prop] = mean_absolute_error(test_df[f'TARGET_{prop}'], preds_stat)
        print(f"  - {prop} Model Validation RMSE: {np.sqrt(mean_squared_error(test_df[f'TARGET_{prop}'], preds_stat)):.3f} | MAE: {mae_values[prop]:.3f}")

    # Calculate combination MAEs
    mae_values['PRA'] = mean_absolute_error(test_df['TARGET_PTS'] + test_df['TARGET_REB'] + test_df['TARGET_AST'], test_preds['PTS'] + test_preds['REB'] + test_preds['AST'])
    mae_values['P+R'] = mean_absolute_error(test_df['TARGET_PTS'] + test_df['TARGET_REB'], test_preds['PTS'] + test_preds['REB'])
    mae_values['P+A'] = mean_absolute_error(test_df['TARGET_PTS'] + test_df['TARGET_AST'], test_preds['PTS'] + test_preds['AST'])
    mae_values['R+A'] = mean_absolute_error(test_df['TARGET_REB'] + test_df['TARGET_AST'], test_preds['REB'] + test_preds['AST'])

    with open(os.path.join(MODEL_DIR, "mae_values.json"), 'w') as f:
        json.dump(mae_values, f, indent=4)
    print(f"✅ MAE values saved to mae_values.json")

    print("\n--- Finalizing Production Models (Re-training on all data) ---")
    imputer_min_final = SimpleImputer(strategy='mean')
    X_min_full_imputed = imputer_min_final.fit_transform(df[min_features])

    minutes_model_final = lgb.LGBMRegressor(**best_params_min, random_state=42, n_jobs=-1)
    minutes_model_final.fit(X_min_full_imputed, df['TARGET_MIN'])

    joblib.dump(minutes_model_final, os.path.join(MODEL_DIR, "minutes_lgbm_model.pkl"))
    joblib.dump(imputer_min_final, os.path.join(MODEL_DIR, "minutes_imputer.pkl"))
    joblib.dump(min_features, os.path.join(MODEL_DIR, "minutes_feature_names.pkl"))

    df['Proj_MIN'] = minutes_model_final.predict(X_min_full_imputed)
    imputer_stats_final = SimpleImputer(strategy='mean')
    X_stats_full_imputed = imputer_stats_final.fit_transform(df[stat_features])

    joblib.dump(imputer_stats_final, os.path.join(MODEL_DIR, "stats_imputer.pkl"))
    joblib.dump(stat_features, os.path.join(MODEL_DIR, "stats_feature_names.pkl"))

    for prop in ['PTS', 'REB', 'AST', 'FG3M']:
        model_final = lgb.LGBMRegressor(**best_params_stats[prop], random_state=42, n_jobs=-1)
        model_final.fit(X_stats_full_imputed, df[f'TARGET_{prop}'])
        joblib.dump(model_final, os.path.join(MODEL_DIR, f"{prop.lower()}_lgbm_model.pkl"))
    print("✅ Final Stats models saved.")

    df['Proj_PTS'] = joblib.load(os.path.join(MODEL_DIR, "pts_lgbm_model.pkl")).predict(X_stats_full_imputed)
    df['Proj_REB'] = joblib.load(os.path.join(MODEL_DIR, "reb_lgbm_model.pkl")).predict(X_stats_full_imputed)
    df['Proj_AST'] = joblib.load(os.path.join(MODEL_DIR, "ast_lgbm_model.pkl")).predict(X_stats_full_imputed)
    classification_features = ['Proj_PTS', 'Proj_REB', 'Proj_AST']

    joblib.dump(classification_features, os.path.join(MODEL_DIR, "classification_feature_names.pkl"))

    for prop in ['DD', 'TD']:
        y_class_full = df[f'TARGET_{prop}']
        if y_class_full.sum() > 50:
            classifier_final = lgb.LGBMClassifier(objective='binary', class_weight='balanced', random_state=42, n_jobs=-1)
            classifier_final.fit(df[classification_features], y_class_full)
            joblib.dump(classifier_final, os.path.join(MODEL_DIR, f"{prop.lower()}_lgbm_model.pkl"))
    print("✅ Final Classification models saved.")

if __name__ == "__main__":
    train_and_validate_models_v8_3()
```

```python
#@title SCRIPT 3: NBA Professional Sharp & ML Engine (v127 - PINNACLE GOLD STANDARD UNABRIDGED)
# Features: 10-Game Streaks, Robust H2H, CLV/RLM/Steam on ALL Sheets, Full Guides
# NEW: Pinnacle-Exclusive Sharp Benchmark for Game Lines & Market Baselines
# NEW: No-Vig Fair Implied Probability Column Directly Next to Odds
# NEW: Dynamic Confidence Score (Empirical 72%-94.5% Scale, Uncapped from Flat 99%)
# NEW: Strict Half-Point & Whole Integer Snapping for Alternate Lines (.5 Steps)
# NEW: Proj vs Sharp Conditional Highlights (>= +0.5 for Over, <= -0.5 for Under)
# NEW: Dynamic Column Autowidth with Built-in Dropdown Filter Arrow Padding
# NEW: Automated Player Prop Grouping & Chronological Matchup Organization
# ==============================================================================

# --- Part 0: Setup & Installs ---
!pip install -q xlsxwriter nba_api scikit-learn pandas==2.2.3 joblib pytz unidecode requests lightgbm python-dateutil tqdm openpyxl scipy sportsdataverse

import os
import re
import time
import joblib
import warnings
import requests
import json
import glob
import traceback
import numpy as np
import pandas as pd
import pytz
from datetime import datetime, timedelta
from dateutil import parser
from unidecode import unidecode
from tqdm import tqdm
from decimal import Decimal
from scipy.stats import poisson, norm
from google.colab import drive
from sklearn.metrics import mean_absolute_error, mean_squared_error
from IPython.display import display, clear_output
from xlsxwriter.utility import xl_col_to_name

from nba_api.stats.static import teams, players
from sportsdataverse.nba import load_nba_schedule, most_recent_nba_season

warnings.filterwarnings("ignore")

# --- Part 1: Configuration ---
GDRIVE_BASE_PATH = "/content/drive/MyDrive/NBA/"
MODEL_DIR = os.path.join(GDRIVE_BASE_PATH, "Training Model Advanced/")
OUTPUT_DIR = os.path.join(GDRIVE_BASE_PATH, "Predictions/")
GAMELOG_LAKE_DIR = os.path.join(GDRIVE_BASE_PATH, "gamelog_lake/")
DATA_DIR = os.path.join(GDRIVE_BASE_PATH, "Historical Data/")
BOXSCORE_DIR = os.path.join(DATA_DIR, "Boxscores/")
INITIAL_LINES_PATH = os.path.join(OUTPUT_DIR, "initial_nba_lines.csv")

os.makedirs(OUTPUT_DIR, exist_ok=True); os.makedirs(GAMELOG_LAKE_DIR, exist_ok=True)

# *** 1. MARKET TOGGLES ***
ENABLE_MARKETS = {
    "Moneyline": True, "Spreads": True, "Totals": True,
    "Points": True, "Rebounds": True, "Assists": True, "Threes": True,
    "Combos": True, "DoubleDouble": True, "TripleDouble": True
}

FILTER_OUT_PLAYERS = True

_GAME_MAP = {"Moneyline": "h2h", "Spreads": "spreads", "Totals": "totals"}
_PROP_MAP = {
    "Points": {'player_points': 'PTS'}, "Rebounds": {'player_rebounds': 'REB'}, "Assists": {'player_assists': 'AST'},
    "Threes": {'player_threes': 'FG3M'},
    "Combos": {'player_points_rebounds_assists': 'PRA', 'player_points_rebounds': 'P+R', 'player_points_assists': 'P+A', 'player_rebounds_assists': 'R+A'},
    "DoubleDouble": {'player_double_double': 'DD'}, "TripleDouble": {'player_triple_double': 'TD'}
}

SIMPLE_PROP_MAP = {
    "Points": "PTS", "PTS": "PTS",
    "Rebounds": "REB", "REB": "REB",
    "Assists": "AST", "AST": "AST",
    "Threes": "FG3M", "FG3M": "FG3M", "3PT": "FG3M",
    "Combos": "PRA", "PRA": "PRA",
    "DoubleDouble": "DD", "DD": "DD",
    "TripleDouble": "TD", "TD": "TD",
    "P+R": "P+R", "P+A": "P+A", "R+A": "R+A"
}

ACTIVE_GAME_MARKETS = [v for k, v in _GAME_MAP.items() if ENABLE_MARKETS.get(k, False)]

MARKET_NAME_MAP = {}
for category, enabled in ENABLE_MARKETS.items():
    if enabled and category in _PROP_MAP:
        MARKET_NAME_MAP.update(_PROP_MAP[category])
ALL_PROP_MARKETS_LIST = list(MARKET_NAME_MAP.keys())

# *** EXACT KEYS ***
ODDS_API_KEYS = [
    "9b96859c5e0781bb4c5809215a5ed94b", "f4cfcea7cc6120a97559f69d6049596e", "7b9405e2febc561e554b7169a42cefd1", "146dff42ae19a83208b059d4855095c9",
    "f8474479351ccc685ccae9f4716c23c5", "39756c093848b6968c899a4e1ded85f9", "9705a1f6b609305047eb8ec9243cc54a", "4217574ccdd90732483d98f24772f66b",
    "66953d96fcb2ec72a47e7f39901258e7", "94304f58fb95fd1d45159b9c8d292724", "e384411e342cac5c9408804c622703f1", "dab5c3da80b0415f64d4d94890974cdf",
    "9705a1f6b609305047eb8ec9243cc54a", "3aae94b51cc3dcb8e0623bf2633a695a", "a4e4d3eef7d360055b96a1351560817f", "1bfcac154fb2b490bf55a80a18275db1",
    "7b64d450f91079f60211cb33033594b4", "ec9dc1e1aa0543c1b3fa3d2576fabffd", "6bbfe433274aef6fee5aa5543ad69055", "1f5e9cdc190854c605b533917b2c9b82",
    "858b7234b5f77661345f205b1c5ec121", "39756c093848b6968c899a4e1ded85f9", "21caebe8058c338a7ba76f6c9fb550cb", "94304f58fb95fd1d45159b9c8d292724",
    "129b9d482b270ff3809e11b0c445ef5c", "a11767c08916b29f824150bffbbf6d9c", "e7e814c95b32c4a9fb5b72acefb7389b", "3a3cd8a78555c46afa842aa8bdf1455e",
    "3b9adf0f0e27964f58de1f9b4af3adfc", "f098da95884f42852df5ff12f5704741", "768fb25c7f33328bc53b61b81e607bd9"
]

# PINNACLE SOLE SHARP BENCHMARK
SHARP_BOOK = "pinnacle"
RETAIL_BOOKS = ["draftkings", "fanduel"]
ALL_BOOKS_FOR_API = list(set([SHARP_BOOK] + RETAIL_BOOKS))
ROLLING_WINDOWS = [5, 10, 20]
CONFIG = {
    "GAMES_PLAYED_THRESHOLD_ACCLIMATED": 15,
    "GAMES_PLAYED_THRESHOLD_LOW_VOLUME": 20,
    "STEAM_THRESHOLD": 1.5,
    "RLM_CONFIDENCE": 0.65
}

# --- Part 2: Helper Functions ---

def run_accuracy_audit():
    print("\n" + "="*60)
    print("🧠 PHASE 0: PRE-FLIGHT ACCURACY AUDIT (Last 1,000 Games)")
    print("="*60)
    try:
        if not os.path.exists('/content/drive'):
            drive.mount('/content/drive', force_remount=True)

        df = pd.read_csv(os.path.join(DATA_DIR, "nba_advanced_training_data_v8.csv"))
        df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])

        models = {
            'min': joblib.load(os.path.join(MODEL_DIR, "minutes_lgbm_model.pkl")),
            'pts': joblib.load(os.path.join(MODEL_DIR, "pts_lgbm_model.pkl")),
            'reb': joblib.load(os.path.join(MODEL_DIR, "reb_lgbm_model.pkl")),
            'ast': joblib.load(os.path.join(MODEL_DIR, "ast_lgbm_model.pkl")),
            'fg3m': joblib.load(os.path.join(MODEL_DIR, "fg3m_lgbm_model.pkl"))
        }

        imputers = {
            'minutes': joblib.load(os.path.join(MODEL_DIR, "minutes_imputer.pkl")),
            'stats': joblib.load(os.path.join(MODEL_DIR, "stats_imputer.pkl"))
        }

        min_feats = joblib.load(os.path.join(MODEL_DIR, "minutes_feature_names.pkl"))
        stat_feats = joblib.load(os.path.join(MODEL_DIR, "stats_feature_names.pkl"))

    except Exception as e:
        print(f"❌ CRITICAL ERROR: Could not load files for audit. Error: {e}")
        return {}

    audit_df = df.sort_values('GAME_DATE', ascending=False).head(1000).copy()

    X_min = audit_df[min_feats]
    X_min_imputed = imputers['minutes'].transform(X_min)
    audit_df['PRED_MIN'] = models['min'].predict(X_min_imputed)

    audit_df['Proj_MIN'] = audit_df['PRED_MIN']
    X_stats = audit_df[stat_feats]
    X_stats_imputed = imputers['stats'].transform(X_stats)

    audit_df['PRED_PTS'] = models['pts'].predict(X_stats_imputed)
    audit_df['PRED_REB'] = models['reb'].predict(X_stats_imputed)
    audit_df['PRED_AST'] = models['ast'].predict(X_stats_imputed)
    audit_df['PRED_FG3M'] = models['fg3m'].predict(X_stats_imputed)

    audit_df['PRED_PRA'] = audit_df['PRED_PTS'] + audit_df['PRED_REB'] + audit_df['PRED_AST']
    audit_df['ACTUAL_PRA'] = audit_df['TARGET_PTS'] + audit_df['TARGET_REB'] + audit_df['TARGET_AST']

    audit_df['PRED_P+R'] = audit_df['PRED_PTS'] + audit_df['PRED_REB']
    audit_df['ACTUAL_P+R'] = audit_df['TARGET_PTS'] + audit_df['TARGET_REB']

    audit_df['PRED_P+A'] = audit_df['PRED_PTS'] + audit_df['PRED_AST']
    audit_df['ACTUAL_P+A'] = audit_df['TARGET_PTS'] + audit_df['TARGET_AST']

    audit_df['PRED_R+A'] = audit_df['PRED_REB'] + audit_df['PRED_AST']
    audit_df['ACTUAL_R+A'] = audit_df['TARGET_REB'] + audit_df['TARGET_AST']

    results = []
    comparison_map = {
        'PTS': ('TARGET_PTS', 'PRED_PTS'),
        'REB': ('TARGET_REB', 'PRED_REB'),
        'AST': ('TARGET_AST', 'PRED_AST'),
        '3PT': ('TARGET_FG3M', 'PRED_FG3M'),
        'PRA': ('ACTUAL_PRA', 'PRED_PRA'),
        'P+R': ('ACTUAL_P+R', 'PRED_P+R'),
        'P+A': ('ACTUAL_P+A', 'PRED_P+A'),
        'R+A': ('ACTUAL_R+A', 'PRED_R+A')
    }

    audit_hurdles = {}
    for label, (actual_col, pred_col) in comparison_map.items():
        mae = mean_absolute_error(audit_df[actual_col], audit_df[pred_col])
        rmse = np.sqrt(mean_squared_error(audit_df[actual_col], audit_df[pred_col]))
        avg_actual = audit_df[actual_col].mean()
        rel_error = (mae / avg_actual) * 100 if avg_actual > 0 else 0

        audit_hurdles[label] = rel_error

        results.append({
            'Prop Type': label,
            'Avg Error (MAE)': round(mae, 2),
            'Volatility (RMSE)': round(rmse, 2),
            'Error %': f"{round(rel_error, 1)}%"
        })

    report = pd.DataFrame(results).sort_values('Avg Error (MAE)')
    print(report.to_string(index=False))
    print("="*60 + "\n")
    return audit_hurdles

def load_master_boxscores():
    files = glob.glob(os.path.join(BOXSCORE_DIR, "boxscores_regular_season_and_playoffs_*.parquet"))
    if not files:
        files = glob.glob(os.path.join(DATA_DIR, "*.parquet"))
    if not files:
        return pd.DataFrame()
    latest_file = sorted(files)[-1]
    print(f"📦 Loading historical boxscores from: {os.path.basename(latest_file)}")
    df = pd.read_parquet(latest_file)

    rename_cols = {
        'personId': 'PLAYER_ID', 'player_id': 'PLAYER_ID',
        'personName': 'PLAYER_NAME', 'player_name': 'PLAYER_NAME',
        'game_date': 'GAME_DATE', 'gameId': 'GAME_ID',
        'teamId': 'TEAM_ID', 'team_id': 'TEAM_ID',
        'teamTricode': 'TEAM_ABBREVIATION', 'team_abbreviation': 'TEAM_ABBREVIATION',
        'minutes': 'MIN', 'points': 'PTS', 'reboundsTotal': 'REB', 'assists': 'AST',
        'threePointersMade': 'FG3M', 'threePointersAttempted': 'FG3A',
        'fieldGoalsMade': 'FGM', 'fieldGoalsAttempted': 'FGA',
        'freeThrowsMade': 'FTM', 'freeThrowsAttempted': 'FTA',
        'steals': 'STL', 'blocks': 'BLK', 'turnovers': 'TOV'
    }
    df.rename(columns={k: v for k, v in rename_cols.items() if k in df.columns}, inplace=True)
    df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])
    return df

def get_defense_ranks_from_boxscores(master_df):
    print("   🛡️ Calculating Opponent Defense Rankings from Historical Boxscores...")
    if master_df.empty: return {}, {}
    try:
        recent = master_df.sort_values('GAME_DATE').tail(15000)
        team_stats = recent.groupby('TEAM_ABBREVIATION').agg({
            'PTS': 'mean', 'REB': 'mean', 'AST': 'mean', 'FG3M': 'mean'
        }).reset_index()

        team_stats['CALC_RANK_PTS'] = team_stats['PTS'].rank(ascending=False)
        team_stats['CALC_RANK_REB'] = team_stats['REB'].rank(ascending=False)
        team_stats['CALC_RANK_AST'] = team_stats['AST'].rank(ascending=False)
        team_stats['CALC_RANK_FG3M'] = team_stats['FG3M'].rank(ascending=False)

        team_stats['PRA'] = team_stats['PTS'] + team_stats['REB'] + team_stats['AST']
        team_stats['P+R'] = team_stats['PTS'] + team_stats['REB']
        team_stats['P+A'] = team_stats['PTS'] + team_stats['AST']
        team_stats['R+A'] = team_stats['REB'] + team_stats['AST']

        team_stats['CALC_RANK_PRA'] = team_stats['PRA'].rank(ascending=False)
        team_stats['CALC_RANK_P+R'] = team_stats['P+R'].rank(ascending=False)
        team_stats['CALC_RANK_P+A'] = team_stats['P+A'].rank(ascending=False)
        team_stats['CALC_RANK_R+A'] = team_stats['R+A'].rank(ascending=False)

        ranks, raw_stats = {}, {}
        abbr_to_id = {t['abbreviation']: str(t['id']) for t in teams.get_teams()}

        for _, row in team_stats.iterrows():
            tid = abbr_to_id.get(row['TEAM_ABBREVIATION'], row['TEAM_ABBREVIATION'])
            ranks[tid] = {
                'PTS': int(row['CALC_RANK_PTS']), 'REB': int(row['CALC_RANK_REB']),
                'AST': int(row['CALC_RANK_AST']), 'FG3M': int(row['CALC_RANK_FG3M']),
                'PRA': int(row['CALC_RANK_PRA']), 'P+R': int(row['CALC_RANK_P+R']),
                'P+A': int(row['CALC_RANK_P+A']), 'R+A': int(row['CALC_RANK_R+A'])
            }
            raw_stats[tid] = {
                'PTS': float(row['PTS']), 'REB': float(row['REB']),
                'AST': float(row['AST']), 'FG3M': float(row['FG3M'])
            }
        return ranks, raw_stats
    except Exception as e:
        print(f"   ⚠️ Could not compute local defense ranks: {e}")
        return {}, {}

def get_team_metrics_from_boxscores(master_df):
    print("   📊 Calculating Team Metrics from Historical Boxscores...")
    if master_df.empty: return {}
    metrics_dict = {}
    try:
        abbr_to_id = {t['abbreviation']: t['id'] for t in teams.get_teams()}
        recent = master_df.sort_values('GAME_DATE').tail(40000)
        game_totals = recent.groupby(['GAME_ID', 'TEAM_ABBREVIATION']).agg({
            'PTS': 'sum', 'FGA': 'sum', 'FTA': 'sum', 'FGM': 'sum', 'FG3M': 'sum'
        }).reset_index()
        grouped = game_totals.groupby('TEAM_ABBREVIATION').agg({
            'PTS': 'mean', 'FGA': 'mean', 'FTA': 'mean', 'FGM': 'mean', 'FG3M': 'mean'
        }).reset_index()

        for _, r in grouped.iterrows():
            tid = abbr_to_id.get(r['TEAM_ABBREVIATION'])
            if tid:
                efg = (r['FGM'] + 0.5 * r['FG3M']) / r['FGA'] if r['FGA'] > 0 else 0.52
                metrics_dict[tid] = {
                    'E_OFF_RATING': r['PTS'], 'E_DEF_RATING': 115.0,
                    'E_NET_RATING': 0.0, 'E_PACE': 100.0,
                    'E_EFG_PCT': efg, 'E_OREB_PCT': 0.28,
                    'GP': 82, 'W': 41, 'L': 41
                }
        return metrics_dict
    except Exception as e:
        print(f"   ⚠️ Could not compute local team metrics: {e}")
        return {}

def assign_live_player_trait_runtime(player_id, player_name, logs, trait_map_df):
    if not trait_map_df.empty:
        match = trait_map_df[trait_map_df['PLAYER_ID'] == int(player_id)]
        if not match.empty:
            return int(match.iloc[-1]['TRAIT'])

    if logs.empty: return 4
    try:
        recent = logs.head(15).copy()
        recent['MIN_dec'] = recent['MIN'].apply(convert_min_to_decimal)
        mins = max(1.0, recent['MIN_dec'].mean())
        pts_pm = recent['PTS'].mean() / mins
        reb_pm = recent['REB'].mean() / mins
        ast_pm = recent['AST'].mean() / mins
        fg3m_pm = recent['FG3M'].mean() / mins
        vector = np.array([pts_pm, reb_pm, ast_pm, fg3m_pm])

        cluster_norms = {
            0: np.array([0.3, 0.2, 0.05, 0.05]),
            1: np.array([0.6, 0.15, 0.2, 0.08]),
            2: np.array([0.4, 0.12, 0.12, 0.06]),
            3: np.array([0.4, 0.08, 0.25, 0.05]),
            4: np.array([0.25, 0.1, 0.05, 0.08]),
            5: np.array([0.5, 0.3, 0.08, 0.01]),
            6: np.array([0.15, 0.05, 0.08, 0.04]),
            7: np.array([0.5, 0.1, 0.12, 0.10]),
            8: np.array([0.3, 0.06, 0.1, 0.08]),
            9: np.array([0.1, 0.04, 0.04, 0.02]),
        }

        best_trait = 4
        min_dist = float('inf')
        for trait_id, center in cluster_norms.items():
            dist = np.linalg.norm(vector - center)
            if dist < min_dist:
                min_dist = dist
                best_trait = trait_id
        return best_trait
    except:
        return 4

def get_coach_games_played(team_abbr, coach_df, logs, season_start_year):
    if coach_df.empty or logs.empty: return 0
    match = coach_df[(coach_df['team'] == team_abbr) & (coach_df['from_year'] <= season_start_year) & (coach_df['to_year'] >= season_start_year)]
    if match.empty: return 0
    season_cutoff = datetime(season_start_year, 9, 1)
    try:
        logs['GAME_DATE_DT'] = pd.to_datetime(logs['GAME_DATE'])
        cur_season_logs = logs[logs['GAME_DATE_DT'] >= season_cutoff]
        return len(cur_season_logs)
    except: return 0

def normalize_player_name(name):
    if name is None: return None
    try: name = str(name)
    except: return None
    name = unidecode(name.lower()); name = re.sub(r'\s+(jr|sr|ii|iii|iv)\.?$', '', name); name = re.sub(r'[^\w\s]', '', name); return re.sub(r'\s+', ' ', name).strip()

def get_team_abbr_map(): return {team['full_name']: team['abbreviation'] for team in teams.get_teams()}

def convert_min_to_decimal(min_str):
    if isinstance(min_str, str) and ':' in min_str:
        try: minutes, seconds = map(int, min_str.split(':')); return minutes + seconds / 60.0
        except: return np.nan
    return pd.to_numeric(min_str, errors='coerce')

def derive_position(row):
    pts = row.get('Pred_PTS', 0); reb = row.get('Pred_REB', 0); ast = row.get('Pred_AST', 0)
    if reb >= 8.0: return 'C'
    if reb >= 6.0 and ast < 2.5: return 'C/F'
    if ast >= 6.0: return 'PG'
    if ast >= 4.5 and reb < 4.0: return 'PG'
    if reb >= 5.0 and pts >= 12.0: return 'F'
    if pts >= 10.0 and reb < 4.5: return 'G'
    return 'G/F'

def get_player_experience(player_id, master_df):
    try:
        p_logs = master_df[master_df['PLAYER_ID'].astype(str) == str(player_id)]
        if p_logs.empty: return "Rookie"
        seasons_played = p_logs['season_year'].nunique() if 'season_year' in p_logs.columns else len(p_logs) // 50
        teams_played = p_logs['TEAM_ABBREVIATION'].nunique()
        if seasons_played <= 1: return "Rookie"
        if teams_played > 1 and len(p_logs.tail(15)['TEAM_ABBREVIATION'].unique()) > 1: return "Traded Veteran"
        if len(p_logs) < CONFIG["GAMES_PLAYED_THRESHOLD_LOW_VOLUME"]: return "Low Volume"
        return "Veteran"
    except Exception:
        return "Veteran"

def fetch_espn_nba_injuries():
    print("\nFetching live injury report from ESPN API (NBA)...")
    try:
        teams_url = "https://site.api.espn.com/apis/site/v2/sports/basketball/nba/teams"
        teams_data = requests.get(teams_url, timeout=10).json()
        team_list = teams_data['sports'][0]['leagues'][0]['teams']
        all_injuries =[]
        for team in tqdm(team_list, desc="Fetching Team Rosters for Injuries"):
            team_id = team['team']['id']
            roster_url = f"https://site.api.espn.com/apis/site/v2/sports/basketball/nba/teams/{team_id}?enable=roster"
            try: roster_data = requests.get(roster_url, timeout=10).json()
            except Exception: continue
            for player in roster_data.get('team', {}).get('athletes',[]):
                if 'injuries' in player and player['injuries']:
                    injury_obj = player['injuries'][0]
                    status = injury_obj.get('status', 'Unknown')
                    all_injuries.append({'Player_ESPN': player.get('displayName'), 'Status': status, 'Injury Details': str(injury_obj.get('date', '')) + " " + str(injury_obj.get('type', {}).get('description', ''))})

        if not all_injuries: print("   ℹ No reported injuries found."); return pd.DataFrame()
        injury_df = pd.DataFrame(all_injuries)
        injury_df['match_key'] = injury_df['Player_ESPN'].apply(normalize_player_name)
        print(f"   ✅ Successfully fetched {len(injury_df)} injury entries.")
        return injury_df
    except Exception as e: print(f"   ⚠️ Could not fetch ESPN injuries: {e}"); return pd.DataFrame()

def calculate_prop_streak(player_id, prop_map, line, master_df):
    try:
        gamelog_file = os.path.join(GAMELOG_LAKE_DIR, f"{player_id}.csv")
        if os.path.exists(gamelog_file):
            df = pd.read_csv(gamelog_file)
        elif not master_df.empty:
            df = master_df[master_df['PLAYER_ID'].astype(str) == str(player_id)].copy()
        else:
            return 0, 0, 0, 0

        if df.empty: return 0, 0, 0, 0
        df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])
        df = df.sort_values(by='GAME_DATE', ascending=False).reset_index(drop=True)
        col_name = SIMPLE_PROP_MAP.get(prop_map, prop_map)

        if col_name == 'PTS': val_series = df['PTS']
        elif col_name == 'REB': val_series = df['REB']
        elif col_name == 'AST': val_series = df['AST']
        elif col_name == 'FG3M': val_series = df['FG3M']
        elif col_name == 'PRA': val_series = df['PTS'] + df['REB'] + df['AST']
        elif col_name == 'P+R': val_series = df['PTS'] + df['REB']
        elif col_name == 'P+A': val_series = df['PTS'] + df['AST']
        elif col_name == 'R+A': val_series = df['REB'] + df['AST']
        elif col_name == 'DD': val_series = ((df['PTS']>=10).astype(int) + (df['REB']>=10).astype(int) + (df['AST']>=10).astype(int)) >= 2
        else: val_series = df.get(col_name)

        if val_series is None: return 0, 0, 0, 0

        l10_slice = val_series.head(10)

        streak_over = 0
        for val in l10_slice:
            if col_name == 'DD':
                if val: streak_over += 1
                else: break
            else:
                if val > line: streak_over += 1
                else: break

        streak_under = 0
        for val in l10_slice:
            if col_name == 'DD':
                if not val: streak_under += 1
                else: break
            else:
                if val < line: streak_under += 1
                else: break

        hits_over_l10 = l10_slice.sum() if col_name == 'DD' else (l10_slice > line).sum()
        hits_under_l10 = (10 - l10_slice.sum()) if col_name == 'DD' else (l10_slice < line).sum()

        return streak_over, streak_under, hits_over_l10, hits_under_l10
    except Exception: return 0, 0, 0, 0

def calculate_head_to_head(player_id, prop_map, line, opponent_abbr, season_year, master_df):
    try:
        gamelog_file = os.path.join(GAMELOG_LAKE_DIR, f"{player_id}.csv")
        if os.path.exists(gamelog_file):
            df = pd.read_csv(gamelog_file)
        elif not master_df.empty:
            df = master_df[master_df['PLAYER_ID'].astype(str) == str(player_id)].copy()
        else:
            return "0/0"

        if df.empty: return "0/0"

        df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])
        cutoff = datetime.now() - timedelta(days=365*2)
        df = df[df['GAME_DATE'] >= cutoff]

        col_name = SIMPLE_PROP_MAP.get(prop_map, prop_map)
        target_opp = str(opponent_abbr).upper()

        abbr_map = {'PHX': 'PHO', 'PHO': 'PHX', 'GSW': 'GS', 'GS': 'GSW', 'NOP': 'NO', 'NO': 'NOP', 'SAS': 'SA', 'SA': 'SAS', 'NYK': 'NY', 'NY': 'NYK', 'UTA': 'UT', 'UT': 'UTA', 'WAS': 'WSH', 'WSH': 'WAS'}
        alt_opp = abbr_map.get(target_opp, target_opp)

        def is_matchup(m_str):
            if pd.isna(m_str): return False
            m_str = str(m_str).replace('.', '').upper()
            tokens = m_str.split()
            actual_opp = tokens[-1] if tokens else ""
            return actual_opp == target_opp or actual_opp == alt_opp

        if 'MATCHUP' in df.columns:
            df_vs = df[df['MATCHUP'].apply(is_matchup)]
        elif 'matchup' in df.columns:
            df_vs = df[df['matchup'].apply(is_matchup)]
        else:
            return "0/0"

        if df_vs.empty: return "0/0"

        if col_name == 'PRA': val_series = df_vs['PTS'] + df_vs['REB'] + df_vs['AST']
        elif col_name == 'P+R': val_series = df_vs['PTS'] + df_vs['REB']
        elif col_name == 'P+A': val_series = df_vs['PTS'] + df_vs['AST']
        elif col_name == 'R+A': val_series = df_vs['REB'] + df_vs['AST']
        else: val_series = df_vs[col_name]

        hits = (val_series > line).sum()
        return f"{int(hits)}/{len(df_vs)}"
    except Exception: return "0/0"

def get_live_schedule_from_sportsdataverse(iso_date):
    schedule = {}
    print(f"--- Fetching Schedule for {iso_date} via SportsDataverse ---")
    try:
        current_year = most_recent_nba_season()
        try:
            df = load_nba_schedule(seasons=[current_year], return_as_pandas=True)
        except Exception:
            df = load_nba_schedule(seasons=[current_year - 1], return_as_pandas=True)

        if df is None or df.empty:
            return get_live_schedule_from_espn(iso_date)

        date_col = 'game_date' if 'game_date' in df.columns else ('date' if 'date' in df.columns else None)
        if not date_col:
            return get_live_schedule_from_espn(iso_date)

        df[date_col] = df[date_col].astype(str)
        todays_games = df[df[date_col].str.contains(iso_date, na=False)]

        if todays_games.empty:
            return get_live_schedule_from_espn(iso_date)

        nba_teams = teams.get_teams()
        abbr_to_id = {t['abbreviation']: str(t['id']) for t in nba_teams}

        valid_count = 0
        for _, game in todays_games.iterrows():
            try:
                h_abbr = game.get('home_team_abbreviation', game.get('home_abbreviation', ''))
                a_abbr = game.get('away_team_abbreviation', game.get('away_abbreviation', ''))

                mapping = {'PHO': 'PHX', 'GS': 'GSW', 'NO': 'NOP', 'WSH': 'WAS', 'UTAH': 'UTA', 'SA': 'SAS', 'NY': 'NYK'}
                h_abbr = mapping.get(h_abbr, h_abbr)
                a_abbr = mapping.get(a_abbr, a_abbr)

                h_id = abbr_to_id.get(h_abbr)
                v_id = abbr_to_id.get(a_abbr)

                if h_id and v_id:
                    schedule[str(h_id)] = str(v_id)
                    schedule[str(v_id)] = str(h_id)
                    valid_count += 1
            except Exception:
                continue

        print(f"✅ Found {valid_count} games via SportsDataverse schedule.")
        return schedule
    except Exception as e:
        print(f"⚠️ SportsDataverse schedule lookup failed, falling back to ESPN: {e}")
        return get_live_schedule_from_espn(iso_date)

def get_live_schedule_from_espn(iso_date):
    schedule = {}
    print(f"--- Fetching Schedule for {iso_date} via ESPN Live API ---")
    try:
        date_clean = iso_date.replace('-', '')
        url = f"https://site.api.espn.com/apis/site/v2/sports/basketball/nba/scoreboard?dates={date_clean}"
        res = requests.get(url, timeout=10).json()
        events = res.get('events', [])

        if not events:
            print(f"ℹ No games scheduled on ESPN for {iso_date}")
            return {}

        nba_teams = teams.get_teams()
        abbr_to_id = {t['abbreviation']: str(t['id']) for t in nba_teams}

        valid_count = 0
        for ev in events:
            try:
                comps = ev.get('competitions', [{}])[0]
                competitors = comps.get('competitors', [])
                if len(competitors) == 2:
                    c_home = competitors[0] if competitors[0].get('homeAway') == 'home' else competitors[1]
                    c_away = competitors[1] if competitors[0].get('homeAway') == 'home' else competitors[0]

                    h_abbr = c_home['team']['abbreviation']
                    a_abbr = c_away['team']['abbreviation']

                    mapping = {'PHO': 'PHX', 'GS': 'GSW', 'NO': 'NOP', 'WSH': 'WAS', 'UTAH': 'UTA', 'SA': 'SAS', 'NY': 'NYK'}
                    h_abbr = mapping.get(h_abbr, h_abbr)
                    a_abbr = mapping.get(a_abbr, a_abbr)

                    h_id = abbr_to_id.get(h_abbr)
                    v_id = abbr_to_id.get(a_abbr)

                    if h_id and v_id:
                        schedule[str(h_id)] = str(v_id)
                        schedule[str(v_id)] = str(h_id)
                        valid_count += 1
            except Exception:
                continue

        print(f"✅ Found {valid_count} games on the live schedule.")
        return schedule
    except Exception as e:
        print(f"❌ Error fetching ESPN schedule: {e}")
        return {}

def get_player_gamelogs_from_master(player_id, master_df):
    gamelog_file = os.path.join(GAMELOG_LAKE_DIR, f"{player_id}.csv")
    if os.path.exists(gamelog_file):
        try:
            return pd.read_csv(gamelog_file)
        except Exception:
            pass

    if not master_df.empty:
        p_df = master_df[master_df['PLAYER_ID'].astype(str) == str(player_id)].copy()
        if not p_df.empty:
            p_df = p_df.sort_values(by="GAME_DATE", ascending=False)
            p_df.to_csv(gamelog_file, index=False)
            return p_df

    return pd.DataFrame()

def calculate_live_features(logs):
    if logs.empty or len(logs) < 5: return {}
    logs_for_calc = logs.copy(); features = {}
    logs_for_calc['MIN'] = logs_for_calc['MIN'].apply(convert_min_to_decimal)
    base_stats =['PTS', 'REB', 'AST', 'FGA', 'FG3A', 'FG3M', 'FTA', 'TOV', 'FGM']
    for col in base_stats:
        if col in logs_for_calc.columns: logs_for_calc[col] = pd.to_numeric(logs_for_calc[col], errors='coerce')
        else: logs_for_calc[col] = np.nan
    logs_for_calc.dropna(subset=['MIN'], inplace=True);
    if logs_for_calc.empty: return {}
    avg_minutes = logs_for_calc['MIN'].mean()
    weights = np.ones(len(logs_for_calc)); weights[logs_for_calc['MIN'] < 0.7 * avg_minutes] = 0.5; weights[logs_for_calc['MIN'] > 1.15 * avg_minutes] = 1.2
    logs_for_calc['WEIGHT'] = weights
    logs_safe = logs_for_calc[logs_for_calc['MIN'] > 0].copy()

    logs_safe['EFG_PCT_RAW'] = (logs_safe['FGM'] + 0.5 * logs_safe['FG3M']) / logs_safe['FGA'].replace(0, np.nan)
    logs_safe['TS_PCT_RAW'] = logs_safe['PTS'] / (2 * (logs_safe['FGA'] + 0.44 * logs_safe['FTA'])).replace(0, np.nan)
    logs_safe[['EFG_PCT_RAW', 'TS_PCT_RAW']] = logs_safe[['EFG_PCT_RAW', 'TS_PCT_RAW']].fillna(0)

    features['EFG_PCT_PREV'] = logs_safe['EFG_PCT_RAW'].iloc[0] if not logs_safe.empty else 0
    features['TS_PCT_PREV'] = logs_safe['TS_PCT_RAW'].iloc[0] if not logs_safe.empty else 0

    stats_for_ratios =['PTS', 'REB', 'AST', 'FGA', 'FG3A', 'FTA', 'TOV', 'FG3M']
    for stat in stats_for_ratios:
        if stat in logs_safe.columns: logs_safe[f'{stat}_PM'] = logs_safe[stat] / logs_safe['MIN']
        else: logs_safe[f'{stat}_PM'] = 0

    stats_to_roll =['PTS', 'REB', 'AST', 'MIN', 'FG3M', 'FG3A', 'EFG_PCT_RAW', 'TS_PCT_RAW'] +[f'{s}_PM' for s in stats_for_ratios]
    for stat in stats_to_roll:
        if stat not in logs_safe.columns: continue
        for w in ROLLING_WINDOWS:
            recent_logs = logs_safe.head(w)
            features[f'W_ROLL_AVG_{stat}_{w}'] = np.average(recent_logs[stat].fillna(0), weights=recent_logs['WEIGHT'].head(len(recent_logs))) if not recent_logs.empty else np.nan
        features[f'TREND_{stat}'] = features.get(f'W_ROLL_AVG_{stat}_{ROLLING_WINDOWS[0]}') - features.get(f'W_ROLL_AVG_{stat}_{ROLLING_WINDOWS[-1]}')

    for stat in['PTS', 'REB', 'AST', 'FG3M']:
        if stat in logs_safe.columns:
            last_20 = logs_safe.head(20)
            if not last_20.empty:
                std_dev = last_20[stat].std(); mean_val = last_20[stat].mean()
                cv = (std_dev / mean_val) if mean_val > 0.5 else 9.9
                features[f'CV_{stat}'] = cv
            else: features[f'CV_{stat}'] = 9.9
    features['DAYS_OF_REST'] = (datetime.utcnow().replace(tzinfo=pytz.utc) - pd.to_datetime(logs_for_calc['GAME_DATE'].iloc[0]).tz_localize('UTC')).days
    for stat in stats_for_ratios: features[f'{stat}_PM'] = logs_safe.head(20)[f'{stat}_PM'].mean() if f'{stat}_PM' in logs_safe.columns else 0
    return features

def apply_live_usage_redistribution(projection_rows, scratched_players, all_rosters):
    if not scratched_players or not projection_rows:
        return projection_rows

    proj_df = pd.DataFrame(projection_rows)
    scratched_rows = all_rosters[all_rosters['PLAYER_NAME'].apply(normalize_player_name).isin(scratched_players)]
    teams_with_injuries = scratched_rows['TEAM_ABBREVIATION'].unique()

    updated_rows = []
    for _, row in proj_df.iterrows():
        team = row['Team']
        if team in teams_with_injuries:
            team_scratches = scratched_rows[scratched_rows['TEAM_ABBREVIATION'] == team]
            num_injuries = len(team_scratches)

            usage_boost = 1.0 + (num_injuries * 0.035)
            min_boost = 1.0 + (num_injuries * 0.02)

            row['Pred_MIN'] = min(42.0, row['Pred_MIN'] * min_boost)
            row['Pred_PTS'] *= usage_boost
            row['Pred_AST'] *= usage_boost
            row['Pred_FGA_PM'] *= usage_boost
            row['Pred_PRA'] = row['Pred_PTS'] + row['Pred_REB'] + row['Pred_AST']
            row['Pred_P+R'] = row['Pred_PTS'] + row['Pred_REB']
            row['Pred_P+A'] = row['Pred_PTS'] + row['Pred_AST']
            row['Is_Star'] = (row['Pred_PTS'] >= 20) or (row['Pred_MIN'] >= 34)

        updated_rows.append(row)
    return updated_rows

def filter_sgp_correlations(sgp_df):
    if sgp_df.empty:
        return sgp_df

    filtered_rows = []
    games = sgp_df['Game'].unique()

    for game in games:
        game_bets = sgp_df[sgp_df['Game'] == game].copy()
        overs = game_bets[game_bets['Bet'] == 'Over']
        unders = game_bets[game_bets['Bet'] == 'Under']

        if not overs.empty:
            filtered_rows.append(overs.sort_values('Score', ascending=False).head(4))
        if not unders.empty:
            filtered_rows.append(unders.sort_values('Score', ascending=False).head(2))

    if filtered_rows:
        return pd.concat(filtered_rows, ignore_index=True)
    return sgp_df

def calculate_ou_probability(row):
    try:
        projection = float(row.get('Projection', 0))
        line = float(row.get('Line', 0))
        prop = row.get('Prop')

        if prop in['DD', 'TD']:
            prob_over = projection / 100.0
            prob_under = 1.0 - prob_over
            return prob_over, prob_under

        if pd.isna(projection) or pd.isna(line) or projection <= 0: return 0.5, 0.5
        prob_over = 1 - poisson.cdf(k=line, mu=projection)
        prob_under = poisson.cdf(k=line - 1, mu=projection)
        return prob_over, prob_under
    except: return 0.5, 0.5

def calculate_no_vig_prob(row):
    try:
        over_odds = float(row.get('Over_Odds', 0))
        under_odds = float(row.get('Under_Odds', 0))
        if over_odds <= 0 or under_odds <= 0 or pd.isna(over_odds) or pd.isna(under_odds):
            return ""

        over_implied = 1 / over_odds
        under_implied = 1 / under_odds
        total_implied = over_implied + under_implied

        if row.get('Bet') == 'Over':
            fair_prob = (over_implied / total_implied) * 100
        elif row.get('Bet') == 'Under':
            fair_prob = (under_implied / total_implied) * 100
        else:
            return ""

        return f"{fair_prob:.1f}%"
    except:
        return ""

def derive_shot_quality_signal(features):
    try:
        recent_efg = features.get('W_ROLL_AVG_EFG_PCT_RAW_5', 0)
        baseline_efg = features.get('W_ROLL_AVG_EFG_PCT_RAW_20', 0)
        fga = features.get('W_ROLL_AVG_FGA_5', 0)
        fg3a = features.get('W_ROLL_AVG_FG3A_5', 0)

        if baseline_efg <= 0: return "-"

        diff = recent_efg - baseline_efg
        if recent_efg > 0.62 and diff > 0.10:
            return "⚠️ Luck Risk"
        elif diff < -0.08 and fga >= 10:
            return "🔥 Due"
        elif fg3a >= 5 and recent_efg >= 0.54:
            return "🎯 High SQ"
        return "-"
    except:
        return "-"

def derive_advanced_matchup(row, def_ranks):
    try:
        pos = str(row.get('Pos', ''))
        prop = str(row.get('Prop', ''))
        bet = str(row.get('Bet', ''))
        oid = str(row.get('OpponentID', ''))

        rank = def_ranks.get(oid, {}).get(SIMPLE_PROP_MAP.get(prop, 'PTS'), 15)

        is_handler = 'G' in pos or 'PG' in pos
        is_big = 'C' in pos or 'F' in pos

        if bet == 'Over':
            if is_handler and prop in ['PTS', 'AST', 'PRA', 'P+A'] and rank >= 22:
                return "🔥 Elite P&R"
            if is_big and prop in ['PTS', 'REB', 'PRA', 'P+R'] and rank >= 22:
                return "🔥 Elite Rim"
            if prop in ['3PT', 'FG3M'] and rank >= 22:
                return "🔥 Elite Perim"
            if rank <= 5:
                return "⚠️ Trap Matchup"
        elif bet == 'Under':
            if is_handler and prop in ['PTS', 'AST', 'PRA'] and rank <= 5:
                return "🔒 Elite Under (P&R)"
            if is_big and prop in ['PTS', 'REB', 'PRA'] and rank <= 5:
                return "🔒 Elite Under (Rim)"
            if prop in ['3PT', 'FG3M'] and rank <= 5:
                return "🔒 Elite Under (Perim)"
        return "-"
    except Exception:
        return "-"

def get_previous_run_data(current_date_str, output_dir):
    try:
        dt_obj = datetime.strptime(current_date_str, '%m-%d-%Y')
        short_date_pattern = f"{dt_obj.month}-{dt_obj.day}-{str(dt_obj.year)[-2:]}"
    except:
        short_date_pattern = None

    candidate_files = []
    if os.path.exists(output_dir):
        for f in os.listdir(output_dir):
            if f.endswith("NBA Pro Dashboard.xlsx"):
                if current_date_str in f or (short_date_pattern and f.startswith(short_date_pattern)):
                    candidate_files.append(os.path.join(output_dir, f))

    if candidate_files:
        candidate_files.sort(key=os.path.getmtime, reverse=True)
        for file_path in candidate_files:
            print(f"✅ Found earlier run from TODAY: {os.path.basename(file_path)}. Calculating Pro vs Info Window movement...")
            try:
                return pd.read_excel(file_path, sheet_name='DraftKings_EV_Bets')
            except Exception:
                continue

    print(f"ℹ No earlier run found for TODAY ({current_date_str}). This is likely the 'Pro Window' (First Run).")
    return pd.DataFrame()

def apply_movement_and_ranking(curr_df, prev_df, sharp_props_map):
    if curr_df.empty: return curr_df
    if not prev_df.empty:
        curr_df['Player'] = curr_df['Player'].astype(str); prev_df['Player'] = prev_df['Player'].astype(str)
        if 'Bookmaker' in prev_df.columns:
            curr_df['Bookmaker'] = curr_df['Bookmaker'].astype(str); prev_df['Bookmaker'] = prev_df['Bookmaker'].astype(str)
            prev_clean = prev_df[['Player', 'Prop', 'Bookmaker', 'Edge (%)']].rename(columns={'Edge (%)': 'Prev_Edge'})
            merged = pd.merge(curr_df, prev_clean, on=['Player', 'Prop', 'Bookmaker'], how='left', suffixes=('', '_prev'))
        else:
            prev_clean = prev_df[['Player', 'Prop', 'Edge (%)']].rename(columns={'Edge (%)': 'Prev_Edge'})
            merged = pd.merge(curr_df, prev_clean, on=['Player', 'Prop'], how='left', suffixes=('', '_prev'))
    else: merged = curr_df.copy(); merged['Prev_Edge'] = np.nan

    def get_sharp_data(row):
        key = (row['match_key'], row['Prop'])
        sharp_lines_list = sharp_props_map.get(key,[])
        if sharp_lines_list:
            avg_sharp_line = sum(sharp_lines_list) / len(sharp_lines_list)
            diff = row['Projection'] - avg_sharp_line
            return round(avg_sharp_line, 1), round(diff, 1)
        return np.nan, np.nan
    merged[['Sharp Line', 'Proj vs Sharp']] = merged.apply(lambda r: pd.Series(get_sharp_data(r)), axis=1)

    def get_signal(row):
        curr_edge, prev_edge = row.get('Edge (%)', 0), row.get('Prev_Edge')
        if pd.isna(prev_edge): return "-"
        delta = curr_edge - prev_edge
        if delta > 3.5: return "PRICE IMPROVED"
        if delta < -3.5 and curr_edge > 5: return "SHARPS AGREE"
        return "-"
    merged['Sharp Signal'] = merged.apply(get_signal, axis=1)

    def get_consistency_badge(row):
        cv_pts = row.get('CV_PTS', 9.9); cv_reb = row.get('CV_REB', 9.9); cv_ast = row.get('CV_AST', 9.9)
        prop = row['Prop']
        cv = 9.9
        if prop == 'PTS': cv = cv_pts
        elif prop == 'REB': cv = cv_reb
        elif prop == 'AST': cv = cv_ast
        elif prop in['FG3M', '3PT']: cv = row.get('CV_FG3M', 9.9)
        elif prop == 'PRA': cv = (cv_pts + cv_reb + cv_ast) / 3
        elif prop == 'P+R': cv = (cv_pts + cv_reb) / 2
        elif prop == 'P+A': cv = (cv_pts + cv_ast) / 2
        elif prop == 'R+A': cv = (cv_reb + cv_ast) / 2
        else: return "-"
        if cv > 5.0: return "-"
        if cv < 0.25: return "🔒 High"
        elif cv < 0.45: return "✅ Med"
        else: return "⚡ Volatile"
    merged['Consistency'] = merged.apply(get_consistency_badge, axis=1)

    def get_ranking(row):
        edge, signal = row.get('Edge (%)', 0), row['Sharp Signal']
        bet_type = row.get('Bet')
        sl, line = row.get('Sharp Line'), row.get('Line')
        if not pd.isna(sl):
            if (bet_type == 'Over' and sl > line) or (bet_type == 'Under' and sl < line):
                return "💎 SHARP ALIGNMENT"
        if edge > 15.0: return "Strong Value"
        if edge > 5.0: return "Solid Play"
        return "-"
    merged['Ranking'] = merged.apply(get_ranking, axis=1)
    return merged

# --- Part 5: Main Execution Workflow ---
def run_daily_projections():
    proj_df = pd.DataFrame()

    try:
        if not os.path.exists('/content/drive'): drive.mount('/content/drive', force_remount=True)
    except Exception as e: print(f"⚠️ Drive mount check skipped: {e}")

    audit_hurdles = run_accuracy_audit()

    eastern = pytz.timezone('US/Eastern')
    SIMULATION_DATE = None

    if 'SIMULATION_DATE' in globals() and globals()['SIMULATION_DATE'] is not None:
         try: datetime.strptime(globals()['SIMULATION_DATE'], "%Y-%m-%d"); SIMULATION_DATE = globals()['SIMULATION_DATE']
         except: SIMULATION_DATE = None

    if SIMULATION_DATE:
        live_now_et = datetime.strptime(f"{SIMULATION_DATE} 12:00:00", "%Y-%m-%d %H:%M:%S").astimezone(eastern)
        print(f"\n⚠️ RUNNING IN SIMULATION MODE for date: {SIMULATION_DATE}")
        iso_today_str = SIMULATION_DATE
        file_today_str = live_now_et.strftime('%m-%d-%Y')
    else:
        live_now_et = datetime.now(pytz.utc).astimezone(eastern)
        print(f"\n📅 Running Live Mode. Current ET Time: {live_now_et.strftime('%Y-%m-%d %H:%M:%S')}")
        iso_today_str = live_now_et.strftime('%Y-%m-%d')
        file_today_str = live_now_et.strftime('%m-%d-%Y')

    run_mode_input = ''
    while run_mode_input not in['i', 'f', 't']:
        run_mode_input = input("Enter run mode ('i' for initial capture, 'f' for final report, 't' for training): ").strip().lower()

    if run_mode_input == 'i':
        run_mode = 'initial'
        print("\n🔵 INITIAL RUN DETECTED. Will save opening lines to CSV.")
    elif run_mode_input == 'f':
        run_mode = 'final'
        print("\n🔴 FINAL RUN DETECTED. Will compare vs Initial Lines for Steam/RLM/CLV.")
    elif run_mode_input == 't':
        run_mode = 'training'
        print("\n🟢 TRAINING MODE DETECTED.")

    if live_now_et.month >= 9: start_year = live_now_et.year; end_year = live_now_et.year + 1
    else: start_year = live_now_et.year - 1; end_year = live_now_et.year
    season = f"{start_year}-{str(end_year)[-2:]}"; prev_season = f"{start_year-1}-{str(start_year)[-2:]}"; season_year_for_url = str(start_year)

    print(f"Running projections for {iso_today_str} (Season: {season})")

    try:
        models = {'minutes': joblib.load(os.path.join(MODEL_DIR, "minutes_lgbm_model.pkl")), 'pts': joblib.load(os.path.join(MODEL_DIR, "pts_lgbm_model.pkl")), 'reb': joblib.load(os.path.join(MODEL_DIR, "reb_lgbm_model.pkl")), 'ast': joblib.load(os.path.join(MODEL_DIR, "ast_lgbm_model.pkl")), 'fg3m': joblib.load(os.path.join(MODEL_DIR, "fg3m_lgbm_model.pkl")), 'dd': joblib.load(os.path.join(MODEL_DIR, "dd_lgbm_model.pkl")), 'td': joblib.load(os.path.join(MODEL_DIR, "td_lgbm_model.pkl"))}
        imputers = {name: joblib.load(os.path.join(MODEL_DIR, f"{name}_imputer.pkl")) for name in['minutes', 'stats']}
        min_feature_names = joblib.load(os.path.join(MODEL_DIR, "minutes_feature_names.pkl")); stat_feature_names = joblib.load(os.path.join(MODEL_DIR, "stats_feature_names.pkl")); mae_values = json.load(open(os.path.join(MODEL_DIR, "mae_values.json")))
        print("✅ All models, imputers, and MAE values loaded successfully.")
    except FileNotFoundError as e: print(f"CRITICAL ERROR: A required model file is missing. Please run Script 2. Details: {e}"); return

    injury_df = fetch_espn_nba_injuries()
    master_df = load_master_boxscores()

    position_map = {}
    POSITIONS_FILE = os.path.join(GDRIVE_BASE_PATH, "Player Positions/2025-26 NBA Player Positions.xlsx")
    if os.path.exists(POSITIONS_FILE):
        try:
            pos_df = pd.read_excel(POSITIONS_FILE)
            if 'Player' in pos_df.columns and 'Pos' in pos_df.columns:
                pos_df['match_key'] = pos_df['Player'].apply(normalize_player_name)
                position_map = pos_df.set_index('match_key')['Pos'].to_dict()
        except Exception: pass

    scratched_players =[]
    if not injury_df.empty:
        scratched_players = injury_df[injury_df['Status'].isin(['Out', 'Injured Reserve', 'Suspended'])]['match_key'].tolist()

    if scratched_players: print(f"   ℹ Detected {len(scratched_players)} ruled-out players. Adjusting team usage automatically...")

    print("\n--- Preparing Context & Stats ---")

    if not master_df.empty:
        recent_cutoff = master_df['GAME_DATE'].max() - pd.Timedelta(days=300)
        recent_roster_df = master_df[master_df['GAME_DATE'] >= recent_cutoff]
        all_rosters = recent_roster_df.sort_values('GAME_DATE').groupby('PLAYER_ID', as_index=False).last()
    else:
        all_rosters = pd.DataFrame()

    def_ranks, def_stats_raw = get_defense_ranks_from_boxscores(master_df)
    team_metrics = get_team_metrics_from_boxscores(master_df)
    schedule = get_live_schedule_from_sportsdataverse(iso_today_str)

    if schedule:
        schedule = {str(k): str(v) for k, v in schedule.items() if k is not None and v is not None}
    if not all_rosters.empty and 'TEAM_ID' in all_rosters.columns:
        all_rosters['TEAM_ID'] = all_rosters['TEAM_ID'].astype(str).str.split('.').str[0]

    coach_path = os.path.join(DATA_DIR, "Coaches 2015 to Present.csv")
    coach_history_df = pd.read_csv(coach_path) if os.path.exists(coach_path) else pd.DataFrame()

    trait_path = os.path.join(DATA_DIR, "player_trait_map.csv")
    player_trait_map = pd.read_csv(trait_path) if os.path.exists(trait_path) else pd.DataFrame()

    props_df, live_game_lines = pd.DataFrame(), {}
    key, game_events = None,[]
    team_abbr_map = get_team_abbr_map()

    print("\n--- Running Initial Projections & Monte Carlo Sims ---")
    projection_rows, model_game_rows = [],[]

    LEAGUE_AVG_PACE = 99.0
    LEAGUE_AVG_OFF = 115.0

    if schedule and not all_rosters.empty:
        abbr_to_id = {t['abbreviation']: str(t['id']) for t in teams.get_teams()}
        id_to_abbr = {str(t['id']): t['abbreviation'] for t in teams.get_teams()}

        for team_id, opponent_id in tqdm(schedule.items(), desc="Generating Projections"):
            t_abbr = id_to_abbr.get(str(team_id), str(team_id))
            roster = all_rosters[(all_rosters['TEAM_ABBREVIATION'] == t_abbr) | (all_rosters['TEAM_ID'].astype(str) == str(team_id))]

            for _, row in roster.iterrows():
                player_id, player_name, team_abbr = row['PLAYER_ID'], row['PLAYER_NAME'], row['TEAM_ABBREVIATION']
                match_key = normalize_player_name(player_name)
                logs = get_player_gamelogs_from_master(player_id, master_df)
                features = calculate_live_features(logs)
                if not features: continue

                try: opp_id_int = int(opponent_id); opp_metrics = team_metrics.get(opp_id_int, {})
                except: opp_metrics = {}

                features['OPP_ALLOWED_PTS_AVG_30'] = opp_metrics.get('E_DEF_RATING', 115.0)
                features['OPP_ALLOWED_REB_AVG_30'] = 44.0; features['OPP_ALLOWED_AST_AVG_30'] = 25.0; features['OPP_ALLOWED_FG3M_AVG_30'] = 12.0
                features['OPP_ALLOWED_EFG_PCT_AVG_30'] = opp_metrics.get('E_EFG_PCT', 0.54)

                if str(opponent_id) in def_stats_raw:
                    features['OPP_ALLOWED_PTS_AVG_30'] = def_stats_raw[str(opponent_id)]['PTS']
                    features['OPP_ALLOWED_REB_AVG_30'] = def_stats_raw[str(opponent_id)]['REB']
                    features['OPP_ALLOWED_AST_AVG_30'] = def_stats_raw[str(opponent_id)]['AST']
                    features['OPP_ALLOWED_FG3M_AVG_30'] = def_stats_raw[str(opponent_id)]['FG3M']

                features['VEGAS_SPREAD'] = 0; features['VEGAS_TOTAL'] = 225

                is_playoff_game = 1 if live_now_et.month in [4, 5, 6] else 0
                features['IS_PLAYOFF_GAME'] = is_playoff_game

                coach_games = get_coach_games_played(team_abbr, coach_history_df, logs, start_year)
                features['COACH_GAMES_PLAYED'] = coach_games

                assigned_trait = assign_live_player_trait_runtime(player_id, player_name, logs, player_trait_map)

                pos_weight_pts = {0: 0.15, 1: 0.25, 2: 0.18, 3: 0.20, 4: 0.15, 5: 0.12, 6: 0.10, 7: 0.22, 8: 0.16, 9: 0.08}.get(assigned_trait, 0.18)
                pos_weight_reb = {0: 0.28, 1: 0.10, 2: 0.18, 3: 0.08, 4: 0.12, 5: 0.35, 6: 0.06, 7: 0.15, 8: 0.10, 9: 0.08}.get(assigned_trait, 0.15)
                pos_weight_ast = {0: 0.08, 1: 0.20, 2: 0.14, 3: 0.38, 4: 0.08, 5: 0.06, 6: 0.12, 7: 0.15, 8: 0.25, 9: 0.06}.get(assigned_trait, 0.15)
                pos_weight_3pm = {0: 0.02, 1: 0.22, 2: 0.18, 3: 0.25, 4: 0.32, 5: 0.01, 6: 0.15, 7: 0.28, 8: 0.20, 9: 0.10}.get(assigned_trait, 0.18)

                opp_pts_avg = features.get('OPP_ALLOWED_PTS_AVG_30', 115.0)
                opp_reb_avg = features.get('OPP_ALLOWED_REB_AVG_30', 44.0)
                opp_ast_avg = features.get('OPP_ALLOWED_AST_AVG_30', 25.0)
                opp_3pm_avg = features.get('OPP_ALLOWED_FG3M_AVG_30', 12.0)

                features['OPP_DvT_PTS'] = opp_pts_avg * pos_weight_pts * 5
                features['OPP_DvT_REB'] = opp_reb_avg * pos_weight_reb * 5
                features['OPP_DvT_AST'] = opp_ast_avg * pos_weight_ast * 5
                features['OPP_DvT_FG3M'] = opp_3pm_avg * pos_weight_3pm * 5

                try:
                    min_input = pd.DataFrame([features]).reindex(columns=min_feature_names, fill_value=0)
                    pred_min = models['minutes'].predict(imputers['minutes'].transform(min_input))[0]; features['Proj_MIN'] = pred_min

                    stat_input_df = pd.DataFrame([features]).reindex(columns=stat_feature_names, fill_value=0)
                    stat_input = imputers['stats'].transform(stat_input_df)

                    p_pts = models['pts'].predict(stat_input)[0]; p_reb = models['reb'].predict(stat_input)[0]
                    p_ast = models['ast'].predict(stat_input)[0]; p_3pm = models['fg3m'].predict(stat_input)[0]

                    temp_proj = {'Pred_PTS': p_pts, 'Pred_REB': p_reb, 'Pred_AST': p_ast}
                    if match_key in position_map: player_pos = position_map[match_key]
                    else: player_pos = derive_position(temp_proj)

                    opp_team_data = teams.find_team_name_by_id(int(opponent_id)) if str(opponent_id).isdigit() else None
                    opp_abbr = opp_team_data['abbreviation'] if opp_team_data else str(opponent_id)

                    exp_level = get_player_experience(player_id, master_df)
                    sq_sig = derive_shot_quality_signal(features)

                    proj = {
                        'PLAYER_ID': player_id, 'match_key': match_key, 'Player': player_name,
                        'Pos': player_pos, 'Experience': exp_level,
                        'Team': team_abbr, 'Opponent': opp_abbr,
                        'OpponentID': str(opponent_id), 'Pred_MIN': pred_min, 'Pred_FGA_PM': features.get('FGA_PM', 0),
                        'Pred_PTS': p_pts, 'Pred_REB': p_reb, 'Pred_AST': p_ast, 'Pred_FG3M': p_3pm, 'EFG_PCT_PREV': features.get('EFG_PCT_PREV', 0), 'TEAM_ID': team_id,
                        'Shot Quality': sq_sig, 'TREND_PTS': features.get('TREND_PTS', 0)
                    }

                    proj['Is_Star'] = (p_pts >= 22) or (pred_min >= 34)
                    proj['CV_PTS'] = features.get('CV_PTS', 9.9); proj['CV_REB'] = features.get('CV_REB', 9.9)
                    proj['CV_AST'] = features.get('CV_AST', 9.9); proj['CV_FG3M'] = features.get('CV_FG3M', 9.9)

                    clf_features = pd.DataFrame([{'Proj_PTS': p_pts, 'Proj_REB': p_reb, 'Proj_AST': p_ast}])
                    proj['Pred_DD'] = models['dd'].predict_proba(clf_features)[0][1] * 100; proj['Pred_TD'] = models['td'].predict_proba(clf_features)[0][1] * 100
                    proj['Pred_PRA'] = proj['Pred_PTS'] + proj['Pred_REB'] + proj['Pred_AST']; proj['Pred_P+R'] = proj['Pred_PTS'] + proj['Pred_REB']; proj['Pred_P+A'] = proj['Pred_PTS'] + proj['Pred_AST']; proj['Pred_R+A'] = proj['Pred_REB'] + proj['Pred_AST']
                    projection_rows.append(proj)
                except Exception: continue

            home_stats = team_metrics.get(int(team_id), {}) if str(team_id).isdigit() else {}
            away_stats = team_metrics.get(int(opponent_id), {}) if str(opponent_id).isdigit() else {}
            if home_stats and away_stats:
                h_off, a_off = home_stats.get('E_OFF_RATING', 110), away_stats.get('E_OFF_RATING', 110)
                h_def, a_def = home_stats.get('E_DEF_RATING', 110), away_stats.get('E_DEF_RATING', 110)
                h_pace, a_pace = home_stats.get('E_PACE', 99), away_stats.get('E_PACE', 99)
                poss = (h_pace + a_pace) / 2; HCA = 2.5

                high_intensity_venues = ['New York Knicks', 'Boston Celtics', 'Philadelphia 76ers', 'Denver Nuggets']
                home_team_info = teams.find_team_name_by_id(int(team_id)) if str(team_id).isdigit() else None
                home_team_name = home_team_info['full_name'] if home_team_info else "Home"
                if home_team_name in high_intensity_venues:
                    HCA = 4.5

                h_exp = ((h_off * a_def) / LEAGUE_AVG_OFF) * (poss / 100.0) + (HCA/2)
                a_exp = ((a_off * h_def) / LEAGUE_AVG_OFF) * (poss / 100.0) - (HCA/2)
                SIMS = 5000
                scores = np.random.multivariate_normal([h_exp, a_exp], [[169, 50],[50, 169]], size=SIMS)
                h_sc, a_sc = scores[:,0], scores[:,1]
                model_spread = -(np.mean(h_sc) - np.mean(a_sc))
                model_total = np.mean(h_sc) + np.mean(a_sc)
                model_win_prob = np.mean(h_sc > a_sc)
                model_game_rows.append({'TeamID': team_id, 'Home_Team_Name': home_team_name, 'Model Spread': round(model_spread, 1), 'Model Total': round(model_total, 1), 'Model Win %': model_win_prob, 'Feng Spread': "-"})

    if scratched_players and projection_rows:
        print("   ⚡ Applying live usage redistribution for scratched players...")
        projection_rows = apply_live_usage_redistribution(projection_rows, scratched_players, all_rosters)

    proj_df = pd.DataFrame(projection_rows)
    if proj_df.empty: print(f"\n❌ NO PROJECTIONS GENERATED."); return
    else: proj_df['match_key'] = proj_df['Player'].apply(normalize_player_name); print(f"✅ Generated {len(proj_df)} player projections.")

    # --- Market Odds Fetching (Pinnacle Benchmark Exclusive) ---
    print("\n--- Fetching Market Odds Data (Pinnacle Benchmark) ---")
    game_lines_rows =[]; sharp_props_map = {}; initial_lines_to_save =[]
    all_props_raw =[]

    key = None
    for k in ODDS_API_KEYS:
        try:
            r = requests.get(f"https://api.the-odds-api.com/v4/sports/basketball_nba/odds?apiKey={k}&regions=us&markets=h2h", timeout=10)
            if r.status_code == 200:
                remaining = int(r.headers.get('x-requests-remaining', 0))
                if remaining > 10:
                    key = k; break
        except: continue

    if key:
        try:
            url_games = f"https://api.the-odds-api.com/v4/sports/basketball_nba/odds/?apiKey={key}&regions=us,eu&markets=h2h,spreads,totals&bookmakers={','.join(ALL_BOOKS_FOR_API)}"
            games_data = requests.get(url_games, timeout=15).json()

            if isinstance(games_data, list):
                for game in games_data:
                    if parser.isoparse(game["commence_time"]).astimezone(eastern).strftime('%Y-%m-%d') != iso_today_str:
                        continue

                    home_abbr = team_abbr_map.get(game['home_team'], game['home_team'])
                    away_abbr = team_abbr_map.get(game['away_team'], game['away_team'])

                    model_data = next((item for item in model_game_rows if item["Home_Team_Name"] == game['home_team']), {})

                    spreads, totals = [], []
                    retail_spread, retail_total, retail_ml = np.nan, np.nan, np.nan

                    for book in game.get('bookmakers',[]):
                        is_pinnacle = (book['key'] == SHARP_BOOK)
                        for market in book.get('markets', []):
                            if market['key'] == 'spreads':
                                for o in market['outcomes']:
                                    if o['name'] == game['home_team']:
                                        if is_pinnacle: spreads.append(o['point'])
                                        if book['key'] in RETAIL_BOOKS: retail_spread = o['point']
                            elif market['key'] == 'totals':
                                for o in market['outcomes']:
                                    if o['name'] == 'Over':
                                        if is_pinnacle: totals.append(o['point'])
                                        if book['key'] in RETAIL_BOOKS: retail_total = o['point']
                            elif market['key'] == 'h2h':
                                for o in market['outcomes']:
                                    if o['name'] == game['home_team'] and book['key'] in RETAIL_BOOKS: retail_ml = o['price']

                    sharp_spread = spreads[0] if spreads else None
                    sharp_total = totals[0] if totals else None

                    mvmt, clv_s, clv_t = "-", "-", "-"
                    if run_mode == 'final' and os.path.exists(INITIAL_LINES_PATH):
                        try:
                            init_df = pd.read_csv(INITIAL_LINES_PATH); matched = init_df[init_df['GameID'] == game['id']]
                            if not matched.empty:
                                init_val = matched.iloc[0]['Initial_Sharp_Spread']
                                if sharp_spread is not None:
                                    clv_s = round(init_val - sharp_spread, 1)
                                    if abs(clv_s) >= 1.5: mvmt = "🔥 Steam Chase"
                                else:
                                    clv_s = 0.0
                        except: pass

                    game_lines_rows.append({
                        'Game': f"{away_abbr} @ {home_abbr}", 'Time': parser.isoparse(game["commence_time"]).astimezone(eastern).strftime('%-I:%M %p'),
                        'Retail Spread': retail_spread, 'Sharp Spread': round(sharp_spread, 1) if sharp_spread else "-",
                        'Model Spread': model_data.get('Model Spread', '-'), 'Retail Total': retail_total,
                        'Sharp Total': round(sharp_total, 1) if sharp_total else "-", 'Model Total': model_data.get('Model Total', '-'),
                        'Retail ML': retail_ml, 'Model Win %': f"{model_data.get('Model Win %', 0)*100:.1f}%" if 'Model Win %' in model_data else "-",
                        'Home_Team_Full': game['home_team'], 'Sort_Time': game['commence_time'], 'Game Move': mvmt, 'Game CLV': clv_s
                    })

                    if run_mode == 'initial':
                        initial_lines_to_save.append({
                            'GameID': game['id'], 'Home': home_abbr, 'Away': away_abbr,
                            'Initial_Sharp_Spread': sharp_spread, 'Initial_Sharp_Total': sharp_total
                        })

                    url_props = f"https://api.the-odds-api.com/v4/sports/basketball_nba/events/{game['id']}/odds?apiKey={key}&regions=us,eu&markets={','.join(ALL_PROP_MARKETS_LIST)}&bookmakers={','.join(ALL_BOOKS_FOR_API)}"
                    try:
                        time.sleep(0.5)
                        payload = requests.get(url_props, timeout=20).json()
                        if isinstance(payload, dict) and 'bookmakers' in payload:
                            for book in payload.get('bookmakers',[]):
                                is_pinnacle = (book['key'] == SHARP_BOOK)
                                for market in book.get('markets',[]):
                                    prop_name = MARKET_NAME_MAP.get(market['key'])
                                    if not prop_name: continue
                                    for outcome in market.get('outcomes',[]):
                                        p_name = outcome.get('description')
                                        if p_name:
                                            norm_name = normalize_player_name(p_name)
                                            if is_pinnacle and outcome.get('point'):
                                                sharp_props_map.setdefault((norm_name, prop_name),[]).append(outcome.get('point'))
                                            if book['key'] in RETAIL_BOOKS:
                                                all_props_raw.append({
                                                    'Game': f"{away_abbr} @ {home_abbr}", 'Game_Time': parser.isoparse(game["commence_time"]).astimezone(eastern).strftime('%-I:%M %p'),
                                                    'Player': p_name, 'Prop': prop_name, 'Line': outcome.get('point'), 'Decimal_Odds': outcome.get('price'),
                                                    'OutcomeName': outcome['name'], 'Bookmaker': book['title'], 'Home_Team_Full': game['home_team'], 'Opponent_Abbr': away_abbr if game['home_team'] == game['home_team'] else home_abbr, 'Sort_Time': game['commence_time'],
                                                    'Game Move': mvmt, 'Game CLV': clv_s
                                                })
                    except: continue
            else:
                print(f"⚠️ API Error or No Games: {games_data}")
        except: traceback.print_exc()

    if run_mode == 'initial':
        try: pd.DataFrame(initial_lines_to_save).to_csv(INITIAL_LINES_PATH, index=False)
        except: pass

    if not all_props_raw:
        print("No props found. This could be due to API issues or no games on the schedule.")
        return

    props_df = pd.DataFrame(all_props_raw)
    props_df['match_key'] = props_df['Player'].apply(normalize_player_name)
    ou_pivot = props_df[props_df['OutcomeName'].isin(['Over', 'Under'])].pivot_table(index=['match_key', 'Player', 'Prop', 'Line', 'Bookmaker', 'Game', 'Game_Time', 'Sort_Time', 'Home_Team_Full', 'Opponent_Abbr', 'Game Move', 'Game CLV'], columns='OutcomeName', values='Decimal_Odds', observed=False).reset_index().rename(columns={'Over': 'Over_Odds', 'Under': 'Under_Odds'})
    merged = pd.merge(ou_pivot, proj_df.drop(columns=['Player'], errors='ignore'), on='match_key')

    merged['Status'] = 'Active'

    def get_display_odds(row):
        bet_type = row.get('Bet', 'Pass')
        dec = row.get('Over_Odds', 0) if bet_type == 'Over' else row.get('Under_Odds', 0)
        if pd.isna(dec) or dec == 0: return ""
        return f"+{int((dec - 1) * 100)}" if dec >= 2.00 else f"{int(-100 / (dec - 1))}"

    def get_proj(row):
        prop = row['Prop']
        if prop == 'PTS': return row.get('Pred_PTS', 0)
        elif prop == 'REB': return row.get('Pred_REB', 0)
        elif prop == 'AST': return row.get('Pred_AST', 0)
        elif prop == '3PT': return row.get('Pred_FG3M', 0)
        elif prop == 'PRA': return row.get('Pred_PRA', 0)
        elif prop == 'P+R': return row.get('Pred_P+R', 0)
        elif prop == 'P+A': return row.get('Pred_P+A', 0)
        elif prop == 'R+A': return row.get('Pred_R+A', 0)
        elif prop == 'DD': return row.get('Pred_DD', 0)
        elif prop == 'TD': return row.get('Pred_TD', 0)
        return 0

    merged['Projection'] = merged.apply(get_proj, axis=1).round(1)
    merged['Proj Range'] = merged.apply(lambda r: f"{max(0, r['Projection'] - mae_values.get(SIMPLE_PROP_MAP.get(r['Prop'], 'PTS'), 1)):.1f}-{r['Projection'] + mae_values.get(SIMPLE_PROP_MAP.get(r['Prop'], 'PTS'), 1):.1f}", axis=1)

    def get_opp_rank(row):
        try: oid = str(int(float(row['OpponentID'])))
        except: return "-"
        rank = def_ranks.get(oid, {}).get(SIMPLE_PROP_MAP.get(row['Prop'], 'PTS'), 99)
        if rank == 99: return "-"
        if rank <= 5: return f"{rank} (Hard)"
        if rank >= 25: return f"{rank} (Easy)"
        return str(rank)

    merged['Opp Rank'] = merged.apply(get_opp_rank, axis=1)
    merged[['Prob_Over', 'Prob_Under']] = merged.apply(lambda r: calculate_ou_probability(r), axis=1, result_type='expand')
    merged['Bet'] = np.where(merged['Prob_Over'] > 0.55, 'Over', np.where(merged['Prob_Under'] > 0.55, 'Under', 'Pass'))
    merged['Fair Prob %'] = merged.apply(calculate_no_vig_prob, axis=1)
    merged['Edge (%)'] = ((np.where(merged['Bet'] == 'Over', merged['Prob_Over']*merged['Over_Odds'], merged['Prob_Under']*merged['Under_Odds']) - 1) * 100).astype(float).round(1)
    merged['Odds'] = merged.apply(get_display_odds, axis=1)

    def get_streak_and_l10(row):
        so, su, ho, hu = calculate_prop_streak(row['PLAYER_ID'], SIMPLE_PROP_MAP.get(row['Prop'], 'PTS'), row['Line'], master_df)
        if row['Bet'] == 'Over':
            return f"W{so}", f"{int(ho)}/10"
        else:
            return f"W{su}", f"{int(hu)}/10"

    streak_l10_df = merged.apply(get_streak_and_l10, axis=1, result_type='expand')
    merged['Streak'] = streak_l10_df[0]
    merged['L10'] = streak_l10_df[1]

    merged['vs Opp'] = merged.apply(lambda r: calculate_head_to_head(r['PLAYER_ID'], r['Prop'], r['Line'], r['Opponent_Abbr'], start_year, master_df), axis=1)

    prev_run_df = get_previous_run_data(file_today_str, OUTPUT_DIR)
    merged = apply_movement_and_ranking(merged, prev_run_df, sharp_props_map)

    merged['Market Disc.'] = (merged['Sharp Line'] - merged['Line']).round(1)
    merged['Model Alpha'] = (merged['Projection'] - merged['Line']).round(1)

    merged['Adv Matchup'] = merged.apply(lambda r: derive_advanced_matchup(r, def_ranks), axis=1)

    def calculate_base_score(row):
        score = 50; bet = row['Bet']; sl = row['Sharp Line']; line = row['Line']
        is_star = row.get('Is_Star', False); trend = row.get('TREND_PTS', 0)
        opp_rank = str(row.get('Opp Rank', ''))
        h2h_str = row.get('vs Opp', '0/0')

        if '/' in h2h_str:
            try:
                hits, total = map(int, h2h_str.split('/'))
                if total >= 2 and hits == total: score += 25
            except: pass

        if trend > 5.0 and not is_star:
            score -= 25
        elif trend > 5.0 and is_star:
            score += 10

        if bet == 'Over' and "(Hard)" in opp_rank:
            if is_star: score -= 10
            else: score -= 40

        if not pd.isna(sl):
            market_diff = abs(sl - line)
            if (bet == 'Over' and sl < line) or (bet == 'Under' and sl > line):
                score -= 40
            elif (bet == 'Over' and sl > line) or (bet == 'Under' and sl < line):
                score += 20
            if market_diff >= 1.5:
                score += 30
        return score

    merged['Score'] = merged.apply(calculate_base_score, axis=1)

    def classify_bet(row):
        sl = row.get('Sharp Line'); line = row.get('Line')
        is_star = row.get('Is_Star', False); trend = row.get('TREND_PTS', 0)
        alpha = row.get('Model Alpha', 0)
        h2h_str = row.get('vs Opp', '0/0')

        if '/' in h2h_str:
            try:
                hits, total = map(int, h2h_str.split('/'))
                if total >= 3 and hits == total: return "🔥 H2H CRUSHER"
            except: pass

        high_intensity_venues = ['New York Knicks', 'Boston Celtics', 'Philadelphia 76ers', 'Denver Nuggets']
        if row['Home_Team_Full'] in high_intensity_venues and row['Team'] == get_team_abbr_map().get(row['Home_Team_Full']):
            if row['Score'] >= 50: return "🏟️ VENUE BOOST"

        if is_star and trend > 2.0 and alpha > 1.5: return "🔥 STANDOUT STAR"
        if not pd.isna(sl) and not pd.isna(line) and abs(sl - line) >= 1.5: return "🚨 SHARP ALERT"

        if row['Score'] >= 70: return "💎 ALPHA (Lock)"
        if row['Score'] >= 50: return "📈 BETA (Parlay)"
        if row['Score'] >= 30: return "🤖 GAMMA (Spec)"
        return "PASS"

    merged['Bet Class'] = merged.apply(classify_bet, axis=1)

    def build_fd_my_way(df):
        my_way_rows =[]
        for _, row in df.iterrows():
            prop = row.get('Prop'); proj = row.get('Projection', 0); bet = row.get('Bet', '')
            sl = row.get('Sharp Line'); line = row.get('Line')
            h2h_str = row.get('vs Opp', '0/0')
            edge = row.get('Edge (%)', 0)
            p_over = row.get('Prob_Over', 0.5); p_under = row.get('Prob_Under', 0.5)

            if not pd.isna(sl):
                if bet == 'Over' and sl < (line - 0.5): continue
                if bet == 'Under' and sl > (line + 0.5): continue

            mae = mae_values.get(SIMPLE_PROP_MAP.get(prop, 'PTS'), 1)
            raw_alt = proj - (0.75 * mae) if bet == 'Over' else proj + (0.75 * mae)

            # Snap to genuine sports betting half-point or whole number step
            alt_line = max(0.5, round(raw_alt * 2.0) / 2.0)

            # Recalibrated dynamic confidence between 72% and 94.5%
            target_prob = p_over if bet == 'Over' else p_under
            dyn_conf = min(94.5, max(72.0, (target_prob * 100.0) + (edge * 0.15) + (2.5 if row.get('Is_Star') else 0)))

            evidence = ["Statistically Safe"]
            if not pd.isna(sl) and ((bet == 'Over' and sl > line) or (bet == 'Under' and sl < line)):
                evidence.append("💎 Sharp Support")
                dyn_conf = min(94.5, dyn_conf + 3.0)
            if row.get('Is_Star'):
                evidence.append("⭐ Star Power")
            if row['Bet Class'] == "🏟️ VENUE BOOST":
                evidence.append("🏟️ Home Intensity")
            if '/' in h2h_str:
                try:
                    hits, total = map(int, h2h_str.split('/'))
                    if total >= 2 and hits == total:
                        evidence.append(f"🎯 H2H {h2h_str}")
                        dyn_conf = min(94.5, dyn_conf + 2.5)
                except: pass

            indicator = "🟢 OVER" if bet == 'Over' else "🔴 UNDER"

            my_way_rows.append({
                'Sort_Time': row.get('Sort_Time'),
                'Game_Time': row.get('Game_Time'),
                'Game': row.get('Game'),
                'Player': row.get('Player'),
                'Prop': prop,
                'Bet': bet,
                'Odds': row.get('Odds'),
                'Fair Prob %': row.get('Fair Prob %'),
                'Standard Line': line,
                'Alt Line': alt_line,
                'Direction': indicator,
                'Projection': proj,
                'Confidence': f"{dyn_conf:.1f}%",
                'Supporting Evidence': ", ".join(evidence)
            })
        return pd.DataFrame(my_way_rows)

    if run_mode == 'final':
        ts_str = live_now_et.strftime('%-m-%-d-%y %-I:%M%p').lower()
        out_file = os.path.join(OUTPUT_DIR, f"{ts_str} NBA Pro Dashboard.xlsx")
    else:
        out_file = os.path.join(OUTPUT_DIR, f"{file_today_str} NBA Pro Dashboard.xlsx")

    with pd.ExcelWriter(out_file, engine='xlsxwriter') as writer:
        workbook = writer.book
        header_fmt = workbook.add_format({'bold': True, 'align': 'center', 'valign': 'vcenter', 'bg_color': '#D9D9D9', 'border': 1})
        center_fmt = workbook.add_format({'align': 'center', 'valign': 'vcenter', 'border': 1})
        green_fmt = workbook.add_format({'bg_color': '#C6EFCE', 'font_color': '#006100', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        red_fmt = workbook.add_format({'bg_color': '#FFC7CE', 'font_color': '#9C0006', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        orange_fmt = workbook.add_format({'bg_color': '#FFEB9C', 'font_color': '#9C5700', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        blue_fmt = workbook.add_format({'bg_color': '#BDD7EE', 'font_color': '#1F4E78', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        purple_fmt = workbook.add_format({'bg_color': '#CCC0DA', 'font_color': '#3F3F76', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        perfect_fmt = workbook.add_format({'bg_color': '#00FF00', 'font_color': '#000000', 'align': 'center', 'valign': 'vcenter', 'border': 1, 'bold': True})
        yellow_fmt = workbook.add_format({'bg_color': '#FFFF00', 'font_color': '#000000', 'align': 'center', 'valign': 'vcenter', 'border': 1})

        green_row_fmt = workbook.add_format({'bg_color': '#C6EFCE', 'font_color': '#006100', 'align': 'left', 'valign': 'vcenter', 'border': 1})
        blue_row_fmt = workbook.add_format({'bg_color': '#BDD7EE', 'font_color': '#1F4E78', 'align': 'left', 'valign': 'vcenter', 'border': 1})
        red_row_fmt = workbook.add_format({'bg_color': '#FFC7CE', 'font_color': '#9C0006', 'align': 'left', 'valign': 'vcenter', 'border': 1})
        yellow_row_fmt = workbook.add_format({'bg_color': '#FFFF00', 'font_color': '#000000', 'align': 'left', 'valign': 'vcenter', 'border': 1})
        purple_row_fmt = workbook.add_format({'bg_color': '#CCC0DA', 'font_color': '#3F3F76', 'align': 'left', 'valign': 'vcenter', 'border': 1})

        # --- ENHANCED HOW TO USE SHEET ---
        ws_howto = workbook.add_worksheet("HOW TO USE")
        ws_howto.set_column(0, 0, 32.0); ws_howto.set_column(1, 1, 110.0); ws_howto.freeze_panes(1, 0)
        row_idx = 0
        ws_howto.write(row_idx, 0, "Metric / Column / Tier", header_fmt); ws_howto.write(row_idx, 1, "Professional Sports Betting Interpretation & Operational Guide", header_fmt); row_idx += 1

        ws_howto.write(row_idx, 0, "💎 ALPHA (Lock)", perfect_fmt); ws_howto.write(row_idx, 1, "Model, Pinnacle Sharp Consensus, and Price all converge with positive expectation (Score >= 70).", perfect_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "🔥 STANDOUT STAR", perfect_fmt); ws_howto.write(row_idx, 1, "Elite usage star on an upward trend with high model alpha. Virtually matchup-proof.", perfect_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "🔥 H2H CRUSHER", perfect_fmt); ws_howto.write(row_idx, 1, "Player holds an undefeated historical record against this specific opponent over the last 2 seasons.", perfect_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "🏟️ VENUE BOOST", blue_fmt); ws_howto.write(row_idx, 1, "Home court intensity boost (+2.0 spread / +50 base score) for elite home arenas (MSG, TD Garden, Wells Fargo).", blue_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "📈 BETA (Parlay Leg)", green_fmt); ws_howto.write(row_idx, 1, "Confirmed sharp market support. Excellent secondary building block for standard parlays (Score 50-70).", green_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "🤖 GAMMA (Speculative)", yellow_fmt); ws_howto.write(row_idx, 1, "High mathematical model edge captured before retail books adjust to sharp price changes (Score 30-50).", yellow_fmt); row_idx += 1
        row_idx += 1

        ws_howto.write(row_idx, 0, "Fair Prob % (No-Vig)", header_fmt); ws_howto.write(row_idx, 1, "The true market-implied win probability after mathematically stripping the sportsbook's vig (juice).", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Market Disc. (Discrepancy)", header_fmt); ws_howto.write(row_idx, 1, "Pinnacle Sharp Line minus Retail Line. Green highlights indicate retail book is lagging the sharp market.", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Model Alpha", header_fmt); ws_howto.write(row_idx, 1, "Model Projection minus Retail Line. Raw statistical projection edge over the retail pricing.", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Proj vs Sharp", header_fmt); ws_howto.write(row_idx, 1, "Model Projection minus Pinnacle Consensus. Highlighted green when at least 0.5 units in your bet's favor.", header_fmt); row_idx += 1
        row_idx += 1

        ws_howto.write(row_idx, 0, "Consistency (Vol Check)", header_fmt); ws_howto.write(row_idx, 1, "Measures coefficient of variation. High (Locked) / Med (Solid) / Volatile (Caution).", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Adv Matchup", header_fmt); ws_howto.write(row_idx, 1, "Exposes scheme strengths: Elite P&R, Elite Rim, Elite Perim vs Trap Matchups.", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Shot Quality", header_fmt); ws_howto.write(row_idx, 1, "Regression indicator: High SQ (open looks), Due (positive regression), Luck Risk (over-performing).", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Game Move & Game CLV", header_fmt); ws_howto.write(row_idx, 1, "Steam Chase indicates rapid sharp market money. Closing Line Value (CLV) confirms beating the closing line.", header_fmt); row_idx += 1
        row_idx += 1

        ws_howto.write(row_idx, 0, "FD My Way Sheet Guide", header_fmt); ws_howto.write(row_idx, 1, "Curated alternate milestone lines rounded to clean half/whole numbers with dynamically calculated confidence.", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Game Lines Sheet Guide", header_fmt); ws_howto.write(row_idx, 1, "Monte Carlo multivariate spread and total simulations benchmarked exclusively against Pinnacle sharp lines.", header_fmt); row_idx += 1

        # --- OUT COLUMNS SEQUENCE ---
        out_cols = [
            'Game_Time', 'Game', 'Status', 'Experience', 'Team', 'Player', 'Prop',
            'Line', 'Bet', 'Odds', 'Fair Prob %', 'Projection', 'Sharp Line', 'Market Disc.', 'Proj vs Sharp',
            'Model Alpha', 'Edge (%)', 'Consistency', 'Opp Rank', 'Adv Matchup', 'Streak', 'L10', 'vs Opp',
            'Shot Quality', 'Game Move', 'Game CLV', 'Sharp Signal', 'Ranking', 'Score', 'Bet Class'
        ]

        def auto_fit_columns(worksheet, dataframe, extra_padding=5):
            for col_idx, col_name in enumerate(dataframe.columns):
                max_len = len(str(col_name))
                if not dataframe.empty:
                    series_max = dataframe[col_name].astype(str).map(len).max()
                    if pd.notna(series_max):
                        max_len = max(max_len, int(series_max))
                worksheet.set_column(col_idx, col_idx, max(11, max_len + extra_padding), center_fmt)

        sheet_config =['DraftKings_EV_Bets', 'DraftKings_EV_SGP', 'FanDuel_EV_Bets', 'FanDuel_EV_SGP']
        for sheet_name in sheet_config:
            bookmaker = 'DraftKings' if 'DraftKings' in sheet_name else 'FanDuel'
            is_sgp = 'SGP' in sheet_name

            if run_mode == 'final':
                book_df = merged[(merged['Bookmaker'] == bookmaker) & (merged['Bet'] != 'Pass') & (merged['Score'] >= 30) & (merged['Bet Class'] != 'PASS')].copy()
            else:
                book_df = merged[(merged['Bookmaker'] == bookmaker) & (merged['Bet'] != 'Pass')].copy()

            if not book_df.empty:
                if 'Fair Prob %' not in book_df.columns:
                    book_df['Fair Prob %'] = ""

                # Group players with multiple props together, ordered by Score
                if is_sgp:
                    final_df = book_df.sort_values(['Sort_Time', 'Game', 'Player', 'Score'], ascending=[True, True, True, False]).reindex(columns=out_cols)
                    final_df = filter_sgp_correlations(final_df)
                else:
                    final_df = book_df.sort_values(['Player', 'Score'], ascending=[True, False]).reindex(columns=out_cols)

                final_df.to_excel(writer, sheet_name=sheet_name, index=False, startrow=1, header=False)
                ws = writer.sheets[sheet_name]
                ws.freeze_panes(1, 6)
                ws.autofilter(0, 0, len(final_df), len(out_cols) - 1)

                for col_idx, val in enumerate(final_df.columns):
                    ws.write(0, col_idx, val, header_fmt)

                auto_fit_columns(ws, final_df, extra_padding=5)

                last_row = len(final_df) + 1
                def get_col(name): return xl_col_to_name(out_cols.index(name))

                ws.set_column(out_cols.index('Score'), out_cols.index('Score'), None, None, {'hidden': True})

                line_col = get_col('Line'); bet_col = get_col('Bet'); sharp_col = get_col('Sharp Line')
                proj_col = get_col('Projection'); opp_rank_col = get_col('Opp Rank'); player_col = get_col('Player')
                score_col = get_col('Score'); odds_col = get_col('Odds'); adv_col = get_col('Adv Matchup')
                class_col = get_col('Bet Class'); v_opp_col = get_col('vs Opp')
                proj_vs_sharp_col = get_col('Proj vs Sharp'); consistency_col = get_col('Consistency')
                m_disc_col = get_col('Market Disc.')

                # Bet Class Highlights
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'STAR', 'format': perfect_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'BOOST', 'format': blue_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'ALPHA', 'format': perfect_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'BETA', 'format': green_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'GAMMA', 'format': yellow_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'ALERT', 'format': orange_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'CRUSHER', 'format': perfect_fmt})

                # Player Score Highlights
                ws.conditional_format(f'{player_col}2:{player_col}{last_row}', {'type': 'formula', 'criteria': f'=${score_col}2>=75', 'format': perfect_fmt})
                ws.conditional_format(f'{player_col}2:{player_col}{last_row}', {'type': 'formula', 'criteria': f'=AND(${score_col}2>=55, ${score_col}2<75)', 'format': green_fmt})
                ws.conditional_format(f'{player_col}2:{player_col}{last_row}', {'type': 'formula', 'criteria': f'=AND(${score_col}2>=30, ${score_col}2<55)', 'format': yellow_fmt})

                # Opponent Rank Matchup Highlights
                green_matchup_criteria = f'=OR(AND(${bet_col}2="Over", ISNUMBER(SEARCH("Easy", ${opp_rank_col}2))), AND(${bet_col}2="Under", ISNUMBER(SEARCH("Hard", ${opp_rank_col}2))))'
                ws.conditional_format(f'{opp_rank_col}2:{opp_rank_col}{last_row}', {'type': 'formula', 'criteria': green_matchup_criteria, 'format': green_fmt})
                ws.conditional_format(f'{bet_col}2:{bet_col}{last_row}', {'type': 'formula', 'criteria': green_matchup_criteria, 'format': green_fmt})

                # Proj vs Sharp In Favor of Bet Type (>= 0.5 Over, <= -0.5 Under)
                proj_sharp_criteria = f'=OR(AND(${bet_col}2="Over", ${proj_vs_sharp_col}2>=0.5), AND(${bet_col}2="Under", ${proj_vs_sharp_col}2<=-0.5))'
                ws.conditional_format(f'{proj_vs_sharp_col}2:{proj_vs_sharp_col}{last_row}', {'type': 'formula', 'criteria': proj_sharp_criteria, 'format': green_fmt})

                # Market Discrepancy Highlighting
                ws.conditional_format(f'{m_disc_col}2:{m_disc_col}{last_row}', {'type': 'cell', 'criteria': '>', 'value': 0, 'format': green_fmt})
                ws.conditional_format(f'{m_disc_col}2:{m_disc_col}{last_row}', {'type': 'cell', 'criteria': '<', 'value': 0, 'format': red_fmt})

                # Consistency Med or Better Green
                ws.conditional_format(f'{consistency_col}2:{consistency_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'High', 'format': green_fmt})
                ws.conditional_format(f'{consistency_col}2:{consistency_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'Med', 'format': green_fmt})

                # Adv Matchup
                ws.conditional_format(f'{adv_col}2:{adv_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': '🔥', 'format': green_fmt})
                ws.conditional_format(f'{adv_col}2:{adv_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': '🔒', 'format': green_fmt})
                ws.conditional_format(f'{adv_col}2:{adv_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': '⚠️', 'format': red_fmt})

                # L10 & vs Opp
                for trend_val in ['7/10', '8/10', '9/10', '10/10']:
                    ws.conditional_format(f'{get_col("L10")}2:{get_col("L10")}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': trend_val, 'format': green_fmt})
                ws.conditional_format(f'{v_opp_col}2:{v_opp_col}{last_row}', {'type': 'formula', 'criteria': f'=AND(IFERROR(FIND("/", ${v_opp_col}2)>0, FALSE), IFERROR(VALUE(LEFT(${v_opp_col}2, FIND("/", ${v_opp_col}2)-1)) = VALUE(MID(${v_opp_col}2, FIND("/", ${v_opp_col}2)+1, 10)), FALSE), IFERROR(VALUE(MID(${v_opp_col}2, FIND("/", ${v_opp_col}2)+1, 10)) > 0, FALSE))', 'format': perfect_fmt})

                # Game Move & CLV
                ws.conditional_format(f'{get_col("Game Move")}2:{get_col("Game Move")}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'Steam', 'format': yellow_fmt})
                ws.conditional_format(f'{get_col("Game CLV")}2:{get_col("Game CLV")}{last_row}', {'type': 'cell', 'criteria': '>', 'value': 0, 'format': green_fmt})

            else: writer.book.add_worksheet(sheet_name).write(0, 0, "No bets found.")

        # --- FD MY WAY SHEET ---
        fd_df = merged[(merged['Bookmaker'] == 'FanDuel') & (merged['Bet'] != 'Pass')].copy()
        if not fd_df.empty:
            my_way_df = build_fd_my_way(fd_df)
            if not my_way_df.empty:
                my_way_df = my_way_df.sort_values(by=['Sort_Time', 'Game', 'Player', 'Confidence'], ascending=[True, True, True, False])
                my_way_df.drop(columns=['Sort_Time'], inplace=True, errors='ignore')

                my_way_df.to_excel(writer, sheet_name='FD My Way', index=False, startrow=1, header=False)
                ws_mw = writer.sheets['FD My Way']
                ws_mw.freeze_panes(1, 4)
                ws_mw.autofilter(0, 0, len(my_way_df), len(my_way_df.columns) - 1)

                for col_idx, val in enumerate(my_way_df.columns):
                    ws_mw.write(0, col_idx, val, header_fmt)

                auto_fit_columns(ws_mw, my_way_df, extra_padding=6)

                lr_mw = len(my_way_df) + 1
                c_dir = xl_col_to_name(my_way_df.columns.get_loc('Direction'))
                c_alt = xl_col_to_name(my_way_df.columns.get_loc('Alt Line'))
                c_conf = xl_col_to_name(my_way_df.columns.get_loc('Confidence'))
                c_evid = xl_col_to_name(my_way_df.columns.get_loc('Supporting Evidence'))

                # Color-code Direction (Soft Green / Soft Red)
                ws_mw.conditional_format(f'{c_dir}2:{c_dir}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'OVER', 'format': green_fmt})
                ws_mw.conditional_format(f'{c_dir}2:{c_dir}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'UNDER', 'format': red_fmt})
                ws_mw.conditional_format(f'{c_alt}2:{c_alt}{lr_mw}', {'type': 'cell', 'criteria': '>', 'value': 0, 'format': center_fmt})

                # Dynamic Confidence Highlighting
                ws_mw.conditional_format(f'{c_conf}2:{c_conf}{lr_mw}', {'type': 'cell', 'criteria': '>=', 'value': '"85.0%"', 'format': perfect_fmt})
                ws_mw.conditional_format(f'{c_conf}2:{c_conf}{lr_mw}', {'type': 'cell', 'criteria': '<', 'value': '"85.0%"', 'format': green_fmt})

                # Supporting Evidence Highlighting
                ws_mw.conditional_format(f'{c_evid}2:{c_evid}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'Sharp Support', 'format': purple_fmt})
                ws_mw.conditional_format(f'{c_evid}2:{c_evid}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'Star Power', 'format': blue_fmt})
                ws_mw.conditional_format(f'{c_evid}2:{c_evid}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'H2H', 'format': perfect_fmt})

        # --- GAME LINES SHEET ---
        if game_lines_rows:
            gl_df = pd.DataFrame(game_lines_rows)
            gl_cols =['Time', 'Game', 'Retail Spread', 'Sharp Spread', 'Model Spread', 'Retail Total', 'Sharp Total', 'Model Total', 'Retail ML', 'Model Win %', 'Home_Team_Full', 'Game Move', 'Game CLV']
            gl_df = gl_df.reindex(columns=gl_cols)
            gl_df.to_excel(writer, sheet_name='Game_Lines', index=False)
            ws_gl = writer.sheets['Game_Lines']
            ws_gl.freeze_panes(1, 0); ws_gl.autofilter(0, 0, len(gl_df), len(gl_cols) - 1)

            for col_idx, val in enumerate(gl_df.columns):
                ws_gl.write(0, col_idx, val, header_fmt)

            auto_fit_columns(ws_gl, gl_df, extra_padding=5)

            lr_gl = len(gl_df) + 1
            col_m_spd = xl_col_to_name(gl_cols.index('Model Spread'))
            col_r_spd = xl_col_to_name(gl_cols.index('Retail Spread'))
            col_m_tot = xl_col_to_name(gl_cols.index('Model Total'))
            col_r_tot = xl_col_to_name(gl_cols.index('Retail Total'))
            col_win = xl_col_to_name(gl_cols.index('Model Win %'))
            col_move = xl_col_to_name(gl_cols.index('Game Move'))
            col_clv = xl_col_to_name(gl_cols.index('Game CLV'))

            # Highlights on significant model edges (>2.0 on spread, >3.0 on total)
            ws_gl.conditional_format(f'{col_m_spd}2:{col_m_spd}{lr_gl}', {'type': 'formula', 'criteria': f'=ABS(${col_m_spd}2-${col_r_spd}2)>=2.0', 'format': green_fmt})
            ws_gl.conditional_format(f'{col_m_tot}2:{col_m_tot}{lr_gl}', {'type': 'formula', 'criteria': f'=ABS(${col_m_tot}2-${col_r_tot}2)>=3.0', 'format': green_fmt})
            ws_gl.conditional_format(f'{col_win}2:{col_win}{lr_gl}', {'type': 'cell', 'criteria': '>=', 'value': '"55.0%"', 'format': perfect_fmt})
            ws_gl.conditional_format(f'{col_move}2:{col_move}{lr_gl}', {'type': 'text', 'criteria': 'containing', 'value': 'Steam', 'format': yellow_fmt})
            ws_gl.conditional_format(f'{col_clv}2:{col_clv}{lr_gl}', {'type': 'cell', 'criteria': '>', 'value': 0, 'format': green_fmt})
            ws_gl.conditional_format(f'{col_clv}2:{col_clv}{lr_gl}', {'type': 'cell', 'criteria': '<', 'value': 0, 'format': red_fmt})

    print(f"✅ Exported: {out_file}")

if __name__ == "__main__":
    run_daily_projections()
```

```python
#@title SCRIPT 3: NBA Professional Sharp & ML Engine (v126 - FULLY ENHANCED & ALIGNED)
# ==============================================================================

# --- Part 0: Setup & Installs ---
!pip install -q xlsxwriter nba_api scikit-learn pandas==2.2.3 joblib pytz unidecode requests lightgbm python-dateutil tqdm openpyxl scipy sportsdataverse

import os
import re
import time
import joblib
import warnings
import requests
import json
import glob
import traceback
import numpy as np
import pandas as pd
import pytz
from datetime import datetime, timedelta
from dateutil import parser
from unidecode import unidecode
from tqdm import tqdm
from decimal import Decimal
from scipy.stats import poisson, norm
from google.colab import drive
from sklearn.metrics import mean_absolute_error, mean_squared_error
from IPython.display import display, clear_output
from xlsxwriter.utility import xl_col_to_name

from nba_api.stats.static import teams, players
from sportsdataverse.nba import load_nba_schedule, most_recent_nba_season

warnings.filterwarnings("ignore")

# --- Part 1: Configuration ---
GDRIVE_BASE_PATH = "/content/drive/MyDrive/NBA/"
MODEL_DIR = os.path.join(GDRIVE_BASE_PATH, "Training Model Advanced/")
OUTPUT_DIR = os.path.join(GDRIVE_BASE_PATH, "Predictions/")
GAMELOG_LAKE_DIR = os.path.join(GDRIVE_BASE_PATH, "gamelog_lake/")
DATA_DIR = os.path.join(GDRIVE_BASE_PATH, "Historical Data/")
BOXSCORE_DIR = os.path.join(DATA_DIR, "Boxscores/")
INITIAL_LINES_PATH = os.path.join(OUTPUT_DIR, "initial_nba_lines.csv")

os.makedirs(OUTPUT_DIR, exist_ok=True); os.makedirs(GAMELOG_LAKE_DIR, exist_ok=True)

ENABLE_MARKETS = {
    "Moneyline": True, "Spreads": True, "Totals": True,
    "Points": True, "Rebounds": True, "Assists": True, "Threes": True,
    "Combos": True, "DoubleDouble": True, "TripleDouble": True
}

FILTER_OUT_PLAYERS = True

_GAME_MAP = {"Moneyline": "h2h", "Spreads": "spreads", "Totals": "totals"}
_PROP_MAP = {
    "Points": {'player_points': 'PTS'}, "Rebounds": {'player_rebounds': 'REB'}, "Assists": {'player_assists': 'AST'},
    "Threes": {'player_threes': 'FG3M'},
    "Combos": {'player_points_rebounds_assists': 'PRA', 'player_points_rebounds': 'P+R', 'player_points_assists': 'P+A', 'player_rebounds_assists': 'R+A'},
    "DoubleDouble": {'player_double_double': 'DD'}, "TripleDouble": {'player_triple_double': 'TD'}
}

SIMPLE_PROP_MAP = {
    "Points": "PTS", "PTS": "PTS",
    "Rebounds": "REB", "REB": "REB",
    "Assists": "AST", "AST": "AST",
    "Threes": "FG3M", "FG3M": "FG3M", "3PT": "FG3M",
    "Combos": "PRA", "PRA": "PRA",
    "DoubleDouble": "DD", "DD": "DD",
    "TripleDouble": "TD", "TD": "TD",
    "P+R": "P+R", "P+A": "P+A", "R+A": "R+A"
}

ACTIVE_GAME_MARKETS = [v for k, v in _GAME_MAP.items() if ENABLE_MARKETS.get(k, False)]

MARKET_NAME_MAP = {}
for category, enabled in ENABLE_MARKETS.items():
    if enabled and category in _PROP_MAP:
        MARKET_NAME_MAP.update(_PROP_MAP[category])
ALL_PROP_MARKETS_LIST = list(MARKET_NAME_MAP.keys())

ODDS_API_KEYS = [
    "9b96859c5e0781bb4c5809215a5ed94b", "f4cfcea7cc6120a97559f69d6049596e", "7b9405e2febc561e554b7169a42cefd1", "146dff42ae19a83208b059d4855095c9",
    "f8474479351ccc685ccae9f4716c23c5", "39756c093848b6968c899a4e1ded85f9", "9705a1f6b609305047eb8ec9243cc54a", "4217574ccdd90732483d98f24772f66b",
    "66953d96fcb2ec72a47e7f39901258e7", "94304f58fb95fd1d45159b9c8d292724", "e384411e342cac5c9408804c622703f1", "dab5c3da80b0415f64d4d94890974cdf",
    "9705a1f6b609305047eb8ec9243cc54a", "3aae94b51cc3dcb8e0623bf2633a695a", "a4e4d3eef7d360055b96a1351560817f", "1bfcac154fb2b490bf55a80a18275db1",
    "7b64d450f91079f60211cb33033594b4", "ec9dc1e1aa0543c1b3fa3d2576fabffd", "6bbfe433274aef6fee5aa5543ad69055", "1f5e9cdc190854c605b533917b2c9b82",
    "858b7234b5f77661345f205b1c5ec121", "39756c093848b6968c899a4e1ded85f9", "21caebe8058c338a7ba76f6c9fb550cb", "94304f58fb95fd1d45159b9c8d292724",
    "129b9d482b270ff3809e11b0c445ef5c", "a11767c08916b29f824150bffbbf6d9c", "e7e814c95b32c4a9fb5b72acefb7389b", "3a3cd8a78555c46afa842aa8bdf1455e",
    "3b9adf0f0e27964f58de1f9b4af3adfc", "f098da95884f42852df5ff12f5704741", "768fb25c7f33328bc53b61b81e607bd9"
]

SHARP_BOOKS = ["pinnacle", "betonlineag", "bovada", "betrivers"]
RETAIL_BOOKS = ["draftkings", "fanduel"]
ALL_BOOKS_FOR_API = list(set(SHARP_BOOKS + RETAIL_BOOKS))
ROLLING_WINDOWS = [5, 10, 20]
CONFIG = {
    "GAMES_PLAYED_THRESHOLD_ACCLIMATED": 15,
    "GAMES_PLAYED_THRESHOLD_LOW_VOLUME": 20,
    "STEAM_THRESHOLD": 1.5,
    "RLM_CONFIDENCE": 0.65
}

# --- Part 2: Helper Functions ---

def run_accuracy_audit():
    print("\n" + "="*60)
    print("🧠 PHASE 0: PRE-FLIGHT ACCURACY AUDIT (Last 1,000 Games)")
    print("="*60)
    try:
        if not os.path.exists('/content/drive'):
            drive.mount('/content/drive', force_remount=True)

        df = pd.read_csv(os.path.join(DATA_DIR, "nba_advanced_training_data_v8.csv"))
        df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])

        models = {
            'min': joblib.load(os.path.join(MODEL_DIR, "minutes_lgbm_model.pkl")),
            'pts': joblib.load(os.path.join(MODEL_DIR, "pts_lgbm_model.pkl")),
            'reb': joblib.load(os.path.join(MODEL_DIR, "reb_lgbm_model.pkl")),
            'ast': joblib.load(os.path.join(MODEL_DIR, "ast_lgbm_model.pkl")),
            'fg3m': joblib.load(os.path.join(MODEL_DIR, "fg3m_lgbm_model.pkl"))
        }

        imputers = {
            'minutes': joblib.load(os.path.join(MODEL_DIR, "minutes_imputer.pkl")),
            'stats': joblib.load(os.path.join(MODEL_DIR, "stats_imputer.pkl"))
        }

        min_feats = joblib.load(os.path.join(MODEL_DIR, "minutes_feature_names.pkl"))
        stat_feats = joblib.load(os.path.join(MODEL_DIR, "stats_feature_names.pkl"))

    except Exception as e:
        print(f"❌ CRITICAL ERROR: Could not load files for audit. Error: {e}")
        return {}

    audit_df = df.sort_values('GAME_DATE', ascending=False).head(1000).copy()

    X_min = audit_df[min_feats]
    X_min_imputed = imputers['minutes'].transform(X_min)
    audit_df['PRED_MIN'] = models['min'].predict(X_min_imputed)

    audit_df['Proj_MIN'] = audit_df['PRED_MIN']
    X_stats = audit_df[stat_feats]
    X_stats_imputed = imputers['stats'].transform(X_stats)

    audit_df['PRED_PTS'] = models['pts'].predict(X_stats_imputed)
    audit_df['PRED_REB'] = models['reb'].predict(X_stats_imputed)
    audit_df['PRED_AST'] = models['ast'].predict(X_stats_imputed)
    audit_df['PRED_FG3M'] = models['fg3m'].predict(X_stats_imputed)

    audit_df['PRED_PRA'] = audit_df['PRED_PTS'] + audit_df['PRED_REB'] + audit_df['PRED_AST']
    audit_df['ACTUAL_PRA'] = audit_df['TARGET_PTS'] + audit_df['TARGET_REB'] + audit_df['TARGET_AST']

    audit_df['PRED_P+R'] = audit_df['PRED_PTS'] + audit_df['PRED_REB']
    audit_df['ACTUAL_P+R'] = audit_df['TARGET_PTS'] + audit_df['TARGET_REB']

    audit_df['PRED_P+A'] = audit_df['PRED_PTS'] + audit_df['PRED_AST']
    audit_df['ACTUAL_P+A'] = audit_df['TARGET_PTS'] + audit_df['TARGET_AST']

    audit_df['PRED_R+A'] = audit_df['PRED_REB'] + audit_df['PRED_AST']
    audit_df['ACTUAL_R+A'] = audit_df['TARGET_REB'] + audit_df['TARGET_AST']

    results = []
    comparison_map = {
        'PTS': ('TARGET_PTS', 'PRED_PTS'),
        'REB': ('TARGET_REB', 'PRED_REB'),
        'AST': ('TARGET_AST', 'PRED_AST'),
        '3PT': ('TARGET_FG3M', 'PRED_FG3M'),
        'PRA': ('ACTUAL_PRA', 'PRED_PRA'),
        'P+R': ('ACTUAL_P+R', 'PRED_P+R'),
        'P+A': ('ACTUAL_P+A', 'PRED_P+A'),
        'R+A': ('ACTUAL_R+A', 'PRED_R+A')
    }

    audit_hurdles = {}
    for label, (actual_col, pred_col) in comparison_map.items():
        mae = mean_absolute_error(audit_df[actual_col], audit_df[pred_col])
        rmse = np.sqrt(mean_squared_error(audit_df[actual_col], audit_df[pred_col]))
        avg_actual = audit_df[actual_col].mean()
        rel_error = (mae / avg_actual) * 100 if avg_actual > 0 else 0

        audit_hurdles[label] = rel_error

        results.append({
            'Prop Type': label,
            'Avg Error (MAE)': round(mae, 2),
            'Volatility (RMSE)': round(rmse, 2),
            'Error %': f"{round(rel_error, 1)}%"
        })

    report = pd.DataFrame(results).sort_values('Avg Error (MAE)')
    print(report.to_string(index=False))
    print("="*60 + "\n")
    return audit_hurdles

def load_master_boxscores():
    files = glob.glob(os.path.join(BOXSCORE_DIR, "boxscores_regular_season_and_playoffs_*.parquet"))
    if not files:
        files = glob.glob(os.path.join(DATA_DIR, "*.parquet"))
    if not files:
        return pd.DataFrame()
    latest_file = sorted(files)[-1]
    print(f"📦 Loading historical boxscores from: {os.path.basename(latest_file)}")
    df = pd.read_parquet(latest_file)

    rename_cols = {
        'personId': 'PLAYER_ID', 'player_id': 'PLAYER_ID',
        'personName': 'PLAYER_NAME', 'player_name': 'PLAYER_NAME',
        'game_date': 'GAME_DATE', 'gameId': 'GAME_ID',
        'teamId': 'TEAM_ID', 'team_id': 'TEAM_ID',
        'teamTricode': 'TEAM_ABBREVIATION', 'team_abbreviation': 'TEAM_ABBREVIATION',
        'minutes': 'MIN', 'points': 'PTS', 'reboundsTotal': 'REB', 'assists': 'AST',
        'threePointersMade': 'FG3M', 'threePointersAttempted': 'FG3A',
        'fieldGoalsMade': 'FGM', 'fieldGoalsAttempted': 'FGA',
        'freeThrowsMade': 'FTM', 'freeThrowsAttempted': 'FTA',
        'steals': 'STL', 'blocks': 'BLK', 'turnovers': 'TOV'
    }
    df.rename(columns={k: v for k, v in rename_cols.items() if k in df.columns}, inplace=True)
    df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])
    return df

def get_defense_ranks_from_boxscores(master_df):
    print("   🛡️ Calculating Opponent Defense Rankings from Historical Boxscores...")
    if master_df.empty: return {}, {}
    try:
        recent = master_df.sort_values('GAME_DATE').tail(15000)
        team_stats = recent.groupby('TEAM_ABBREVIATION').agg({
            'PTS': 'mean', 'REB': 'mean', 'AST': 'mean', 'FG3M': 'mean'
        }).reset_index()

        team_stats['CALC_RANK_PTS'] = team_stats['PTS'].rank(ascending=False)
        team_stats['CALC_RANK_REB'] = team_stats['REB'].rank(ascending=False)
        team_stats['CALC_RANK_AST'] = team_stats['AST'].rank(ascending=False)
        team_stats['CALC_RANK_FG3M'] = team_stats['FG3M'].rank(ascending=False)

        team_stats['PRA'] = team_stats['PTS'] + team_stats['REB'] + team_stats['AST']
        team_stats['P+R'] = team_stats['PTS'] + team_stats['REB']
        team_stats['P+A'] = team_stats['PTS'] + team_stats['AST']
        team_stats['R+A'] = team_stats['REB'] + team_stats['AST']

        team_stats['CALC_RANK_PRA'] = team_stats['PRA'].rank(ascending=False)
        team_stats['CALC_RANK_P+R'] = team_stats['P+R'].rank(ascending=False)
        team_stats['CALC_RANK_P+A'] = team_stats['P+A'].rank(ascending=False)
        team_stats['CALC_RANK_R+A'] = team_stats['R+A'].rank(ascending=False)

        ranks, raw_stats = {}, {}
        abbr_to_id = {t['abbreviation']: str(t['id']) for t in teams.get_teams()}

        for _, row in team_stats.iterrows():
            tid = abbr_to_id.get(row['TEAM_ABBREVIATION'], row['TEAM_ABBREVIATION'])
            ranks[tid] = {
                'PTS': int(row['CALC_RANK_PTS']), 'REB': int(row['CALC_RANK_REB']),
                'AST': int(row['CALC_RANK_AST']), 'FG3M': int(row['CALC_RANK_FG3M']),
                'PRA': int(row['CALC_RANK_PRA']), 'P+R': int(row['CALC_RANK_P+R']),
                'P+A': int(row['CALC_RANK_P+A']), 'R+A': int(row['CALC_RANK_R+A'])
            }
            raw_stats[tid] = {
                'PTS': float(row['PTS']), 'REB': float(row['REB']),
                'AST': float(row['AST']), 'FG3M': float(row['FG3M'])
            }
        return ranks, raw_stats
    except Exception as e:
        print(f"   ⚠️ Could not compute local defense ranks: {e}")
        return {}, {}

def get_team_metrics_from_boxscores(master_df):
    print("   📊 Calculating Team Metrics from Historical Boxscores...")
    if master_df.empty: return {}
    metrics_dict = {}
    try:
        abbr_to_id = {t['abbreviation']: t['id'] for t in teams.get_teams()}
        recent = master_df.sort_values('GAME_DATE').tail(40000)
        game_totals = recent.groupby(['GAME_ID', 'TEAM_ABBREVIATION']).agg({
            'PTS': 'sum', 'FGA': 'sum', 'FTA': 'sum', 'FGM': 'sum', 'FG3M': 'sum'
        }).reset_index()
        grouped = game_totals.groupby('TEAM_ABBREVIATION').agg({
            'PTS': 'mean', 'FGA': 'mean', 'FTA': 'mean', 'FGM': 'mean', 'FG3M': 'mean'
        }).reset_index()

        for _, r in grouped.iterrows():
            tid = abbr_to_id.get(r['TEAM_ABBREVIATION'])
            if tid:
                efg = (r['FGM'] + 0.5 * r['FG3M']) / r['FGA'] if r['FGA'] > 0 else 0.52
                metrics_dict[tid] = {
                    'E_OFF_RATING': r['PTS'], 'E_DEF_RATING': 115.0,
                    'E_NET_RATING': 0.0, 'E_PACE': 100.0,
                    'E_EFG_PCT': efg, 'E_OREB_PCT': 0.28,
                    'GP': 82, 'W': 41, 'L': 41
                }
        return metrics_dict
    except Exception as e:
        print(f"   ⚠️ Could not compute local team metrics: {e}")
        return {}

def assign_live_player_trait_runtime(player_id, player_name, logs, trait_map_df):
    if not trait_map_df.empty:
        match = trait_map_df[trait_map_df['PLAYER_ID'] == int(player_id)]
        if not match.empty:
            return int(match.iloc[-1]['TRAIT'])

    if logs.empty: return 4
    try:
        recent = logs.head(15).copy()
        recent['MIN_dec'] = recent['MIN'].apply(convert_min_to_decimal)
        mins = max(1.0, recent['MIN_dec'].mean())
        pts_pm = recent['PTS'].mean() / mins
        reb_pm = recent['REB'].mean() / mins
        ast_pm = recent['AST'].mean() / mins
        fg3m_pm = recent['FG3M'].mean() / mins
        vector = np.array([pts_pm, reb_pm, ast_pm, fg3m_pm])

        cluster_norms = {
            0: np.array([0.3, 0.2, 0.05, 0.05]),
            1: np.array([0.6, 0.15, 0.2, 0.08]),
            2: np.array([0.4, 0.12, 0.12, 0.06]),
            3: np.array([0.4, 0.08, 0.25, 0.05]),
            4: np.array([0.25, 0.1, 0.05, 0.08]),
            5: np.array([0.5, 0.3, 0.08, 0.01]),
            6: np.array([0.15, 0.05, 0.08, 0.04]),
            7: np.array([0.5, 0.1, 0.12, 0.10]),
            8: np.array([0.3, 0.06, 0.1, 0.08]),
            9: np.array([0.1, 0.04, 0.04, 0.02]),
        }

        best_trait = 4
        min_dist = float('inf')
        for trait_id, center in cluster_norms.items():
            dist = np.linalg.norm(vector - center)
            if dist < min_dist:
                min_dist = dist
                best_trait = trait_id
        return best_trait
    except:
        return 4

def get_coach_games_played(team_abbr, coach_df, logs, season_start_year):
    if coach_df.empty or logs.empty: return 0
    match = coach_df[(coach_df['team'] == team_abbr) & (coach_df['from_year'] <= season_start_year) & (coach_df['to_year'] >= season_start_year)]
    if match.empty: return 0
    season_cutoff = datetime(season_start_year, 9, 1)
    try:
        logs['GAME_DATE_DT'] = pd.to_datetime(logs['GAME_DATE'])
        cur_season_logs = logs[logs['GAME_DATE_DT'] >= season_cutoff]
        return len(cur_season_logs)
    except: return 0

def normalize_player_name(name):
    if name is None: return None
    try: name = str(name)
    except: return None
    name = unidecode(name.lower()); name = re.sub(r'\s+(jr|sr|ii|iii|iv)\.?$', '', name); name = re.sub(r'[^\w\s]', '', name); return re.sub(r'\s+', ' ', name).strip()

def get_team_abbr_map(): return {team['full_name']: team['abbreviation'] for team in teams.get_teams()}

def convert_min_to_decimal(min_str):
    if isinstance(min_str, str) and ':' in min_str:
        try: minutes, seconds = map(int, min_str.split(':')); return minutes + seconds / 60.0
        except: return np.nan
    return pd.to_numeric(min_str, errors='coerce')

def derive_position(row):
    pts = row.get('Pred_PTS', 0); reb = row.get('Pred_REB', 0); ast = row.get('Pred_AST', 0)
    if reb >= 8.0: return 'C'
    if reb >= 6.0 and ast < 2.5: return 'C/F'
    if ast >= 6.0: return 'PG'
    if ast >= 4.5 and reb < 4.0: return 'PG'
    if reb >= 5.0 and pts >= 12.0: return 'F'
    if pts >= 10.0 and reb < 4.5: return 'G'
    return 'G/F'

def get_player_experience(player_id, master_df):
    try:
        p_logs = master_df[master_df['PLAYER_ID'].astype(str) == str(player_id)]
        if p_logs.empty: return "Rookie"
        seasons_played = p_logs['season_year'].nunique() if 'season_year' in p_logs.columns else len(p_logs) // 50
        teams_played = p_logs['TEAM_ABBREVIATION'].nunique()
        if seasons_played <= 1: return "Rookie"
        if teams_played > 1 and len(p_logs.tail(15)['TEAM_ABBREVIATION'].unique()) > 1: return "Traded Veteran"
        if len(p_logs) < CONFIG["GAMES_PLAYED_THRESHOLD_LOW_VOLUME"]: return "Low Volume"
        return "Veteran"
    except Exception:
        return "Veteran"

def fetch_espn_nba_injuries():
    print("\nFetching live injury report from ESPN API (NBA)...")
    try:
        teams_url = "https://site.api.espn.com/apis/site/v2/sports/basketball/nba/teams"
        teams_data = requests.get(teams_url, timeout=10).json()
        team_list = teams_data['sports'][0]['leagues'][0]['teams']
        all_injuries =[]
        for team in tqdm(team_list, desc="Fetching Team Rosters for Injuries"):
            team_id = team['team']['id']
            roster_url = f"https://site.api.espn.com/apis/site/v2/sports/basketball/nba/teams/{team_id}?enable=roster"
            try: roster_data = requests.get(roster_url, timeout=10).json()
            except Exception: continue
            for player in roster_data.get('team', {}).get('athletes',[]):
                if 'injuries' in player and player['injuries']:
                    injury_obj = player['injuries'][0]
                    status = injury_obj.get('status', 'Unknown')
                    all_injuries.append({'Player_ESPN': player.get('displayName'), 'Status': status, 'Injury Details': str(injury_obj.get('date', '')) + " " + str(injury_obj.get('type', {}).get('description', ''))})

        if not all_injuries: print("   ℹ No reported injuries found."); return pd.DataFrame()
        injury_df = pd.DataFrame(all_injuries)
        injury_df['match_key'] = injury_df['Player_ESPN'].apply(normalize_player_name)
        print(f"   ✅ Successfully fetched {len(injury_df)} injury entries.")
        return injury_df
    except Exception as e: print(f"   ⚠️ Could not fetch ESPN injuries: {e}"); return pd.DataFrame()

def calculate_prop_streak(player_id, prop_map, line, master_df):
    try:
        gamelog_file = os.path.join(GAMELOG_LAKE_DIR, f"{player_id}.csv")
        if os.path.exists(gamelog_file):
            df = pd.read_csv(gamelog_file)
        elif not master_df.empty:
            df = master_df[master_df['PLAYER_ID'].astype(str) == str(player_id)].copy()
        else:
            return 0, 0, 0, 0

        if df.empty: return 0, 0, 0, 0
        df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])
        df = df.sort_values(by='GAME_DATE', ascending=False).reset_index(drop=True)
        col_name = SIMPLE_PROP_MAP.get(prop_map, prop_map)

        if col_name == 'PTS': val_series = df['PTS']
        elif col_name == 'REB': val_series = df['REB']
        elif col_name == 'AST': val_series = df['AST']
        elif col_name == 'FG3M': val_series = df['FG3M']
        elif col_name == 'PRA': val_series = df['PTS'] + df['REB'] + df['AST']
        elif col_name == 'P+R': val_series = df['PTS'] + df['REB']
        elif col_name == 'P+A': val_series = df['PTS'] + df['AST']
        elif col_name == 'R+A': val_series = df['REB'] + df['AST']
        elif col_name == 'DD': val_series = ((df['PTS']>=10).astype(int) + (df['REB']>=10).astype(int) + (df['AST']>=10).astype(int)) >= 2
        else: val_series = df.get(col_name)

        if val_series is None: return 0, 0, 0, 0

        l10_slice = val_series.head(10)

        streak_over = 0
        for val in l10_slice:
            if col_name == 'DD':
                if val: streak_over += 1
                else: break
            else:
                if val > line: streak_over += 1
                else: break

        streak_under = 0
        for val in l10_slice:
            if col_name == 'DD':
                if not val: streak_under += 1
                else: break
            else:
                if val < line: streak_under += 1
                else: break

        hits_over_l10 = l10_slice.sum() if col_name == 'DD' else (l10_slice > line).sum()
        hits_under_l10 = (10 - l10_slice.sum()) if col_name == 'DD' else (l10_slice < line).sum()

        return streak_over, streak_under, hits_over_l10, hits_under_l10
    except Exception: return 0, 0, 0, 0

def calculate_head_to_head(player_id, prop_map, line, opponent_abbr, season_year, master_df):
    try:
        gamelog_file = os.path.join(GAMELOG_LAKE_DIR, f"{player_id}.csv")
        if os.path.exists(gamelog_file):
            df = pd.read_csv(gamelog_file)
        elif not master_df.empty:
            df = master_df[master_df['PLAYER_ID'].astype(str) == str(player_id)].copy()
        else:
            return "0/0"

        if df.empty: return "0/0"

        df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])
        cutoff = datetime.now() - timedelta(days=365*2)
        df = df[df['GAME_DATE'] >= cutoff]

        col_name = SIMPLE_PROP_MAP.get(prop_map, prop_map)
        target_opp = str(opponent_abbr).upper()

        abbr_map = {'PHX': 'PHO', 'PHO': 'PHX', 'GSW': 'GS', 'GS': 'GSW', 'NOP': 'NO', 'NO': 'NOP', 'SAS': 'SA', 'SA': 'SAS', 'NYK': 'NY', 'NY': 'NYK', 'UTA': 'UT', 'UT': 'UTA', 'WAS': 'WSH', 'WSH': 'WAS'}
        alt_opp = abbr_map.get(target_opp, target_opp)

        def is_matchup(m_str):
            if pd.isna(m_str): return False
            m_str = str(m_str).replace('.', '').upper()
            tokens = m_str.split()
            actual_opp = tokens[-1] if tokens else ""
            return actual_opp == target_opp or actual_opp == alt_opp

        if 'MATCHUP' in df.columns:
            df_vs = df[df['MATCHUP'].apply(is_matchup)]
        elif 'matchup' in df.columns:
            df_vs = df[df['matchup'].apply(is_matchup)]
        else:
            return "0/0"

        if df_vs.empty: return "0/0"

        if col_name == 'PRA': val_series = df_vs['PTS'] + df_vs['REB'] + df_vs['AST']
        elif col_name == 'P+R': val_series = df_vs['PTS'] + df_vs['REB']
        elif col_name == 'P+A': val_series = df_vs['PTS'] + df_vs['AST']
        elif col_name == 'R+A': val_series = df_vs['REB'] + df_vs['AST']
        else: val_series = df_vs[col_name]

        hits = (val_series > line).sum()
        return f"{int(hits)}/{len(df_vs)}"
    except Exception: return "0/0"

def get_live_schedule_from_sportsdataverse(iso_date):
    schedule = {}
    print(f"--- Fetching Schedule for {iso_date} via SportsDataverse ---")
    try:
        current_year = most_recent_nba_season()
        try:
            df = load_nba_schedule(seasons=[current_year], return_as_pandas=True)
        except Exception:
            df = load_nba_schedule(seasons=[current_year - 1], return_as_pandas=True)

        if df is None or df.empty:
            return get_live_schedule_from_espn(iso_date)

        date_col = 'game_date' if 'game_date' in df.columns else ('date' if 'date' in df.columns else None)
        if not date_col:
            return get_live_schedule_from_espn(iso_date)

        df[date_col] = df[date_col].astype(str)
        todays_games = df[df[date_col].str.contains(iso_date, na=False)]

        if todays_games.empty:
            return get_live_schedule_from_espn(iso_date)

        nba_teams = teams.get_teams()
        abbr_to_id = {t['abbreviation']: str(t['id']) for t in nba_teams}

        valid_count = 0
        for _, game in todays_games.iterrows():
            try:
                h_abbr = game.get('home_team_abbreviation', game.get('home_abbreviation', ''))
                a_abbr = game.get('away_team_abbreviation', game.get('away_abbreviation', ''))

                mapping = {'PHO': 'PHX', 'GS': 'GSW', 'NO': 'NOP', 'WSH': 'WAS', 'UTAH': 'UTA', 'SA': 'SAS', 'NY': 'NYK'}
                h_abbr = mapping.get(h_abbr, h_abbr)
                a_abbr = mapping.get(a_abbr, a_abbr)

                h_id = abbr_to_id.get(h_abbr)
                v_id = abbr_to_id.get(a_abbr)

                if h_id and v_id:
                    schedule[str(h_id)] = str(v_id)
                    schedule[str(v_id)] = str(h_id)
                    valid_count += 1
            except Exception:
                continue

        print(f"✅ Found {valid_count} games via SportsDataverse schedule.")
        return schedule
    except Exception as e:
        print(f"⚠️ SportsDataverse schedule lookup failed, falling back to ESPN: {e}")
        return get_live_schedule_from_espn(iso_date)

def get_live_schedule_from_espn(iso_date):
    schedule = {}
    print(f"--- Fetching Schedule for {iso_date} via ESPN Live API ---")
    try:
        date_clean = iso_date.replace('-', '')
        url = f"https://site.api.espn.com/apis/site/v2/sports/basketball/nba/scoreboard?dates={date_clean}"
        res = requests.get(url, timeout=10).json()
        events = res.get('events', [])

        if not events:
            print(f"ℹ No games scheduled on ESPN for {iso_date}")
            return {}

        nba_teams = teams.get_teams()
        abbr_to_id = {t['abbreviation']: str(t['id']) for t in nba_teams}

        valid_count = 0
        for ev in events:
            try:
                comps = ev.get('competitions', [{}])[0]
                competitors = comps.get('competitors', [])
                if len(competitors) == 2:
                    c_home = competitors[0] if competitors[0].get('homeAway') == 'home' else competitors[1]
                    c_away = competitors[1] if competitors[0].get('homeAway') == 'home' else competitors[0]

                    h_abbr = c_home['team']['abbreviation']
                    a_abbr = c_away['team']['abbreviation']

                    mapping = {'PHO': 'PHX', 'GS': 'GSW', 'NO': 'NOP', 'WSH': 'WAS', 'UTAH': 'UTA', 'SA': 'SAS', 'NY': 'NYK'}
                    h_abbr = mapping.get(h_abbr, h_abbr)
                    a_abbr = mapping.get(a_abbr, a_abbr)

                    h_id = abbr_to_id.get(h_abbr)
                    v_id = abbr_to_id.get(a_abbr)

                    if h_id and v_id:
                        schedule[str(h_id)] = str(v_id)
                        schedule[str(v_id)] = str(h_id)
                        valid_count += 1
            except Exception:
                continue

        print(f"✅ Found {valid_count} games on the live schedule.")
        return schedule
    except Exception as e:
        print(f"❌ Error fetching ESPN schedule: {e}")
        return {}

def get_player_gamelogs_from_master(player_id, master_df):
    gamelog_file = os.path.join(GAMELOG_LAKE_DIR, f"{player_id}.csv")
    if os.path.exists(gamelog_file):
        try:
            return pd.read_csv(gamelog_file)
        except Exception:
            pass

    if not master_df.empty:
        p_df = master_df[master_df['PLAYER_ID'].astype(str) == str(player_id)].copy()
        if not p_df.empty:
            p_df = p_df.sort_values(by="GAME_DATE", ascending=False)
            p_df.to_csv(gamelog_file, index=False)
            return p_df

    return pd.DataFrame()

def calculate_live_features(logs):
    if logs.empty or len(logs) < 5: return {}
    logs_for_calc = logs.copy(); features = {}
    logs_for_calc['MIN'] = logs_for_calc['MIN'].apply(convert_min_to_decimal)
    base_stats =['PTS', 'REB', 'AST', 'FGA', 'FG3A', 'FG3M', 'FTA', 'TOV', 'FGM']
    for col in base_stats:
        if col in logs_for_calc.columns: logs_for_calc[col] = pd.to_numeric(logs_for_calc[col], errors='coerce')
        else: logs_for_calc[col] = np.nan
    logs_for_calc.dropna(subset=['MIN'], inplace=True);
    if logs_for_calc.empty: return {}
    avg_minutes = logs_for_calc['MIN'].mean()
    weights = np.ones(len(logs_for_calc)); weights[logs_for_calc['MIN'] < 0.7 * avg_minutes] = 0.5; weights[logs_for_calc['MIN'] > 1.15 * avg_minutes] = 1.2
    logs_for_calc['WEIGHT'] = weights
    logs_safe = logs_for_calc[logs_for_calc['MIN'] > 0].copy()

    logs_safe['EFG_PCT_RAW'] = (logs_safe['FGM'] + 0.5 * logs_safe['FG3M']) / logs_safe['FGA'].replace(0, np.nan)
    logs_safe['TS_PCT_RAW'] = logs_safe['PTS'] / (2 * (logs_safe['FGA'] + 0.44 * logs_safe['FTA'])).replace(0, np.nan)
    logs_safe[['EFG_PCT_RAW', 'TS_PCT_RAW']] = logs_safe[['EFG_PCT_RAW', 'TS_PCT_RAW']].fillna(0)

    features['EFG_PCT_PREV'] = logs_safe['EFG_PCT_RAW'].iloc[0] if not logs_safe.empty else 0
    features['TS_PCT_PREV'] = logs_safe['TS_PCT_RAW'].iloc[0] if not logs_safe.empty else 0

    stats_for_ratios =['PTS', 'REB', 'AST', 'FGA', 'FG3A', 'FTA', 'TOV', 'FG3M']
    for stat in stats_for_ratios:
        if stat in logs_safe.columns: logs_safe[f'{stat}_PM'] = logs_safe[stat] / logs_safe['MIN']
        else: logs_safe[f'{stat}_PM'] = 0

    stats_to_roll =['PTS', 'REB', 'AST', 'MIN', 'FG3M', 'FG3A', 'EFG_PCT_RAW', 'TS_PCT_RAW'] +[f'{s}_PM' for s in stats_for_ratios]
    for stat in stats_to_roll:
        if stat not in logs_safe.columns: continue
        for w in ROLLING_WINDOWS:
            recent_logs = logs_safe.head(w)
            features[f'W_ROLL_AVG_{stat}_{w}'] = np.average(recent_logs[stat].fillna(0), weights=recent_logs['WEIGHT'].head(len(recent_logs))) if not recent_logs.empty else np.nan
        features[f'TREND_{stat}'] = features.get(f'W_ROLL_AVG_{stat}_{ROLLING_WINDOWS[0]}') - features.get(f'W_ROLL_AVG_{stat}_{ROLLING_WINDOWS[-1]}')

    for stat in['PTS', 'REB', 'AST', 'FG3M']:
        if stat in logs_safe.columns:
            last_20 = logs_safe.head(20)
            if not last_20.empty:
                std_dev = last_20[stat].std(); mean_val = last_20[stat].mean()
                cv = (std_dev / mean_val) if mean_val > 0.5 else 9.9
                features[f'CV_{stat}'] = cv
            else: features[f'CV_{stat}'] = 9.9
    features['DAYS_OF_REST'] = (datetime.utcnow().replace(tzinfo=pytz.utc) - pd.to_datetime(logs_for_calc['GAME_DATE'].iloc[0]).tz_localize('UTC')).days
    for stat in stats_for_ratios: features[f'{stat}_PM'] = logs_safe.head(20)[f'{stat}_PM'].mean() if f'{stat}_PM' in logs_safe.columns else 0
    return features

def apply_live_usage_redistribution(projection_rows, scratched_players, all_rosters):
    if not scratched_players or not projection_rows:
        return projection_rows

    proj_df = pd.DataFrame(projection_rows)
    scratched_rows = all_rosters[all_rosters['PLAYER_NAME'].apply(normalize_player_name).isin(scratched_players)]
    teams_with_injuries = scratched_rows['TEAM_ABBREVIATION'].unique()

    updated_rows = []
    for _, row in proj_df.iterrows():
        team = row['Team']
        if team in teams_with_injuries:
            team_scratches = scratched_rows[scratched_rows['TEAM_ABBREVIATION'] == team]
            num_injuries = len(team_scratches)

            usage_boost = 1.0 + (num_injuries * 0.035)
            min_boost = 1.0 + (num_injuries * 0.02)

            row['Pred_MIN'] = min(42.0, row['Pred_MIN'] * min_boost)
            row['Pred_PTS'] *= usage_boost
            row['Pred_AST'] *= usage_boost
            row['Pred_FGA_PM'] *= usage_boost
            row['Pred_PRA'] = row['Pred_PTS'] + row['Pred_REB'] + row['Pred_AST']
            row['Pred_P+R'] = row['Pred_PTS'] + row['Pred_REB']
            row['Pred_P+A'] = row['Pred_PTS'] + row['Pred_AST']
            row['Is_Star'] = (row['Pred_PTS'] >= 20) or (row['Pred_MIN'] >= 34)

        updated_rows.append(row)
    return updated_rows

def filter_sgp_correlations(sgp_df):
    if sgp_df.empty:
        return sgp_df

    filtered_rows = []
    games = sgp_df['Game'].unique()

    for game in games:
        game_bets = sgp_df[sgp_df['Game'] == game].copy()
        overs = game_bets[game_bets['Bet'] == 'Over']
        unders = game_bets[game_bets['Bet'] == 'Under']

        if not overs.empty:
            filtered_rows.append(overs.sort_values('Score', ascending=False).head(4))
        if not unders.empty:
            filtered_rows.append(unders.sort_values('Score', ascending=False).head(2))

    if filtered_rows:
        return pd.concat(filtered_rows, ignore_index=True)
    return sgp_df

def calculate_ou_probability(row):
    try:
        projection = float(row.get('Projection', 0))
        line = float(row.get('Line', 0))
        prop = row.get('Prop')

        if prop in['DD', 'TD']:
            prob_over = projection / 100.0
            prob_under = 1.0 - prob_over
            return prob_over, prob_under

        if pd.isna(projection) or pd.isna(line) or projection <= 0: return 0.5, 0.5
        prob_over = 1 - poisson.cdf(k=line, mu=projection)
        prob_under = poisson.cdf(k=line - 1, mu=projection)
        return prob_over, prob_under
    except: return 0.5, 0.5

def calculate_no_vig_prob(row):
    try:
        over_odds = float(row.get('Over_Odds', 0))
        under_odds = float(row.get('Under_Odds', 0))
        if over_odds <= 0 or under_odds <= 0 or pd.isna(over_odds) or pd.isna(under_odds):
            return ""

        over_implied = 1 / over_odds
        under_implied = 1 / under_odds
        total_implied = over_implied + under_implied

        if row.get('Bet') == 'Over':
            fair_prob = (over_implied / total_implied) * 100
        elif row.get('Bet') == 'Under':
            fair_prob = (under_implied / total_implied) * 100
        else:
            return ""

        return f"{fair_prob:.1f}%"
    except:
        return ""

def derive_shot_quality_signal(features):
    try:
        recent_efg = features.get('W_ROLL_AVG_EFG_PCT_RAW_5', 0)
        baseline_efg = features.get('W_ROLL_AVG_EFG_PCT_RAW_20', 0)
        fga = features.get('W_ROLL_AVG_FGA_5', 0)
        fg3a = features.get('W_ROLL_AVG_FG3A_5', 0)

        if baseline_efg <= 0: return "-"

        diff = recent_efg - baseline_efg
        if recent_efg > 0.62 and diff > 0.10:
            return "⚠️ Luck Risk"
        elif diff < -0.08 and fga >= 10:
            return "🔥 Due"
        elif fg3a >= 5 and recent_efg >= 0.54:
            return "🎯 High SQ"
        return "-"
    except:
        return "-"

def derive_advanced_matchup(row, def_ranks):
    try:
        pos = str(row.get('Pos', ''))
        prop = str(row.get('Prop', ''))
        bet = str(row.get('Bet', ''))
        oid = str(row.get('OpponentID', ''))

        rank = def_ranks.get(oid, {}).get(SIMPLE_PROP_MAP.get(prop, 'PTS'), 15)

        is_handler = 'G' in pos or 'PG' in pos
        is_big = 'C' in pos or 'F' in pos

        if bet == 'Over':
            if is_handler and prop in ['PTS', 'AST', 'PRA', 'P+A'] and rank >= 22:
                return "🔥 Elite P&R"
            if is_big and prop in ['PTS', 'REB', 'PRA', 'P+R'] and rank >= 22:
                return "🔥 Elite Rim"
            if prop in ['3PT', 'FG3M'] and rank >= 22:
                return "🔥 Elite Perim"
            if rank <= 5:
                return "⚠️ Trap Matchup"
        elif bet == 'Under':
            if is_handler and prop in ['PTS', 'AST', 'PRA'] and rank <= 5:
                return "🔒 Elite Under (P&R)"
            if is_big and prop in ['PTS', 'REB', 'PRA'] and rank <= 5:
                return "🔒 Elite Under (Rim)"
            if prop in ['3PT', 'FG3M'] and rank <= 5:
                return "🔒 Elite Under (Perim)"
        return "-"
    except Exception:
        return "-"

def get_previous_run_data(current_date_str, output_dir):
    try:
        dt_obj = datetime.strptime(current_date_str, '%m-%d-%Y')
        short_date_pattern = f"{dt_obj.month}-{dt_obj.day}-{str(dt_obj.year)[-2:]}"
    except:
        short_date_pattern = None

    candidate_files = []
    if os.path.exists(output_dir):
        for f in os.listdir(output_dir):
            if f.endswith("NBA Pro Dashboard.xlsx"):
                if current_date_str in f or (short_date_pattern and f.startswith(short_date_pattern)):
                    candidate_files.append(os.path.join(output_dir, f))

    if candidate_files:
        candidate_files.sort(key=os.path.getmtime, reverse=True)
        for file_path in candidate_files:
            print(f"✅ Found earlier run from TODAY: {os.path.basename(file_path)}. Calculating Pro vs Info Window movement...")
            try:
                return pd.read_excel(file_path, sheet_name='DraftKings_EV_Bets')
            except Exception:
                continue

    print(f"ℹ No earlier run found for TODAY ({current_date_str}). This is likely the 'Pro Window' (First Run).")
    return pd.DataFrame()

def apply_movement_and_ranking(curr_df, prev_df, sharp_props_map):
    if curr_df.empty: return curr_df
    if not prev_df.empty:
        curr_df['Player'] = curr_df['Player'].astype(str); prev_df['Player'] = prev_df['Player'].astype(str)
        if 'Bookmaker' in prev_df.columns:
            curr_df['Bookmaker'] = curr_df['Bookmaker'].astype(str); prev_df['Bookmaker'] = prev_df['Bookmaker'].astype(str)
            prev_clean = prev_df[['Player', 'Prop', 'Bookmaker', 'Edge (%)']].rename(columns={'Edge (%)': 'Prev_Edge'})
            merged = pd.merge(curr_df, prev_clean, on=['Player', 'Prop', 'Bookmaker'], how='left', suffixes=('', '_prev'))
        else:
            prev_clean = prev_df[['Player', 'Prop', 'Edge (%)']].rename(columns={'Edge (%)': 'Prev_Edge'})
            merged = pd.merge(curr_df, prev_clean, on=['Player', 'Prop'], how='left', suffixes=('', '_prev'))
    else: merged = curr_df.copy(); merged['Prev_Edge'] = np.nan

    def get_sharp_data(row):
        key = (row['match_key'], row['Prop'])
        sharp_lines_list = sharp_props_map.get(key,[])
        if sharp_lines_list:
            avg_sharp_line = sum(sharp_lines_list) / len(sharp_lines_list)
            diff = row['Projection'] - avg_sharp_line
            return round(avg_sharp_line, 1), round(diff, 1)
        return np.nan, np.nan
    merged[['Sharp Line', 'Proj vs Sharp']] = merged.apply(lambda r: pd.Series(get_sharp_data(r)), axis=1)

    def get_signal(row):
        curr_edge, prev_edge = row.get('Edge (%)', 0), row.get('Prev_Edge')
        if pd.isna(prev_edge): return "-"
        delta = curr_edge - prev_edge
        if delta > 3.5: return "PRICE IMPROVED"
        if delta < -3.5 and curr_edge > 5: return "SHARPS AGREE"
        return "-"
    merged['Sharp Signal'] = merged.apply(get_signal, axis=1)

    def get_consistency_badge(row):
        cv_pts = row.get('CV_PTS', 9.9); cv_reb = row.get('CV_REB', 9.9); cv_ast = row.get('CV_AST', 9.9)
        prop = row['Prop']
        cv = 9.9
        if prop == 'PTS': cv = cv_pts
        elif prop == 'REB': cv = cv_reb
        elif prop == 'AST': cv = cv_ast
        elif prop in['FG3M', '3PT']: cv = row.get('CV_FG3M', 9.9)
        elif prop == 'PRA': cv = (cv_pts + cv_reb + cv_ast) / 3
        elif prop == 'P+R': cv = (cv_pts + cv_reb) / 2
        elif prop == 'P+A': cv = (cv_pts + cv_ast) / 2
        elif prop == 'R+A': cv = (cv_reb + cv_ast) / 2
        else: return "-"
        if cv > 5.0: return "-"
        if cv < 0.25: return "🔒 High"
        elif cv < 0.45: return "✅ Med"
        else: return "⚡ Volatile"
    merged['Consistency'] = merged.apply(get_consistency_badge, axis=1)

    def get_ranking(row):
        edge, signal = row.get('Edge (%)', 0), row['Sharp Signal']
        bet_type = row.get('Bet')
        sl, line = row.get('Sharp Line'), row.get('Line')
        if not pd.isna(sl):
            if (bet_type == 'Over' and sl > line) or (bet_type == 'Under' and sl < line):
                return "💎 SHARP ALIGNMENT"
        if edge > 15.0: return "Strong Value"
        if edge > 5.0: return "Solid Play"
        return "-"
    merged['Ranking'] = merged.apply(get_ranking, axis=1)
    return merged

# --- Part 5: Main Execution Workflow ---
def run_daily_projections():
    proj_df = pd.DataFrame()

    try:
        if not os.path.exists('/content/drive'): drive.mount('/content/drive', force_remount=True)
    except Exception as e: print(f"⚠️ Drive mount check skipped: {e}")

    audit_hurdles = run_accuracy_audit()

    eastern = pytz.timezone('US/Eastern')
    SIMULATION_DATE = None

    if 'SIMULATION_DATE' in globals() and globals()['SIMULATION_DATE'] is not None:
         try: datetime.strptime(globals()['SIMULATION_DATE'], "%Y-%m-%d"); SIMULATION_DATE = globals()['SIMULATION_DATE']
         except: SIMULATION_DATE = None

    if SIMULATION_DATE:
        live_now_et = datetime.strptime(f"{SIMULATION_DATE} 12:00:00", "%Y-%m-%d %H:%M:%S").astimezone(eastern)
        print(f"\n⚠️ RUNNING IN SIMULATION MODE for date: {SIMULATION_DATE}")
        iso_today_str = SIMULATION_DATE
        file_today_str = live_now_et.strftime('%m-%d-%Y')
    else:
        live_now_et = datetime.now(pytz.utc).astimezone(eastern)
        print(f"\n📅 Running Live Mode. Current ET Time: {live_now_et.strftime('%Y-%m-%d %H:%M:%S')}")
        iso_today_str = live_now_et.strftime('%Y-%m-%d')
        file_today_str = live_now_et.strftime('%m-%d-%Y')

    run_mode_input = ''
    while run_mode_input not in['i', 'f', 't']:
        run_mode_input = input("Enter run mode ('i' for initial capture, 'f' for final report, 't' for training): ").strip().lower()

    if run_mode_input == 'i':
        run_mode = 'initial'
        print("\n🔵 INITIAL RUN DETECTED. Will save opening lines to CSV.")
    elif run_mode_input == 'f':
        run_mode = 'final'
        print("\n🔴 FINAL RUN DETECTED. Will compare vs Initial Lines for Steam/RLM/CLV.")
    elif run_mode_input == 't':
        run_mode = 'training'
        print("\n🟢 TRAINING MODE DETECTED.")

    if live_now_et.month >= 9: start_year = live_now_et.year; end_year = live_now_et.year + 1
    else: start_year = live_now_et.year - 1; end_year = live_now_et.year
    season = f"{start_year}-{str(end_year)[-2:]}"; prev_season = f"{start_year-1}-{str(start_year)[-2:]}"; season_year_for_url = str(start_year)

    print(f"Running projections for {iso_today_str} (Season: {season})")

    try:
        models = {'minutes': joblib.load(os.path.join(MODEL_DIR, "minutes_lgbm_model.pkl")), 'pts': joblib.load(os.path.join(MODEL_DIR, "pts_lgbm_model.pkl")), 'reb': joblib.load(os.path.join(MODEL_DIR, "reb_lgbm_model.pkl")), 'ast': joblib.load(os.path.join(MODEL_DIR, "ast_lgbm_model.pkl")), 'fg3m': joblib.load(os.path.join(MODEL_DIR, "fg3m_lgbm_model.pkl")), 'dd': joblib.load(os.path.join(MODEL_DIR, "dd_lgbm_model.pkl")), 'td': joblib.load(os.path.join(MODEL_DIR, "td_lgbm_model.pkl"))}
        imputers = {name: joblib.load(os.path.join(MODEL_DIR, f"{name}_imputer.pkl")) for name in['minutes', 'stats']}
        min_feature_names = joblib.load(os.path.join(MODEL_DIR, "minutes_feature_names.pkl")); stat_feature_names = joblib.load(os.path.join(MODEL_DIR, "stats_feature_names.pkl")); mae_values = json.load(open(os.path.join(MODEL_DIR, "mae_values.json")))
        print("✅ All models, imputers, and MAE values loaded successfully.")
    except FileNotFoundError as e: print(f"CRITICAL ERROR: A required model file is missing. Please run Script 2. Details: {e}"); return

    injury_df = fetch_espn_nba_injuries()
    master_df = load_master_boxscores()

    position_map = {}
    POSITIONS_FILE = os.path.join(GDRIVE_BASE_PATH, "Player Positions/2025-26 NBA Player Positions.xlsx")
    if os.path.exists(POSITIONS_FILE):
        try:
            pos_df = pd.read_excel(POSITIONS_FILE)
            if 'Player' in pos_df.columns and 'Pos' in pos_df.columns:
                pos_df['match_key'] = pos_df['Player'].apply(normalize_player_name)
                position_map = pos_df.set_index('match_key')['Pos'].to_dict()
        except Exception: pass

    scratched_players =[]
    if not injury_df.empty:
        scratched_players = injury_df[injury_df['Status'].isin(['Out', 'Injured Reserve', 'Suspended'])]['match_key'].tolist()

    if scratched_players: print(f"   ℹ Detected {len(scratched_players)} ruled-out players. Adjusting team usage automatically...")

    print("\n--- Preparing Context & Stats ---")

    if not master_df.empty:
        recent_cutoff = master_df['GAME_DATE'].max() - pd.Timedelta(days=300)
        recent_roster_df = master_df[master_df['GAME_DATE'] >= recent_cutoff]
        all_rosters = recent_roster_df.sort_values('GAME_DATE').groupby('PLAYER_ID', as_index=False).last()
    else:
        all_rosters = pd.DataFrame()

    def_ranks, def_stats_raw = get_defense_ranks_from_boxscores(master_df)
    team_metrics = get_team_metrics_from_boxscores(master_df)
    schedule = get_live_schedule_from_sportsdataverse(iso_today_str)

    if schedule:
        schedule = {str(k): str(v) for k, v in schedule.items() if k is not None and v is not None}
    if not all_rosters.empty and 'TEAM_ID' in all_rosters.columns:
        all_rosters['TEAM_ID'] = all_rosters['TEAM_ID'].astype(str).str.split('.').str[0]

    coach_path = os.path.join(DATA_DIR, "Coaches 2015 to Present.csv")
    coach_history_df = pd.read_csv(coach_path) if os.path.exists(coach_path) else pd.DataFrame()

    trait_path = os.path.join(DATA_DIR, "player_trait_map.csv")
    player_trait_map = pd.read_csv(trait_path) if os.path.exists(trait_path) else pd.DataFrame()

    props_df, live_game_lines = pd.DataFrame(), {}
    key, game_events = None,[]
    team_abbr_map = get_team_abbr_map()

    print("\n--- Running Initial Projections & Monte Carlo Sims ---")
    projection_rows, model_game_rows = [],[]

    LEAGUE_AVG_PACE = 99.0
    LEAGUE_AVG_OFF = 115.0

    if schedule and not all_rosters.empty:
        abbr_to_id = {t['abbreviation']: str(t['id']) for t in teams.get_teams()}
        id_to_abbr = {str(t['id']): t['abbreviation'] for t in teams.get_teams()}

        for team_id, opponent_id in tqdm(schedule.items(), desc="Generating Projections"):
            t_abbr = id_to_abbr.get(str(team_id), str(team_id))
            roster = all_rosters[(all_rosters['TEAM_ABBREVIATION'] == t_abbr) | (all_rosters['TEAM_ID'].astype(str) == str(team_id))]

            for _, row in roster.iterrows():
                player_id, player_name, team_abbr = row['PLAYER_ID'], row['PLAYER_NAME'], row['TEAM_ABBREVIATION']
                match_key = normalize_player_name(player_name)
                logs = get_player_gamelogs_from_master(player_id, master_df)
                features = calculate_live_features(logs)
                if not features: continue

                try: opp_id_int = int(opponent_id); opp_metrics = team_metrics.get(opp_id_int, {})
                except: opp_metrics = {}

                features['OPP_ALLOWED_PTS_AVG_30'] = opp_metrics.get('E_DEF_RATING', 115.0)
                features['OPP_ALLOWED_REB_AVG_30'] = 44.0; features['OPP_ALLOWED_AST_AVG_30'] = 25.0; features['OPP_ALLOWED_FG3M_AVG_30'] = 12.0
                features['OPP_ALLOWED_EFG_PCT_AVG_30'] = opp_metrics.get('E_EFG_PCT', 0.54)

                if str(opponent_id) in def_stats_raw:
                    features['OPP_ALLOWED_PTS_AVG_30'] = def_stats_raw[str(opponent_id)]['PTS']
                    features['OPP_ALLOWED_REB_AVG_30'] = def_stats_raw[str(opponent_id)]['REB']
                    features['OPP_ALLOWED_AST_AVG_30'] = def_stats_raw[str(opponent_id)]['AST']
                    features['OPP_ALLOWED_FG3M_AVG_30'] = def_stats_raw[str(opponent_id)]['FG3M']

                features['VEGAS_SPREAD'] = 0; features['VEGAS_TOTAL'] = 225

                is_playoff_game = 1 if live_now_et.month in [4, 5, 6] else 0
                features['IS_PLAYOFF_GAME'] = is_playoff_game

                coach_games = get_coach_games_played(team_abbr, coach_history_df, logs, start_year)
                features['COACH_GAMES_PLAYED'] = coach_games

                assigned_trait = assign_live_player_trait_runtime(player_id, player_name, logs, player_trait_map)

                pos_weight_pts = {0: 0.15, 1: 0.25, 2: 0.18, 3: 0.20, 4: 0.15, 5: 0.12, 6: 0.10, 7: 0.22, 8: 0.16, 9: 0.08}.get(assigned_trait, 0.18)
                pos_weight_reb = {0: 0.28, 1: 0.10, 2: 0.18, 3: 0.08, 4: 0.12, 5: 0.35, 6: 0.06, 7: 0.15, 8: 0.10, 9: 0.08}.get(assigned_trait, 0.15)
                pos_weight_ast = {0: 0.08, 1: 0.20, 2: 0.14, 3: 0.38, 4: 0.08, 5: 0.06, 6: 0.12, 7: 0.15, 8: 0.25, 9: 0.06}.get(assigned_trait, 0.15)
                pos_weight_3pm = {0: 0.02, 1: 0.22, 2: 0.18, 3: 0.25, 4: 0.32, 5: 0.01, 6: 0.15, 7: 0.28, 8: 0.20, 9: 0.10}.get(assigned_trait, 0.18)

                opp_pts_avg = features.get('OPP_ALLOWED_PTS_AVG_30', 115.0)
                opp_reb_avg = features.get('OPP_ALLOWED_REB_AVG_30', 44.0)
                opp_ast_avg = features.get('OPP_ALLOWED_AST_AVG_30', 25.0)
                opp_3pm_avg = features.get('OPP_ALLOWED_FG3M_AVG_30', 12.0)

                features['OPP_DvT_PTS'] = opp_pts_avg * pos_weight_pts * 5
                features['OPP_DvT_REB'] = opp_reb_avg * pos_weight_reb * 5
                features['OPP_DvT_AST'] = opp_ast_avg * pos_weight_ast * 5
                features['OPP_DvT_FG3M'] = opp_3pm_avg * pos_weight_3pm * 5

                try:
                    min_input = pd.DataFrame([features]).reindex(columns=min_feature_names, fill_value=0)
                    pred_min = models['minutes'].predict(imputers['minutes'].transform(min_input))[0]; features['Proj_MIN'] = pred_min

                    stat_input_df = pd.DataFrame([features]).reindex(columns=stat_feature_names, fill_value=0)
                    stat_input = imputers['stats'].transform(stat_input_df)

                    p_pts = models['pts'].predict(stat_input)[0]; p_reb = models['reb'].predict(stat_input)[0]
                    p_ast = models['ast'].predict(stat_input)[0]; p_3pm = models['fg3m'].predict(stat_input)[0]

                    temp_proj = {'Pred_PTS': p_pts, 'Pred_REB': p_reb, 'Pred_AST': p_ast}
                    if match_key in position_map: player_pos = position_map[match_key]
                    else: player_pos = derive_position(temp_proj)

                    opp_team_data = teams.find_team_name_by_id(int(opponent_id)) if str(opponent_id).isdigit() else None
                    opp_abbr = opp_team_data['abbreviation'] if opp_team_data else str(opponent_id)

                    exp_level = get_player_experience(player_id, master_df)
                    sq_sig = derive_shot_quality_signal(features)

                    proj = {
                        'PLAYER_ID': player_id, 'match_key': match_key, 'Player': player_name,
                        'Pos': player_pos, 'Experience': exp_level,
                        'Team': team_abbr, 'Opponent': opp_abbr,
                        'OpponentID': str(opponent_id), 'Pred_MIN': pred_min, 'Pred_FGA_PM': features.get('FGA_PM', 0),
                        'Pred_PTS': p_pts, 'Pred_REB': p_reb, 'Pred_AST': p_ast, 'Pred_FG3M': p_3pm, 'EFG_PCT_PREV': features.get('EFG_PCT_PREV', 0), 'TEAM_ID': team_id,
                        'Shot Quality': sq_sig, 'TREND_PTS': features.get('TREND_PTS', 0)
                    }

                    proj['Is_Star'] = (p_pts >= 22) or (pred_min >= 34)
                    proj['CV_PTS'] = features.get('CV_PTS', 9.9); proj['CV_REB'] = features.get('CV_REB', 9.9)
                    proj['CV_AST'] = features.get('CV_AST', 9.9); proj['CV_FG3M'] = features.get('CV_FG3M', 9.9)

                    clf_features = pd.DataFrame([{'Proj_PTS': p_pts, 'Proj_REB': p_reb, 'Proj_AST': p_ast}])
                    proj['Pred_DD'] = models['dd'].predict_proba(clf_features)[0][1] * 100; proj['Pred_TD'] = models['td'].predict_proba(clf_features)[0][1] * 100
                    proj['Pred_PRA'] = proj['Pred_PTS'] + proj['Pred_REB'] + proj['Pred_AST']; proj['Pred_P+R'] = proj['Pred_PTS'] + proj['Pred_REB']; proj['Pred_P+A'] = proj['Pred_PTS'] + proj['Pred_AST']; proj['Pred_R+A'] = proj['Pred_REB'] + proj['Pred_AST']
                    projection_rows.append(proj)
                except Exception: continue

            home_stats = team_metrics.get(int(team_id), {}) if str(team_id).isdigit() else {}
            away_stats = team_metrics.get(int(opponent_id), {}) if str(opponent_id).isdigit() else {}
            if home_stats and away_stats:
                h_off, a_off = home_stats.get('E_OFF_RATING', 110), away_stats.get('E_OFF_RATING', 110)
                h_def, a_def = home_stats.get('E_DEF_RATING', 110), away_stats.get('E_DEF_RATING', 110)
                h_pace, a_pace = home_stats.get('E_PACE', 99), away_stats.get('E_PACE', 99)
                poss = (h_pace + a_pace) / 2; HCA = 2.5

                high_intensity_venues = ['New York Knicks', 'Boston Celtics', 'Philadelphia 76ers', 'Denver Nuggets']
                home_team_info = teams.find_team_name_by_id(int(team_id)) if str(team_id).isdigit() else None
                home_team_name = home_team_info['full_name'] if home_team_info else "Home"
                if home_team_name in high_intensity_venues:
                    HCA = 4.5

                h_exp = ((h_off * a_def) / LEAGUE_AVG_OFF) * (poss / 100.0) + (HCA/2)
                a_exp = ((a_off * h_def) / LEAGUE_AVG_OFF) * (poss / 100.0) - (HCA/2)
                SIMS = 5000
                scores = np.random.multivariate_normal([h_exp, a_exp], [[169, 50],[50, 169]], size=SIMS)
                h_sc, a_sc = scores[:,0], scores[:,1]
                model_spread = -(np.mean(h_sc) - np.mean(a_sc))
                model_total = np.mean(h_sc) + np.mean(a_sc)
                model_win_prob = np.mean(h_sc > a_sc)
                model_game_rows.append({'TeamID': team_id, 'Home_Team_Name': home_team_name, 'Model Spread': round(model_spread, 1), 'Model Total': round(model_total, 1), 'Model Win %': model_win_prob, 'Feng Spread': "-"})

    if scratched_players and projection_rows:
        print("   ⚡ Applying live usage redistribution for scratched players...")
        projection_rows = apply_live_usage_redistribution(projection_rows, scratched_players, all_rosters)

    proj_df = pd.DataFrame(projection_rows)
    if proj_df.empty: print(f"\n❌ NO PROJECTIONS GENERATED."); return
    else: proj_df['match_key'] = proj_df['Player'].apply(normalize_player_name); print(f"✅ Generated {len(proj_df)} player projections.")

    # --- Market Odds Fetching ---
    print("\n--- Fetching Market Odds Data ---")
    game_lines_rows =[]; sharp_props_map = {}; initial_lines_to_save =[]
    all_props_raw =[]

    key = None
    for k in ODDS_API_KEYS:
        try:
            r = requests.get(f"https://api.the-odds-api.com/v4/sports/basketball_nba/odds?apiKey={k}&regions=us&markets=h2h", timeout=10)
            if r.status_code == 200:
                remaining = int(r.headers.get('x-requests-remaining', 0))
                if remaining > 10:
                    key = k; break
        except: continue

    if key:
        try:
            url_games = f"https://api.the-odds-api.com/v4/sports/basketball_nba/odds/?apiKey={key}&regions=us,eu&markets=h2h,spreads,totals&bookmakers={','.join(ALL_BOOKS_FOR_API)}"
            games_data = requests.get(url_games, timeout=15).json()

            if isinstance(games_data, list):
                for game in games_data:
                    if parser.isoparse(game["commence_time"]).astimezone(eastern).strftime('%Y-%m-%d') != iso_today_str:
                        continue

                    home_abbr = team_abbr_map.get(game['home_team'], game['home_team'])
                    away_abbr = team_abbr_map.get(game['away_team'], game['away_team'])

                    model_data = next((item for item in model_game_rows if item["Home_Team_Name"] == game['home_team']), {})

                    spreads, totals = [], []
                    retail_spread, retail_total, retail_ml = np.nan, np.nan, np.nan

                    for book in game.get('bookmakers',[]):
                        for market in book.get('markets', []):
                            if market['key'] == 'spreads':
                                for o in market['outcomes']:
                                    if o['name'] == game['home_team']:
                                        if book['key'] in SHARP_BOOKS: spreads.append(o['point'])
                                        if book['key'] in RETAIL_BOOKS: retail_spread = o['point']
                            elif market['key'] == 'totals':
                                for o in market['outcomes']:
                                    if o['name'] == 'Over':
                                        if book['key'] in SHARP_BOOKS: totals.append(o['point'])
                                        if book['key'] in RETAIL_BOOKS: retail_total = o['point']
                            elif market['key'] == 'h2h':
                                for o in market['outcomes']:
                                    if o['name'] == game['home_team'] and book['key'] in RETAIL_BOOKS: retail_ml = o['price']

                    sharp_spread = sum(spreads)/len(spreads) if spreads else None
                    sharp_total = sum(totals)/len(totals) if totals else None

                    mvmt, clv_s, clv_t = "-", "-", "-"
                    if run_mode == 'final' and os.path.exists(INITIAL_LINES_PATH):
                        try:
                            init_df = pd.read_csv(INITIAL_LINES_PATH); matched = init_df[init_df['GameID'] == game['id']]
                            if not matched.empty:
                                init_val = matched.iloc[0]['Initial_Sharp_Spread']
                                if sharp_spread is not None:
                                    clv_s = round(init_val - sharp_spread, 1)
                                    if abs(clv_s) >= 1.5: mvmt = "🔥 Steam Chase"
                                else:
                                    clv_s = 0.0
                        except: pass

                    game_lines_rows.append({
                        'Game': f"{away_abbr} @ {home_abbr}", 'Time': parser.isoparse(game["commence_time"]).astimezone(eastern).strftime('%-I:%M %p'),
                        'Retail Spread': retail_spread, 'Sharp Spread': round(sharp_spread, 1) if sharp_spread else "-",
                        'Model Spread': model_data.get('Model Spread', '-'), 'Retail Total': retail_total,
                        'Sharp Total': round(sharp_total, 1) if sharp_total else "-", 'Model Total': model_data.get('Model Total', '-'),
                        'Retail ML': retail_ml, 'Model Win %': f"{model_data.get('Model Win %', 0)*100:.1f}%" if 'Model Win %' in model_data else "-",
                        'Home_Team_Full': game['home_team'], 'Sort_Time': game['commence_time'], 'Game Move': mvmt, 'Game CLV': clv_s
                    })

                    if run_mode == 'initial':
                        initial_lines_to_save.append({
                            'GameID': game['id'], 'Home': home_abbr, 'Away': away_abbr,
                            'Initial_Sharp_Spread': sharp_spread, 'Initial_Sharp_Total': sharp_total
                        })

                    url_props = f"https://api.the-odds-api.com/v4/sports/basketball_nba/events/{game['id']}/odds?apiKey={key}&regions=us,eu&markets={','.join(ALL_PROP_MARKETS_LIST)}&bookmakers={','.join(ALL_BOOKS_FOR_API)}"
                    try:
                        time.sleep(0.5)
                        payload = requests.get(url_props, timeout=20).json()
                        if isinstance(payload, dict) and 'bookmakers' in payload:
                            for book in payload.get('bookmakers',[]):
                                for market in book.get('markets',[]):
                                    prop_name = MARKET_NAME_MAP.get(market['key'])
                                    if not prop_name: continue
                                    for outcome in market.get('outcomes',[]):
                                        p_name = outcome.get('description')
                                        if p_name:
                                            norm_name = normalize_player_name(p_name)
                                            if book['key'] in SHARP_BOOKS and outcome.get('point'):
                                                sharp_props_map.setdefault((norm_name, prop_name),[]).append(outcome.get('point'))
                                            if book['key'] in RETAIL_BOOKS:
                                                all_props_raw.append({
                                                    'Game': f"{away_abbr} @ {home_abbr}", 'Game_Time': parser.isoparse(game["commence_time"]).astimezone(eastern).strftime('%-I:%M %p'),
                                                    'Player': p_name, 'Prop': prop_name, 'Line': outcome.get('point'), 'Decimal_Odds': outcome.get('price'),
                                                    'OutcomeName': outcome['name'], 'Bookmaker': book['title'], 'Home_Team_Full': game['home_team'], 'Opponent_Abbr': away_abbr if game['home_team'] == game['home_team'] else home_abbr, 'Sort_Time': game['commence_time'],
                                                    'Game Move': mvmt, 'Game CLV': clv_s
                                                })
                    except: continue
            else:
                print(f"⚠️ API Error or No Games: {games_data}")
        except: traceback.print_exc()

    if run_mode == 'initial':
        try: pd.DataFrame(initial_lines_to_save).to_csv(INITIAL_LINES_PATH, index=False)
        except: pass

    if not all_props_raw:
        print("No props found. This could be due to API issues or no games on the schedule.")
        return

    props_df = pd.DataFrame(all_props_raw)
    props_df['match_key'] = props_df['Player'].apply(normalize_player_name)
    ou_pivot = props_df[props_df['OutcomeName'].isin(['Over', 'Under'])].pivot_table(index=['match_key', 'Player', 'Prop', 'Line', 'Bookmaker', 'Game', 'Game_Time', 'Sort_Time', 'Home_Team_Full', 'Opponent_Abbr', 'Game Move', 'Game CLV'], columns='OutcomeName', values='Decimal_Odds', observed=False).reset_index().rename(columns={'Over': 'Over_Odds', 'Under': 'Under_Odds'})
    merged = pd.merge(ou_pivot, proj_df.drop(columns=['Player'], errors='ignore'), on='match_key')

    merged['Status'] = 'Active'

    def get_display_odds(row):
        bet_type = row.get('Bet', 'Pass')
        dec = row.get('Over_Odds', 0) if bet_type == 'Over' else row.get('Under_Odds', 0)
        if pd.isna(dec) or dec == 0: return ""
        return f"+{int((dec - 1) * 100)}" if dec >= 2.00 else f"{int(-100 / (dec - 1))}"

    def get_proj(row):
        prop = row['Prop']
        if prop == 'PTS': return row.get('Pred_PTS', 0)
        elif prop == 'REB': return row.get('Pred_REB', 0)
        elif prop == 'AST': return row.get('Pred_AST', 0)
        elif prop == '3PT': return row.get('Pred_FG3M', 0)
        elif prop == 'PRA': return row.get('Pred_PRA', 0)
        elif prop == 'P+R': return row.get('Pred_P+R', 0)
        elif prop == 'P+A': return row.get('Pred_P+A', 0)
        elif prop == 'R+A': return row.get('Pred_R+A', 0)
        elif prop == 'DD': return row.get('Pred_DD', 0)
        elif prop == 'TD': return row.get('Pred_TD', 0)
        return 0

    merged['Projection'] = merged.apply(get_proj, axis=1).round(1)
    merged['Proj Range'] = merged.apply(lambda r: f"{max(0, r['Projection'] - mae_values.get(SIMPLE_PROP_MAP.get(r['Prop'], 'PTS'), 1)):.1f}-{r['Projection'] + mae_values.get(SIMPLE_PROP_MAP.get(r['Prop'], 'PTS'), 1):.1f}", axis=1)

    def get_opp_rank(row):
        try: oid = str(int(float(row['OpponentID'])))
        except: return "-"
        rank = def_ranks.get(oid, {}).get(SIMPLE_PROP_MAP.get(row['Prop'], 'PTS'), 99)
        if rank == 99: return "-"
        if rank <= 5: return f"{rank} (Hard)"
        if rank >= 25: return f"{rank} (Easy)"
        return str(rank)

    merged['Opp Rank'] = merged.apply(get_opp_rank, axis=1)
    merged[['Prob_Over', 'Prob_Under']] = merged.apply(lambda r: calculate_ou_probability(r), axis=1, result_type='expand')
    merged['Bet'] = np.where(merged['Prob_Over'] > 0.55, 'Over', np.where(merged['Prob_Under'] > 0.55, 'Under', 'Pass'))
    merged['Fair Prob %'] = merged.apply(calculate_no_vig_prob, axis=1)
    merged['Edge (%)'] = ((np.where(merged['Bet'] == 'Over', merged['Prob_Over']*merged['Over_Odds'], merged['Prob_Under']*merged['Under_Odds']) - 1) * 100).astype(float).round(1)
    merged['Odds'] = merged.apply(get_display_odds, axis=1)

    def get_streak_and_l10(row):
        so, su, ho, hu = calculate_prop_streak(row['PLAYER_ID'], SIMPLE_PROP_MAP.get(row['Prop'], 'PTS'), row['Line'], master_df)
        if row['Bet'] == 'Over':
            return f"W{so}", f"{int(ho)}/10"
        else:
            return f"W{su}", f"{int(hu)}/10"

    streak_l10_df = merged.apply(get_streak_and_l10, axis=1, result_type='expand')
    merged['Streak'] = streak_l10_df[0]
    merged['L10'] = streak_l10_df[1]

    merged['vs Opp'] = merged.apply(lambda r: calculate_head_to_head(r['PLAYER_ID'], r['Prop'], r['Line'], r['Opponent_Abbr'], start_year, master_df), axis=1)

    prev_run_df = get_previous_run_data(file_today_str, OUTPUT_DIR)
    merged = apply_movement_and_ranking(merged, prev_run_df, sharp_props_map)

    merged['Market Disc.'] = (merged['Sharp Line'] - merged['Line']).round(1)
    merged['Model Alpha'] = (merged['Projection'] - merged['Line']).round(1)

    merged['Adv Matchup'] = merged.apply(lambda r: derive_advanced_matchup(r, def_ranks), axis=1)

    def calculate_base_score(row):
        score = 50; bet = row['Bet']; sl = row['Sharp Line']; line = row['Line']
        is_star = row.get('Is_Star', False); trend = row.get('TREND_PTS', 0)
        opp_rank = str(row.get('Opp Rank', ''))
        h2h_str = row.get('vs Opp', '0/0')

        if '/' in h2h_str:
            try:
                hits, total = map(int, h2h_str.split('/'))
                if total >= 2 and hits == total: score += 25
            except: pass

        if trend > 5.0 and not is_star:
            score -= 25
        elif trend > 5.0 and is_star:
            score += 10

        if bet == 'Over' and "(Hard)" in opp_rank:
            if is_star: score -= 10
            else: score -= 40

        if not pd.isna(sl):
            market_diff = abs(sl - line)
            if (bet == 'Over' and sl < line) or (bet == 'Under' and sl > line):
                score -= 40
            elif (bet == 'Over' and sl > line) or (bet == 'Under' and sl < line):
                score += 20
            if market_diff >= 1.5:
                score += 30
        return score

    merged['Score'] = merged.apply(calculate_base_score, axis=1)

    def classify_bet(row):
        sl = row.get('Sharp Line'); line = row.get('Line')
        is_star = row.get('Is_Star', False); trend = row.get('TREND_PTS', 0)
        alpha = row.get('Model Alpha', 0)
        h2h_str = row.get('vs Opp', '0/0')

        if '/' in h2h_str:
            try:
                hits, total = map(int, h2h_str.split('/'))
                if total >= 3 and hits == total: return "🔥 H2H CRUSHER"
            except: pass

        high_intensity_venues = ['New York Knicks', 'Boston Celtics', 'Philadelphia 76ers', 'Denver Nuggets']
        if row['Home_Team_Full'] in high_intensity_venues and row['Team'] == get_team_abbr_map().get(row['Home_Team_Full']):
            if row['Score'] >= 50: return "🏟️ VENUE BOOST"

        if is_star and trend > 2.0 and alpha > 1.5: return "🔥 STANDOUT STAR"
        if not pd.isna(sl) and not pd.isna(line) and abs(sl - line) >= 1.5: return "🚨 SHARP ALERT"

        if row['Score'] >= 70: return "💎 ALPHA (Lock)"
        if row['Score'] >= 50: return "📈 BETA (Parlay)"
        if row['Score'] >= 30: return "🤖 GAMMA (Spec)"
        return "PASS"

    merged['Bet Class'] = merged.apply(classify_bet, axis=1)

    def build_fd_my_way(df):
        my_way_rows =[]
        for _, row in df.iterrows():
            prop = row.get('Prop'); proj = row.get('Projection', 0); bet = row.get('Bet', '')
            sl = row.get('Sharp Line'); line = row.get('Line')
            h2h_str = row.get('vs Opp', '0/0')
            edge = row.get('Edge (%)', 0)
            p_over = row.get('Prob_Over', 0.5); p_under = row.get('Prob_Under', 0.5)

            if not pd.isna(sl):
                if bet == 'Over' and sl < (line - 0.5): continue
                if bet == 'Under' and sl > (line + 0.5): continue

            mae = mae_values.get(SIMPLE_PROP_MAP.get(prop, 'PTS'), 1)
            raw_alt = proj - (0.75 * mae) if bet == 'Over' else proj + (0.75 * mae)

            # Snap to genuine sports betting half-point or whole number step
            alt_line = max(0.5, round(raw_alt * 2.0) / 2.0)

            # Recalibrated dynamic confidence between 72% and 94%
            target_prob = p_over if bet == 'Over' else p_under
            dyn_conf = min(94.5, max(72.0, (target_prob * 100.0) + (edge * 0.15) + (2.5 if row.get('Is_Star') else 0)))

            evidence = ["Statistically Safe"]
            if not pd.isna(sl) and ((bet == 'Over' and sl > line) or (bet == 'Under' and sl < line)):
                evidence.append("💎 Sharp Support")
                dyn_conf = min(94.5, dyn_conf + 3.0)
            if row.get('Is_Star'):
                evidence.append("⭐ Star Power")
            if row['Bet Class'] == "🏟️ VENUE BOOST":
                evidence.append("🏟️ Home Intensity")
            if '/' in h2h_str:
                try:
                    hits, total = map(int, h2h_str.split('/'))
                    if total >= 2 and hits == total:
                        evidence.append(f"🎯 H2H {h2h_str}")
                        dyn_conf = min(94.5, dyn_conf + 2.5)
                except: pass

            indicator = "🟢 OVER" if bet == 'Over' else "🔴 UNDER"

            my_way_rows.append({
                'Sort_Time': row.get('Sort_Time'),
                'Game_Time': row.get('Game_Time'),
                'Game': row.get('Game'),
                'Player': row.get('Player'),
                'Prop': prop,
                'Bet': bet,
                'Odds': row.get('Odds'),
                'Fair Prob %': row.get('Fair Prob %'),
                'Standard Line': line,
                'Alt Line': alt_line,
                'Direction': indicator,
                'Projection': proj,
                'Confidence': f"{dyn_conf:.1f}%",
                'Supporting Evidence': ", ".join(evidence)
            })
        return pd.DataFrame(my_way_rows)

    if run_mode == 'final':
        ts_str = live_now_et.strftime('%-m-%-d-%y %-I:%M%p').lower()
        out_file = os.path.join(OUTPUT_DIR, f"{ts_str} NBA Pro Dashboard.xlsx")
    else:
        out_file = os.path.join(OUTPUT_DIR, f"{file_today_str} NBA Pro Dashboard.xlsx")

    with pd.ExcelWriter(out_file, engine='xlsxwriter') as writer:
        workbook = writer.book
        header_fmt = workbook.add_format({'bold': True, 'align': 'center', 'valign': 'vcenter', 'bg_color': '#D9D9D9', 'border': 1})
        center_fmt = workbook.add_format({'align': 'center', 'valign': 'vcenter', 'border': 1})
        green_fmt = workbook.add_format({'bg_color': '#C6EFCE', 'font_color': '#006100', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        red_fmt = workbook.add_format({'bg_color': '#FFC7CE', 'font_color': '#9C0006', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        orange_fmt = workbook.add_format({'bg_color': '#FFEB9C', 'font_color': '#9C5700', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        blue_fmt = workbook.add_format({'bg_color': '#BDD7EE', 'font_color': '#1F4E78', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        purple_fmt = workbook.add_format({'bg_color': '#CCC0DA', 'font_color': '#3F3F76', 'align': 'center', 'valign': 'vcenter', 'border': 1})
        perfect_fmt = workbook.add_format({'bg_color': '#00FF00', 'font_color': '#000000', 'align': 'center', 'valign': 'vcenter', 'border': 1, 'bold': True})
        yellow_fmt = workbook.add_format({'bg_color': '#FFFF00', 'font_color': '#000000', 'align': 'center', 'valign': 'vcenter', 'border': 1})

        green_row_fmt = workbook.add_format({'bg_color': '#C6EFCE', 'font_color': '#006100', 'align': 'left', 'valign': 'vcenter', 'border': 1})
        blue_row_fmt = workbook.add_format({'bg_color': '#BDD7EE', 'font_color': '#1F4E78', 'align': 'left', 'valign': 'vcenter', 'border': 1})
        red_row_fmt = workbook.add_format({'bg_color': '#FFC7CE', 'font_color': '#9C0006', 'align': 'left', 'valign': 'vcenter', 'border': 1})
        yellow_row_fmt = workbook.add_format({'bg_color': '#FFFF00', 'font_color': '#000000', 'align': 'left', 'valign': 'vcenter', 'border': 1})
        purple_row_fmt = workbook.add_format({'bg_color': '#CCC0DA', 'font_color': '#3F3F76', 'align': 'left', 'valign': 'vcenter', 'border': 1})

        # --- ENHANCED HOW TO USE SHEET ---
        ws_howto = workbook.add_worksheet("HOW TO USE")
        ws_howto.set_column(0, 0, 32.0); ws_howto.set_column(1, 1, 110.0); ws_howto.freeze_panes(1, 0)
        row_idx = 0
        ws_howto.write(row_idx, 0, "Metric / Column / Tier", header_fmt); ws_howto.write(row_idx, 1, "Professional Sports Betting Interpretation & Operational Guide", header_fmt); row_idx += 1

        ws_howto.write(row_idx, 0, "💎 ALPHA (Lock)", perfect_fmt); ws_howto.write(row_idx, 1, "Model, Sharps, and Price all converge with positive expectation. Top priority plays (Score >= 70).", perfect_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "🔥 STANDOUT STAR", perfect_fmt); ws_howto.write(row_idx, 1, "Elite usage player on an upward trend with high model alpha. Virtually matchup-proof.", perfect_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "🔥 H2H CRUSHER", perfect_fmt); ws_howto.write(row_idx, 1, "Player holds an undefeated historical record against this specific opponent over the last 2 seasons.", perfect_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "🏟️ VENUE BOOST", blue_fmt); ws_howto.write(row_idx, 1, "Home court intensity boost (+2.0 spread / +50 base score) for elite home arenas (MSG, TD Garden, Wells Fargo).", blue_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "📈 BETA (Parlay Leg)", green_fmt); ws_howto.write(row_idx, 1, "Confirmed sharp market support. Excellent secondary building block for standard parlays (Score 50-70).", green_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "🤖 GAMMA (Speculative)", yellow_fmt); ws_howto.write(row_idx, 1, "High mathematical model edge captured before sharp books move retail prices (Score 30-50).", yellow_fmt); row_idx += 1
        row_idx += 1

        ws_howto.write(row_idx, 0, "Fair Prob % (No-Vig)", header_fmt); ws_howto.write(row_idx, 1, "The true market-implied win probability after mathematically stripping the sportsbook's vig (juice).", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Market Disc. (Discrepancy)", header_fmt); ws_howto.write(row_idx, 1, "Sharp Consensus Line minus Retail Line. Positive indicates retail book is trailing sharp markets.", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Model Alpha", header_fmt); ws_howto.write(row_idx, 1, "Model Projection minus Retail Line. Raw statistical projection edge over the retail pricing.", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Proj vs Sharp", header_fmt); ws_howto.write(row_idx, 1, "Model Projection minus Sharp Consensus. Highlighted green when at least 0.5 units in your bet's favor.", header_fmt); row_idx += 1
        row_idx += 1

        ws_howto.write(row_idx, 0, "Consistency (Vol Check)", header_fmt); ws_howto.write(row_idx, 1, "Measures coefficient of variation. High (Locked) / Med (Solid) / Volatile (Caution).", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Adv Matchup", header_fmt); ws_howto.write(row_idx, 1, "Exposes scheme strengths: Elite P&R, Elite Rim, Elite Perim vs Trap Matchups.", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Shot Quality", header_fmt); ws_howto.write(row_idx, 1, "Regression indicator: High SQ (open looks), Due (positive regression), Luck Risk (over-performing).", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Game Move & Game CLV", header_fmt); ws_howto.write(row_idx, 1, "Steam Chase indicates rapid sharp market money. Closing Line Value (CLV) confirms beating the closing line.", header_fmt); row_idx += 1
        row_idx += 1

        ws_howto.write(row_idx, 0, "FD My Way Sheet Guide", header_fmt); ws_howto.write(row_idx, 1, "Curated alternate milestone lines rounded to clean half/whole numbers with dynamically calculated confidence.", header_fmt); row_idx += 1
        ws_howto.write(row_idx, 0, "Game Lines Sheet Guide", header_fmt); ws_howto.write(row_idx, 1, "Monte Carlo multivariate spread and total simulations benchmarked against sharp market lines.", header_fmt); row_idx += 1

        # --- OUT COLUMNS SEQUENCE ---
        out_cols = [
            'Game_Time', 'Game', 'Status', 'Experience', 'Team', 'Player', 'Prop',
            'Line', 'Bet', 'Odds', 'Fair Prob %', 'Projection', 'Sharp Line', 'Market Disc.', 'Proj vs Sharp',
            'Model Alpha', 'Edge (%)', 'Consistency', 'Opp Rank', 'Adv Matchup', 'Streak', 'L10', 'vs Opp',
            'Shot Quality', 'Game Move', 'Game CLV', 'Sharp Signal', 'Ranking', 'Score', 'Bet Class'
        ]

        def auto_fit_columns(worksheet, dataframe, extra_padding=5):
            for col_idx, col_name in enumerate(dataframe.columns):
                max_len = len(str(col_name))
                if not dataframe.empty:
                    series_max = dataframe[col_name].astype(str).map(len).max()
                    if pd.notna(series_max):
                        max_len = max(max_len, int(series_max))
                worksheet.set_column(col_idx, col_idx, max(10, max_len + extra_padding), center_fmt)

        sheet_config =['DraftKings_EV_Bets', 'DraftKings_EV_SGP', 'FanDuel_EV_Bets', 'FanDuel_EV_SGP']
        for sheet_name in sheet_config:
            bookmaker = 'DraftKings' if 'DraftKings' in sheet_name else 'FanDuel'
            is_sgp = 'SGP' in sheet_name

            if run_mode == 'final':
                book_df = merged[(merged['Bookmaker'] == bookmaker) & (merged['Bet'] != 'Pass') & (merged['Score'] >= 30) & (merged['Bet Class'] != 'PASS')].copy()
            else:
                book_df = merged[(merged['Bookmaker'] == bookmaker) & (merged['Bet'] != 'Pass')].copy()

            if not book_df.empty:
                if 'Fair Prob %' not in book_df.columns:
                    book_df['Fair Prob %'] = ""

                # Group players with multiple props together, ordered by Score
                if is_sgp:
                    final_df = book_df.sort_values(['Sort_Time', 'Game', 'Player', 'Score'], ascending=[True, True, True, False]).reindex(columns=out_cols)
                    final_df = filter_sgp_correlations(final_df)
                else:
                    final_df = book_df.sort_values(['Player', 'Score'], ascending=[True, False]).reindex(columns=out_cols)

                final_df.to_excel(writer, sheet_name=sheet_name, index=False, startrow=1, header=False)
                ws = writer.sheets[sheet_name]
                ws.freeze_panes(1, 6)
                ws.autofilter(0, 0, len(final_df), len(out_cols) - 1)

                for col_idx, val in enumerate(final_df.columns):
                    ws.write(0, col_idx, val, header_fmt)

                auto_fit_columns(ws, final_df, extra_padding=5)

                last_row = len(final_df) + 1
                def get_col(name): return xl_col_to_name(out_cols.index(name))

                ws.set_column(out_cols.index('Score'), out_cols.index('Score'), None, None, {'hidden': True})

                line_col = get_col('Line'); bet_col = get_col('Bet'); sharp_col = get_col('Sharp Line')
                proj_col = get_col('Projection'); opp_rank_col = get_col('Opp Rank'); player_col = get_col('Player')
                score_col = get_col('Score'); odds_col = get_col('Odds'); adv_col = get_col('Adv Matchup')
                class_col = get_col('Bet Class'); v_opp_col = get_col('vs Opp')
                proj_vs_sharp_col = get_col('Proj vs Sharp'); consistency_col = get_col('Consistency')

                # Bet Class Highlights
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'STAR', 'format': perfect_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'BOOST', 'format': blue_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'ALPHA', 'format': perfect_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'BETA', 'format': green_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'GAMMA', 'format': yellow_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'ALERT', 'format': orange_fmt})
                ws.conditional_format(f'{class_col}2:{class_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'CRUSHER', 'format': perfect_fmt})

                # Player Score Highlights
                ws.conditional_format(f'{player_col}2:{player_col}{last_row}', {'type': 'formula', 'criteria': f'=${score_col}2>=75', 'format': perfect_fmt})
                ws.conditional_format(f'{player_col}2:{player_col}{last_row}', {'type': 'formula', 'criteria': f'=AND(${score_col}2>=55, ${score_col}2<75)', 'format': green_fmt})
                ws.conditional_format(f'{player_col}2:{player_col}{last_row}', {'type': 'formula', 'criteria': f'=AND(${score_col}2>=30, ${score_col}2<55)', 'format': yellow_fmt})

                # Opponent Rank Matchup Highlights
                green_matchup_criteria = f'=OR(AND(${bet_col}2="Over", ISNUMBER(SEARCH("Easy", ${opp_rank_col}2))), AND(${bet_col}2="Under", ISNUMBER(SEARCH("Hard", ${opp_rank_col}2))))'
                ws.conditional_format(f'{opp_rank_col}2:{opp_rank_col}{last_row}', {'type': 'formula', 'criteria': green_matchup_criteria, 'format': green_fmt})
                ws.conditional_format(f'{bet_col}2:{bet_col}{last_row}', {'type': 'formula', 'criteria': green_matchup_criteria, 'format': green_fmt})

                # Proj vs Sharp In Favor of Bet Type (>= 0.5 Over, <= -0.5 Under)
                proj_sharp_criteria = f'=OR(AND(${bet_col}2="Over", ${proj_vs_sharp_col}2>=0.5), AND(${bet_col}2="Under", ${proj_vs_sharp_col}2<=-0.5))'
                ws.conditional_format(f'{proj_vs_sharp_col}2:{proj_vs_sharp_col}{last_row}', {'type': 'formula', 'criteria': proj_sharp_criteria, 'format': green_fmt})

                # Consistency Med or Better Green
                ws.conditional_format(f'{consistency_col}2:{consistency_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'High', 'format': green_fmt})
                ws.conditional_format(f'{consistency_col}2:{consistency_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'Med', 'format': green_fmt})

                # Adv Matchup
                ws.conditional_format(f'{adv_col}2:{adv_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': '🔥', 'format': green_fmt})
                ws.conditional_format(f'{adv_col}2:{adv_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': '🔒', 'format': green_fmt})
                ws.conditional_format(f'{adv_col}2:{adv_col}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': '⚠️', 'format': red_fmt})

                # L10 & vs Opp
                for trend_val in ['7/10', '8/10', '9/10', '10/10']:
                    ws.conditional_format(f'{get_col("L10")}2:{get_col("L10")}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': trend_val, 'format': green_fmt})
                ws.conditional_format(f'{v_opp_col}2:{v_opp_col}{last_row}', {'type': 'formula', 'criteria': f'=AND(IFERROR(FIND("/", ${v_opp_col}2)>0, FALSE), IFERROR(VALUE(LEFT(${v_opp_col}2, FIND("/", ${v_opp_col}2)-1)) = VALUE(MID(${v_opp_col}2, FIND("/", ${v_opp_col}2)+1, 10)), FALSE), IFERROR(VALUE(MID(${v_opp_col}2, FIND("/", ${v_opp_col}2)+1, 10)) > 0, FALSE))', 'format': perfect_fmt})

                # Game Move & CLV
                ws.conditional_format(f'{get_col("Game Move")}2:{get_col("Game Move")}{last_row}', {'type': 'text', 'criteria': 'containing', 'value': 'Steam', 'format': yellow_fmt})
                ws.conditional_format(f'{get_col("Game CLV")}2:{get_col("Game CLV")}{last_row}', {'type': 'cell', 'criteria': '>', 'value': 0, 'format': green_fmt})

            else: writer.book.add_worksheet(sheet_name).write(0, 0, "No bets found.")

        # --- FD MY WAY SHEET ---
        fd_df = merged[(merged['Bookmaker'] == 'FanDuel') & (merged['Bet'] != 'Pass')].copy()
        if not fd_df.empty:
            my_way_df = build_fd_my_way(fd_df)
            if not my_way_df.empty:
                # Group by earliest game time, matchup, player name
                my_way_df = my_way_df.sort_values(by=['Sort_Time', 'Game', 'Player', 'Confidence'], ascending=[True, True, True, False])
                my_way_df.drop(columns=['Sort_Time'], inplace=True, errors='ignore')

                my_way_df.to_excel(writer, sheet_name='FD My Way', index=False, startrow=1, header=False)
                ws_mw = writer.sheets['FD My Way']
                ws_mw.freeze_panes(1, 4)
                ws_mw.autofilter(0, 0, len(my_way_df), len(my_way_df.columns) - 1)

                for col_idx, val in enumerate(my_way_df.columns):
                    ws_mw.write(0, col_idx, val, header_fmt)

                auto_fit_columns(ws_mw, my_way_df, extra_padding=6)

                lr_mw = len(my_way_df) + 1
                c_dir = xl_col_to_name(my_way_df.columns.get_loc('Direction'))
                c_alt = xl_col_to_name(my_way_df.columns.get_loc('Alt Line'))
                c_conf = xl_col_to_name(my_way_df.columns.get_loc('Confidence'))
                c_evid = xl_col_to_name(my_way_df.columns.get_loc('Supporting Evidence'))

                # Color-code Over (Soft Green) and Under (Soft Red) on Direction & Alt Line
                ws_mw.conditional_format(f'{c_dir}2:{c_dir}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'OVER', 'format': green_fmt})
                ws_mw.conditional_format(f'{c_dir}2:{c_dir}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'UNDER', 'format': red_fmt})
                ws_mw.conditional_format(f'{c_alt}2:{c_alt}{lr_mw}', {'type': 'cell', 'criteria': '>', 'value': 0, 'format': center_fmt})

                # Confidence Highlighting (>85% Perfect Green, >75% Light Green)
                ws_mw.conditional_format(f'{c_conf}2:{c_conf}{lr_mw}', {'type': 'cell', 'criteria': '>=', 'value': '"85.0%"', 'format': perfect_fmt})
                ws_mw.conditional_format(f'{c_conf}2:{c_conf}{lr_mw}', {'type': 'cell', 'criteria': '<', 'value': '"85.0%"', 'format': green_fmt})

                # Supporting Evidence Highlighting
                ws_mw.conditional_format(f'{c_evid}2:{c_evid}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'Sharp Support', 'format': purple_fmt})
                ws_mw.conditional_format(f'{c_evid}2:{c_evid}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'Star Power', 'format': blue_fmt})
                ws_mw.conditional_format(f'{c_evid}2:{c_evid}{lr_mw}', {'type': 'text', 'criteria': 'containing', 'value': 'H2H', 'format': perfect_fmt})

        # --- GAME LINES SHEET ---
        if game_lines_rows:
            gl_df = pd.DataFrame(game_lines_rows)
            gl_cols =['Time', 'Game', 'Retail Spread', 'Sharp Spread', 'Model Spread', 'Retail Total', 'Sharp Total', 'Model Total', 'Retail ML', 'Model Win %', 'Home_Team_Full', 'Game Move', 'Game CLV']
            gl_df = gl_df.reindex(columns=gl_cols)
            gl_df.to_excel(writer, sheet_name='Game_Lines', index=False)
            ws_gl = writer.sheets['Game_Lines']
            ws_gl.freeze_panes(1, 0)
            ws_gl.autofilter(0, 0, len(gl_df), len(gl_cols) - 1)

            for col_idx, val in enumerate(gl_df.columns):
                ws_gl.write(0, col_idx, val, header_fmt)

            auto_fit_columns(ws_gl, gl_df, extra_padding=5)

            lr_gl = len(gl_df) + 1
            col_m_spd = xl_col_to_name(gl_cols.index('Model Spread'))
            col_r_spd = xl_col_to_name(gl_cols.index('Retail Spread'))
            col_m_tot = xl_col_to_name(gl_cols.index('Model Total'))
            col_r_tot = xl_col_to_name(gl_cols.index('Retail Total'))
            col_win = xl_col_to_name(gl_cols.index('Model Win %'))
            col_move = xl_col_to_name(gl_cols.index('Game Move'))
            col_clv = xl_col_to_name(gl_cols.index('Game CLV'))

            # Highlights on significant model edges (>2.0 on spread, >3.0 on total)
            ws_gl.conditional_format(f'{col_m_spd}2:{col_m_spd}{lr_gl}', {'type': 'formula', 'criteria': f'=ABS(${col_m_spd}2-${col_r_spd}2)>=2.0', 'format': green_fmt})
            ws_gl.conditional_format(f'{col_m_tot}2:{col_m_tot}{lr_gl}', {'type': 'formula', 'criteria': f'=ABS(${col_m_tot}2-${col_r_tot}2)>=3.0', 'format': green_fmt})
            ws_gl.conditional_format(f'{col_win}2:{col_win}{lr_gl}', {'type': 'cell', 'criteria': '>=', 'value': '"55.0%"', 'format': perfect_fmt})
            ws_gl.conditional_format(f'{col_move}2:{col_move}{lr_gl}', {'type': 'text', 'criteria': 'containing', 'value': 'Steam', 'format': yellow_fmt})
            ws_gl.conditional_format(f'{col_clv}2:{col_clv}{lr_gl}', {'type': 'cell', 'criteria': '>', 'value': 0, 'format': green_fmt})
            ws_gl.conditional_format(f'{col_clv}2:{col_clv}{lr_gl}', {'type': 'cell', 'criteria': '<', 'value': 0, 'format': red_fmt})

    print(f"✅ Exported: {out_file}")

if __name__ == "__main__":
    run_daily_projections()
```

```python
#@title SCRIPT 6: HISTORICAL PROP INGESTION & CACHING ENGINE (v1.2 - PAID KEY ONLY)
# Purpose: Downloads historical closing player props from The Odds API,
#          caches them safely to Google Drive, and prevents duplicate token usage.
# ==============================================================================
import os
import time
import requests
import pandas as pd
from datetime import datetime, timedelta
from dateutil import parser
from google.colab import drive
from tqdm import tqdm

# --- 1. CONFIGURATION ---
PAID_API_KEY = "YOUR_PAID_API_KEY_HERE"

# File Paths
BASE_DIR = "/content/drive/MyDrive/NBA/"
DATA_DIR = os.path.join(BASE_DIR, "Historical Data/")
CACHE_PATH = os.path.join(DATA_DIR, "persistent_odds_cache.csv")

# Prop Categories we want to fetch
MARKETS = [
    'player_points', 'player_rebounds', 'player_assists', 'player_threes',
    'player_points_rebounds_assists', 'player_points_rebounds',
    'player_points_assists', 'player_rebounds_assists'
]

# Map for short names inside cache
MARKET_SHORT_MAP = {
    'player_points': 'PTS', 'player_rebounds': 'REB', 'player_assists': 'AST',
    'player_threes': 'FG3M', 'player_points_rebounds_assists': 'PRA',
    'player_points_rebounds': 'P+R', 'player_points_assists': 'P+A', 'player_rebounds_assists': 'R+A'
}

# ==============================================================================
# MAIN DOWNLOAD LOOP
# ==============================================================================
def run_historical_ingestion():
    drive.mount('/content/drive', force_remount=True)
    os.makedirs(DATA_DIR, exist_ok=True)

    # Load pre-existing cache to identify already downloaded dates
    if os.path.exists(CACHE_PATH):
        try:
            cache_df = pd.read_csv(CACHE_PATH)
            # Parse commence times to dates to identify completed dates
            cache_df['date_str'] = cache_df['Game_Time'].apply(lambda x: parser.isoparse(x).strftime('%Y-%m-%d') if pd.notna(x) else "")
            completed_dates = set(cache_df['date_str'].unique())
            print(f"📂 Found existing cache at {CACHE_PATH} with {len(cache_df)} entries.")
            print(f"📅 {len(completed_dates)} dates are already completely cached.")
        except Exception as e:
            print(f"⚠️ Error reading cache: {e}. Starting fresh.")
            cache_df = pd.DataFrame()
            completed_dates = set()
    else:
        cache_df = pd.DataFrame()
        completed_dates = set()

    # Define Ingestion Range (From start of 2023-24 season to today)
    start_date = datetime(2023, 10, 24)
    end_date = datetime.now() - timedelta(days=1)

    # Generate list of days to download
    delta = end_date - start_date
    days_to_process = []
    for i in range(delta.days + 1):
        day = start_date + timedelta(days=i)
        day_str = day.strftime('%Y-%m-%d')
        # Skip dates that have already been completely cached
        if day_str not in completed_dates:
            days_to_process.append(day)

    if not days_to_process:
        print("✅ Your historical cache is fully up to date! No new dates to ingest.")
        return

    print(f"🚀 Ingesting {len(days_to_process)} missing dates using Paid API Key. Let's begin.")

    new_records = []

    for target_day in tqdm(days_to_process, desc="Downloading Dates"):
        target_day_str = target_day.strftime('%Y-%m-%d')
        # We query the snapshot at 23:00 UTC (6 PM EST) for that day, which captures the final lines before game time
        snapshot_timestamp = f"{target_day_str}T23:00:00Z"

        success = False
        attempts = 0
        events_list = []

        while not success and attempts < 3:
            url_events = f"https://api.the-odds-api.com/v4/historical/sports/basketball_nba/events?apiKey={PAID_API_KEY}&date={snapshot_timestamp}"

            try:
                r = requests.get(url_events, timeout=15)
                if r.status_code == 200:
                    events_list = r.json().get('data', [])
                    success = True
                elif r.status_code == 429:
                    print("\n❌ Error: Your Paid API Key quota has been exhausted.")
                    save_accumulated_records(new_records)
                    return
                elif r.status_code in [401, 403]:
                    print("\n❌ Error: Your Paid API Key is invalid or unauthorized for historical snapshots.")
                    save_accumulated_records(new_records)
                    return
                else:
                    attempts += 1
                    time.sleep(1)
            except Exception:
                attempts += 1
                time.sleep(2)

        if not success or not events_list:
            continue # Skip day if we could not retrieve events list

        # Loop through each game on that date
        for event in events_list:
            event_id = event['id']
            home_team = event['home_team']
            away_team = event['away_team']
            commence_time = event['commence_time']

            # Fetch historical prop lines for this specific game
            success_odds = False
            odds_attempts = 0
            payload = {}

            while not success_odds and odds_attempts < 3:
                url_odds = f"https://api.the-odds-api.com/v4/historical/sports/basketball_nba/events/{event_id}/odds?apiKey={PAID_API_KEY}&regions=us&markets={','.join(MARKETS)}&date={snapshot_timestamp}&bookmakers=draftkings,fanduel"

                try:
                    r_odds = requests.get(url_odds, timeout=20)
                    if r_odds.status_code == 200:
                        payload = r_odds.json().get('data', {})
                        success_odds = True
                    elif r_odds.status_code == 429:
                        print("\n❌ Error: Your Paid API Key quota has been exhausted during odds retrieval.")
                        save_accumulated_records(new_records)
                        return
                    elif r_odds.status_code in [401, 403]:
                        print("\n❌ Error: Your Paid API Key is invalid or unauthorized during odds retrieval.")
                        save_accumulated_records(new_records)
                        return
                    else:
                        odds_attempts += 1
                        time.sleep(1)
                except Exception:
                    odds_attempts += 1
                    time.sleep(2)

            if not success_odds or not payload:
                continue

            # Parse the bookmakers and markets from response
            for book in payload.get('bookmakers', []):
                book_title = book['title']
                for market in book.get('markets', []):
                    mkey = market['key']
                    prop_short = MARKET_SHORT_MAP.get(mkey)
                    if not prop_short:
                        continue

                    for outcome in market.get('outcomes', []):
                        player_name = outcome.get('description') or outcome.get('name')
                        line = outcome.get('point')
                        price = outcome.get('price')
                        outcome_name = outcome.get('name')

                        # We only want the consensus 'Over' outcome to represent the standard line
                        if outcome_name == 'Over' and player_name and line:
                            new_records.append({
                                'Game': f"{away_team} @ {home_team}",
                                'Game_Time': commence_time,
                                'Player': player_name,
                                'Prop': prop_short,
                                'Line': float(line),
                                'Decimal_Odds': float(price),
                                'OutcomeName': outcome_name,
                                'Bookmaker': book_title,
                                'event_id': event_id
                            })

        # Save every 5 processed dates to prevent loss of progress during connection cuts
        if len(new_records) > 500:
            save_accumulated_records(new_records)
            new_records = []

    # Final save of remaining items
    save_accumulated_records(new_records)
    print("\n🎉 Ingestion sequence completed successfully! Cache is fully populated.")

def save_accumulated_records(records_list):
    """Safely appends and saves new entries to the Google Drive database."""
    if not records_list:
        return

    new_df = pd.DataFrame(records_list)
    new_df['cache_timestamp'] = datetime.utcnow().strftime('%Y-%m-%d %H:%M:%S')

    if os.path.exists(CACHE_PATH):
        try:
            cache_df = pd.read_csv(CACHE_PATH)
            # Combine and drop exact duplicates
            combined_df = pd.concat([cache_df, new_df], ignore_index=True).drop_duplicates(subset=['Game_Time', 'Player', 'Prop', 'Bookmaker'])
        except Exception:
            combined_df = new_df
    else:
        combined_df = new_df

    combined_df.to_csv(CACHE_PATH, index=False)
    print(f"\n💾 Database Synced: Appended {len(new_df)} new lines to your Google Drive cache.")

if __name__ == "__main__":
    run_historical_ingestion()
```
